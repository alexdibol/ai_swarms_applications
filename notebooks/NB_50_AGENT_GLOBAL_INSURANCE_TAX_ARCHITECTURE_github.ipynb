# NB — 50-Agent Swarm for Global Insurance Corporate & Tax Architecture

## A synthetic, compliance-first experiment in organizational design

This notebook asks a deliberately difficult question: **how might a very large, London-listed insurance group simplify an extraordinarily fragmented international structure into one group headquarters and three regional operating subsidiaries while improving tax efficiency, regulatory coherence, capital mobility, resilience, and governance?** The purpose is not to discover a tax “loophole.” It is to demonstrate how a swarm of specialized AI agents can reason over a problem in which tax is only one layer of a much larger institutional system.

We construct a synthetic insurer, **Atlas Global Insurance Group plc (AGIG)**. AGIG is assumed to serve clients in roughly 150 countries, maintain regulated insurance entities in at least 20 important jurisdictions, write property and casualty, specialty, catastrophe, marine, cyber, health, life and retirement business, and purchase and provide reinsurance through a dense internal and external network. Its shares are listed in London and its shareholder base is globally distributed. The legacy group grew through acquisitions, leaving behind duplicated legal entities, uneven capitalization, multiple service companies, trapped capital, intercompany balances, intellectual-property arrangements, local tax attributes, and a web of reinsurance treaties.

The redesign target is intentionally severe: a **listed parent / group headquarters plus three principal regional platforms — Europe, Americas and Asia-Pacific**. Local regulated carriers are not magically eliminated. Insurance law, licensing, policyholder protection, local capital requirements and supervisory expectations often require legal entities to remain in place. Therefore “three subsidiaries” means three principal regional holding and operating platforms beneath which necessary regulated local entities may continue to exist. The swarm must distinguish economic simplification from fictional legal simplification.

The notebook treats the company as a **multilayer graph**. Nodes represent legal entities and jurisdictions. Edges represent ownership, premiums, claims, reinsurance, dividends, service charges, financing and operational dependencies. Each jurisdiction receives synthetic attributes covering corporate tax, Pillar Two exposure, withholding friction, regulatory intensity, capital fungibility, insurance-market depth, treaty-network quality, operating cost, talent, political and legal stability, and the practical ability to demonstrate management substance. These values are deliberately illustrative. They are not a database of current tax law.

A second layer models business risk. Synthetic exposures are generated by geography and line of business, including catastrophe sensitivity, life-duration risk, cyber concentration, health volatility and specialty-tail risk. A third layer models reinsurance. The notebook creates ceded-risk relationships among operating entities and regional reinsurance centers, allowing the swarm to reason about diversification, counterparty concentration and the possibility that a tax-efficient structure may be economically inferior if it weakens capital mobility or creates regulatory fragmentation.

The analytical architecture uses **50 specialist agents**. They are not fifty copies of the same prompt. The swarm contains international-tax specialists, Pillar Two specialists, transfer-pricing experts, insurance-regulation experts, solvency and capital experts, reinsurance specialists, corporate lawyers, governance specialists, treasury specialists, accounting experts, operational-resilience specialists, data and technology specialists, geopolitical-risk analysts and institutional-design critics. Each agent receives a bounded mandate and the same structured evidence pack. Claude is asked to return machine-readable judgments rather than free-form advocacy.

The swarm is organized as a deliberative institution. First, deterministic code creates the synthetic company and calculates candidate architectures. Second, agents inspect the evidence from their disciplines. Third, a synthesis layer aggregates their objections, constraints and preferences. Fourth, a red-team stage searches for designs that appear tax-efficient only because they ignore substance, transfer pricing, minimum-tax rules, regulatory capital, policyholder protection, governance or implementation costs. Finally, a constrained scoring engine identifies a **Pareto-style shortlist**, not a magical single answer produced by an opaque language model.

Tax efficiency is therefore defined carefully. The objective is to reduce **lawful structural friction**: unnecessary withholding leakage, duplicated tax costs, avoidable entity complexity, stranded losses, inefficient capital routing, inconsistent transfer-pricing arrangements and administrative burden. The notebook explicitly penalizes structures whose apparent advantage depends on very low effective tax rates, weak substance, artificial profit allocation, regulatory arbitrage or assumptions inconsistent with the OECD/G20 Pillar Two environment. For a group of this scale, a 15% jurisdictional minimum-tax framework can materially reduce the value of locating profits in low-tax jurisdictions; the design problem increasingly becomes one of **substance, capital efficiency, treaty access, supervisory credibility and operating coherence**, rather than simply choosing the lowest headline rate.

This is especially important for insurance. A global insurer cannot be optimized as if it were a collection of invoices. Underwriting authority, claims management, actuarial judgment, investment management, reinsurance purchasing, risk ownership, board oversight, capital and policyholder obligations must align with the legal architecture. A regional hub that books risk but lacks people, decision rights, systems and capital is not an economically credible solution. The notebook therefore treats “substance” as an explicit constraint.

The intended output is a board-level architecture: the recommended location of the parent/headquarters and the three regional platforms; the logic for retaining local regulated entities; a proposed pattern for reinsurance and shared services; an estimate of structural tax and capital friction under synthetic assumptions; a map of critical legal and regulatory dependencies; dissenting agent views; and a staged implementation roadmap. The notebook does **not** produce legal or tax advice and should not be used to execute a real restructuring without jurisdiction-specific advice, current tax data, actuarial analysis, regulatory engagement and board approval.

Methodologically, the exercise is also about advanced autonomous systems. The 50-agent swarm demonstrates that difficult institutional problems are not solved merely by adding more model calls. The useful unit is a governed system: **data → graph → constraints → specialists → debate → red team → deterministic validation → recommendation → provenance**. The language model supplies judgment and hypothesis generation; deterministic code supplies arithmetic, reproducibility and constraint enforcement. The result is an auditable collaboration between statistical computation and machine reasoning.

## Pedagogical orientation

The central lesson is that corporate architecture is a **multi-objective constrained optimization problem**. A naïve tax optimizer might minimize an estimated tax rate. A realistic insurer must simultaneously consider taxation, regulatory capital, solvency, legal enforceability, reinsurance economics, operational resilience, transfer pricing, governance, market access, talent, reporting, political risk and implementation cost. Some of these dimensions can be quantified; others require expert judgment. The swarm is designed to combine both.

Students should pay particular attention to the distinction between **objective functions and constraints**. A lower tax burden can be an objective, but legal compliance, minimum regulatory capital, policyholder protection, credible economic substance and arm’s-length pricing are constraints. A solution that violates a constraint is not “high return”; it is infeasible. This distinction is fundamental to responsible AI in finance.

The second lesson concerns graphs. A multinational insurer is not naturally represented by a flat spreadsheet. Ownership links, premium flows, claims, reinsurance, dividends and services create networks. Network representations reveal concentration, centrality, bottlenecks and dependencies that disappear in consolidated financial statements. By collapsing a legacy graph into a regional architecture, we can measure whether simplification genuinely reduces complexity or merely hides it.

The third lesson is institutional pluralism. Different specialists optimize different things. A tax lawyer may value treaty access; a chief risk officer may value capital fungibility; a supervisor may value local resolvability; a reinsurance expert may value diversification; a CFO may value distributable cash; an operations executive may value talent and infrastructure. The 50-agent design makes these tensions explicit rather than forcing one general-purpose model to impersonate consensus.

Finally, the notebook uses AI as a **decision-support layer**, not as the final legal authority. Claude can identify issues, challenge assumptions and synthesize competing perspectives, but the numerical scoring and hard feasibility gates remain inspectable. In a real project, every material assumption would be replaced with verified legal, tax, actuarial, accounting and regulatory data, and each proposed transaction would require professional sign-off.

### Code Unit 2 — Constructing the synthetic global insurer

This cell creates the synthetic institutional footprint. Twenty-four regulated jurisdictions are assigned to Europe, the Americas or Asia-Pacific, and the group is assumed to serve customers in approximately 150 countries through these carriers, branches, fronting arrangements and cross-border permissions. Each regulated entity receives synthetic premiums, claims, reserves, capital, employees and business-mix weights. The resulting table is not intended to mimic any named insurer; it is a pedagogical “mammoth” large enough to make restructuring genuinely difficult.

The design deliberately separates **regulated entities** from **client countries**. A company can have customers in far more jurisdictions than it has incorporated insurance carriers. This distinction matters because tax nexus, insurance licensing, permanent-establishment questions and policyholder obligations do not map one-for-one to customer geography.

Financial quantities are generated with correlated logic rather than independently. Premium scale influences reserves and capital; claims are derived from a loss ratio; employee counts scale with activity; and each entity is assigned a business line that affects its risk characteristics. This gives the synthetic data enough internal structure for later analysis.

The cell also assigns every entity a unique identifier and regional classification. In a real engagement, this table would be assembled from legal-entity management systems, general ledgers, regulatory returns, actuarial systems, tax provision data and reinsurance administration systems. The notebook intentionally creates it in one place so students can see exactly what the agents are allowed to know. A central principle of the experiment is **evidence symmetry**: every specialist receives the same factual base, while their mandates differ. That reduces the risk that apparent disagreement among agents is simply caused by inconsistent input data.

## Infographic — Code Unit 2




```python
# CODE UNIT 2 — Synthetic insurer: 24 regulated jurisdictions, ~150 client countries
regulated = [
("United Kingdom","Europe"),("Ireland","Europe"),("France","Europe"),("Germany","Europe"),
("Netherlands","Europe"),("Spain","Europe"),("Italy","Europe"),("Switzerland","Europe"),
("United States","Americas"),("Canada","Americas"),("Mexico","Americas"),("Brazil","Americas"),
("Chile","Americas"),("Colombia","Americas"),("Argentina","Americas"),("Panama","Americas"),
("Singapore","Asia-Pacific"),("Hong Kong","Asia-Pacific"),("Japan","Asia-Pacific"),("Australia","Asia-Pacific"),
("India","Asia-Pacific"),("South Korea","Asia-Pacific"),("Malaysia","Asia-Pacific"),("Indonesia","Asia-Pacific")
]
lines = ["Property","Casualty","Life","Health","Specialty","Marine","Cyber","Retirement"]
rows = []
for i,(country,region) in enumerate(regulated,1):
    premiums = np.random.lognormal(mean=7.0, sigma=.55)
    loss_ratio = np.clip(np.random.normal(.66,.09),.40,.90)
    rows.append({
        "entity_id":f"AGIG_{i:02d}","country":country,"region":region,
        "primary_line":random.choice(lines),
        "gross_premium_usdm":round(premiums,1),
        "claims_usdm":round(premiums*loss_ratio,1),
        "reserves_usdm":round(premiums*np.random.uniform(1.2,4.5),1),
        "reg_capital_usdm":round(premiums*np.random.uniform(.18,.42),1),
        "employees":int(max(80,premiums*np.random.uniform(.25,.65))),
        "cat_exposure":round(np.random.beta(2,3),3),
        "duration_risk":round(np.random.beta(2,3),3),
        "cyber_concentration":round(np.random.beta(2,4),3)
    })
entities = pd.DataFrame(rows)
client_countries = [f"ClientCountry_{i:03d}" for i in range(1,151)]

print("Synthetic AGIG regulated entities:", len(entities))
print("Synthetic client-country footprint:", len(client_countries))
print("Gross premium (USD bn):", round(entities.gross_premium_usdm.sum()/1000,2))
display(entities.head(10))
```

### Code Unit 3 — Jurisdiction attributes and compliance gates

The third unit creates the jurisdiction-level decision surface. Candidate locations are described by synthetic values for corporate-tax burden, withholding friction, treaty-network quality, insurance regulatory depth, capital fungibility, talent, infrastructure, legal stability, operating cost and the feasibility of maintaining genuine management substance. The tax rates are explicitly illustrative; they must not be interpreted as current statutory rates. The point is to build a reusable optimization grammar, not a tax database.

The cell introduces a crucial variable: `pillar2_floor`. For a group of the assumed scale, a design that derives its attraction from an effective rate materially below the global minimum-tax framework should be treated skeptically. The notebook therefore calculates a simplified “minimum-tax-adjusted” burden and adds a penalty where the synthetic tax advantage would otherwise depend on a sub-floor rate. This is not a GloBE calculator. Real Pillar Two computations are jurisdictional, accounting-intensive and full of detailed adjustments, exclusions, safe harbours and local implementation rules. The simplified variable exists to stop the optimization from learning the wrong lesson.

The compliance gate also requires sufficient legal stability, regulatory depth and substance capacity. These are hard feasibility screens rather than soft preferences. A location can be inexpensive and still be rejected if it cannot credibly host the functions attributed to it.

This separation between **raw economic attractiveness** and **legal feasibility** is one of the most important architectural choices in the notebook. It prevents a weighted-average score from compensating for an unacceptable compliance weakness. In institutional design, some dimensions are not tradeable. A board cannot rationally say that weak substance is acceptable because office costs are low. The code makes that principle computationally explicit.

## Infographic — Code Unit 3



```python
# CODE UNIT 3 — Synthetic jurisdiction decision attributes and hard compliance gates
countries = [x[0] for x in regulated]
region_map = dict(regulated)
rng = np.random.default_rng(SEED+3)

jur = pd.DataFrame({"country":countries})
jur["region"] = jur.country.map(region_map)
# ILLUSTRATIVE synthetic inputs — NOT current statutory tax data.
jur["synthetic_corp_tax"] = rng.uniform(.12,.32,len(jur)).round(3)
jur["withholding_friction"] = rng.uniform(.02,.14,len(jur)).round(3)
jur["treaty_network"] = rng.uniform(.55,.95,len(jur)).round(3)
jur["regulatory_depth"] = rng.uniform(.55,.98,len(jur)).round(3)
jur["capital_fungibility"] = rng.uniform(.45,.95,len(jur)).round(3)
jur["talent_infrastructure"] = rng.uniform(.55,.98,len(jur)).round(3)
jur["legal_stability"] = rng.uniform(.60,.99,len(jur)).round(3)
jur["substance_capacity"] = rng.uniform(.55,.98,len(jur)).round(3)
jur["operating_cost"] = rng.uniform(.35,.95,len(jur)).round(3)

# Give established international centers plausible synthetic institutional depth.
boost = {
"United Kingdom":(.96,.90,.96,.96),"Ireland":(.90,.87,.90,.91),
"Switzerland":(.95,.90,.93,.94),"Netherlands":(.91,.88,.92,.91),
"United States":(.96,.88,.97,.95),"Canada":(.91,.88,.92,.94),
"Singapore":(.96,.94,.97,.96),"Hong Kong":(.91,.88,.94,.91),
"Japan":(.94,.84,.94,.96),"Australia":(.92,.86,.93,.95)
}
for c,(regd,cap,talent,legal) in boost.items():
    jur.loc[jur.country==c,["regulatory_depth","capital_fungibility","talent_infrastructure","legal_stability"]] = [regd,cap,talent,legal]

PILLAR2_FLOOR = .15
jur["min_tax_adjusted"] = jur.synthetic_corp_tax.clip(lower=PILLAR2_FLOOR)
jur["subfloor_penalty"] = (PILLAR2_FLOOR-jur.synthetic_corp_tax).clip(lower=0)*3
jur["hard_feasible"] = (
    (jur.regulatory_depth>=.70) &
    (jur.legal_stability>=.72) &
    (jur.substance_capacity>=.68)
)
display(jur.sort_values(["hard_feasible","regulatory_depth"],ascending=False).head(12))
```

### Code Unit 4 — Building the multilayer corporate and reinsurance graph

This unit converts the synthetic insurer from a table into a network. Each regulated carrier becomes a node. The London-listed parent is represented as the root owner, and edges capture ownership relationships. Additional directed edges represent synthetic reinsurance cessions among entities. These reinsurance links are weighted by ceded premium, allowing the notebook to distinguish a trivial legal connection from a financially significant dependency.

The graph is intentionally simplified but conceptually powerful. Real insurance groups contain several overlapping networks: legal ownership, cash remittances, capital support, guarantees, reinsurance, service agreements, investment management, data processing and operational dependencies. A single legal organization chart captures only one of them. Network analysis allows us to ask different questions: Which entities are central to risk transfer? Where are the bottlenecks? How concentrated is reinsurance? Which nodes connect otherwise separate regional clusters? What would happen if a proposed hub became unavailable?

The cell calculates degree-based centrality and summarizes the most connected nodes. It also computes a simple complexity measure from the number of entities and edges. Later, the proposed regional architecture can be compared with this legacy baseline.

Students should notice that the graph does not “solve tax.” Instead, it changes the representation of the institution so that tax recommendations can be evaluated in the context of risk and operations. This is a recurring theme in advanced AI systems: better reasoning often begins with a better **state representation**. If the model sees only tax rates, it will optimize tax rates. If it sees a network of policyholder obligations, risk transfers, capital and management functions, it is more likely to reason about an insurer as an insurer rather than as an abstract collection of taxable profits.

## Infographic — Code Unit 4



```python
# CODE UNIT 4 — Multilayer ownership + reinsurance graph
G = nx.MultiDiGraph()
G.add_node("AGIG_PLC", kind="parent", country="United Kingdom", region="Europe")
for _,r in entities.iterrows():
    G.add_node(r.entity_id, kind="regulated_carrier", country=r.country, region=r.region)
    G.add_edge("AGIG_PLC", r.entity_id, layer="ownership", weight=1.0)

# Synthetic internal reinsurance network: regional + cross-regional diversification.
for _, cedent in entities.iterrows():
    possible = entities[entities.entity_id != cedent.entity_id]
    same_region = possible[possible.region == cedent.region]
    targets = same_region.sample(min(2,len(same_region)), random_state=SEED+int(cedent.entity_id[-2:]))
    if np.random.rand() > .35:
        targets = pd.concat([targets, possible.sample(1, random_state=SEED+100+int(cedent.entity_id[-2:]))]).drop_duplicates("entity_id")
    for _, reinsurer in targets.iterrows():
        ceded = float(cedent.gross_premium_usdm*np.random.uniform(.03,.18))
        G.add_edge(cedent.entity_id, reinsurer.entity_id, layer="reinsurance", weight=round(ceded,1))

simple = nx.DiGraph()
for u,v,d in G.edges(data=True):
    simple.add_edge(u,v,weight=simple.get_edge_data(u,v,{}).get("weight",0)+d["weight"])
centrality = nx.degree_centrality(simple)
top_nodes = sorted(centrality.items(), key=lambda x:x[1], reverse=True)[:8]
legacy_complexity = G.number_of_nodes() + G.number_of_edges()

print("Nodes:",G.number_of_nodes(),"Edges:",G.number_of_edges(),"Legacy complexity index:",legacy_complexity)
print("Most connected nodes:",top_nodes)

plt.figure(figsize=(12,8))
pos = nx.spring_layout(simple, seed=SEED, k=.7)
nx.draw_networkx(simple,pos,with_labels=False,node_size=120,arrows=False)
plt.title("Synthetic AGIG ownership and reinsurance network")
plt.axis("off")
plt.show()
```

### Code Unit 5 — Generating candidate headquarters and regional architectures

The fifth unit defines the restructuring search space. The listed parent is allowed to remain associated with London while the optimization evaluates a manageable set of plausible headquarters and regional-platform locations. Europe, the Americas and Asia-Pacific each have their own candidate hubs. The code then forms combinations of one headquarters plus one hub per region.

Every architecture receives deterministic scores for tax friction, regulatory quality, capital mobility, substance, operating capability, legal stability and implementation complexity. A synthetic withholding matrix introduces friction when cash or service flows cross borders. Importantly, the tax objective is not “lowest corporate rate.” It combines minimum-tax-adjusted burden with cross-border friction and a penalty for weak substance. This turns the problem from rate shopping into institutional architecture.

The candidate generator also enforces regional logic: a European platform must be located in the European candidate set, and so forth. In a real restructuring, additional constraints would include existing licenses, portfolio-transfer law, employee consultation, regulatory change-of-control approvals, debt covenants, tax attributes, deferred tax, stamp duties, exit taxes and the legal feasibility of moving contracts.

The output is a dataframe in which each row is a complete architecture. This is a critical interface between deterministic optimization and the swarm. The agents will not invent arbitrary jurisdictions from scratch; they will evaluate a bounded, auditable set of alternatives. This reduces hallucination and makes dissent interpretable. The approach is analogous to generating a feasible frontier before asking experts to judge it. Language models are strongest when they reason over explicit alternatives and constraints rather than when they are asked to manufacture an entire transaction structure in unconstrained prose.

## Infographic — Code Unit 5



```python
# CODE UNIT 5 — Candidate HQ + Europe + Americas + Asia-Pacific architectures
hq_candidates = ["United Kingdom","Ireland","Switzerland","Singapore"]
eu_candidates = ["United Kingdom","Ireland","Netherlands","Switzerland","Germany"]
am_candidates = ["United States","Canada","Mexico","Brazil"]
ap_candidates = ["Singapore","Hong Kong","Japan","Australia"]

J = jur.set_index("country")

def architecture_metrics(hq,eu,am,ap):
    locs=[hq,eu,am,ap]
    x=J.loc[locs]
    tax = x.min_tax_adjusted.mean() + x.subfloor_penalty.mean()
    wh = np.mean([J.loc[a,"withholding_friction"]*0.5 + J.loc[b,"withholding_friction"]*0.5
                  for a,b in [(hq,eu),(hq,am),(hq,ap)]])
    reg = x.regulatory_depth.mean()
    cap = x.capital_fungibility.mean()
    substance = x.substance_capacity.mean()
    talent = x.talent_infrastructure.mean()
    legal = x.legal_stability.mean()
    cost = x.operating_cost.mean()
    feasible = bool(x.hard_feasible.all())
    # Lower is better. Tax matters, but cannot dominate institutional quality.
    score = (
        2.2*tax + 1.2*wh + .55*cost
        -1.35*reg -1.20*cap -1.15*substance -0.85*talent -0.90*legal
    )
    return tax,wh,reg,cap,substance,talent,legal,cost,feasible,score

records=[]
for i,(hq,eu,am,ap) in enumerate(itertools.product(hq_candidates,eu_candidates,am_candidates,ap_candidates),1):
    vals=architecture_metrics(hq,eu,am,ap)
    records.append([f"ARCH_{i:03d}",hq,eu,am,ap,*vals])

cols=["arch_id","hq","europe","americas","asia_pacific","tax_friction","withholding_friction",
      "regulatory_depth","capital_fungibility","substance","talent","legal_stability",
      "operating_cost","hard_feasible","det_score"]
architectures=pd.DataFrame(records,columns=cols)
shortlist=architectures[architectures.hard_feasible].sort_values("det_score").head(12).copy()
display(shortlist)
```

### Code Unit 6 — Defining the 50-agent expert institution

This cell creates the swarm itself. Fifty agents are assigned explicit professional roles. The distribution is intentionally heterogeneous: international tax, Pillar Two, transfer pricing, insurance regulation, solvency, actuarial risk, reinsurance, corporate law, securities law, accounting, treasury, capital management, governance, operations, technology, cyber resilience, geopolitical risk, implementation and red-team disciplines are all represented. Several roles appear more than once with different emphases, reflecting the reality that important decisions deserve independent scrutiny.

Each agent receives a mandate that tells it what to protect and what evidence to prioritize. The agents are not asked to “find the lowest tax.” Instead, they are asked to identify feasibility constraints, material risks, trade-offs and preferred architectures from their disciplinary perspective. This makes the swarm closer to an investment committee, supervisory college or multidisciplinary transaction team than to a voting contest among identical bots.

The code also assigns each role a domain weight. These weights do not determine the answer by themselves; they help the synthesis layer understand which objections are particularly material. For example, a regulatory infeasibility flagged by a solvency specialist should not be neutralized by several generic operational preferences.

A major pedagogical point is that **agent count is not intelligence**. Fifty poorly differentiated agents can create fifty correlated errors. Diversity of mandates, structured evidence, adversarial review and deterministic validation are what make the swarm useful. The notebook therefore records agent identity, domain and mandate as first-class data. In production, one would also track prompt versions, model versions, timestamps, token usage, confidence calibration and human reviewer dispositions. Those provenance elements turn an impressive demonstration into a governable decision process.

### Code Unit 7 — Claude Sonnet 5 deliberation with structured outputs

This unit provides the live LLM interface. When `LIVE_LLM=False`, it produces deterministic proxy opinions so that the notebook remains executable without an API key or cost. When live mode is enabled, each of the 50 specialists receives the same compact evidence pack: the synthetic group profile, the leading deterministic architectures, the compliance principles and its own mandate. The Anthropic API is called with the configured Claude Sonnet 5 model.

The prompt requires a JSON object containing the agent’s preferred architecture identifier, confidence, principal rationale, hard constraints, principal risk, and any objection to the deterministic ranking. This is deliberately narrower than a traditional consulting prompt. Structured outputs allow the subsequent code to aggregate results, count objections and preserve dissent.

The system instruction is compliance-first. It prohibits recommendations that depend on concealment, sham substance, artificial profit shifting, false transfer pricing or evasion. It also reminds the agent that local insurance entities may need to remain because the target architecture refers to principal regional platforms rather than the literal disappearance of regulated carriers. These constraints are not cosmetic. They define the legitimate design space.

The deterministic fallback is also pedagogically useful. It demonstrates that the orchestration layer should not collapse when the model is unavailable. The proxy opinions use domain-specific perturbations of the candidate scores, creating disagreement without pretending that the fallback is expert legal advice.

Students can inspect `agent_opinions` to see whether the swarm converges or fragments. A fragmented result is informative: it indicates that the problem contains unresolved trade-offs or insufficient evidence. The notebook does not force unanimity. ### Code Unit 8 — Aggregation, dissent and red-team challenge

The eighth unit converts fifty individual judgments into an institutional view. It counts architecture preferences, weights them by agent confidence and domain materiality, and merges this signal with the deterministic candidate table. The result is not simple majority rule. An architecture can attract many votes yet still fail if it violates a hard feasibility gate or accumulates severe objections from regulation, solvency, legal or tax-compliance specialists.

The red-team layer searches for characteristic failure modes: excessive reliance on a low synthetic tax rate, weak management substance, poor capital fungibility, low regulatory depth, high withholding friction, excessive implementation complexity and concentration of critical functions. Each candidate receives a red-team penalty and an explicit list of flags. This is important because optimization systems are often better at finding loopholes in their objective function than at understanding institutional intent. Red teaming asks, in effect, “How could this score look good while the real-world structure is bad?”

The aggregation code preserves dissent. It reports both the leading consensus and the number of material flags. A board should be able to see that a recommendation was selected despite a particular objection and then decide whether additional evidence is required.

This cell illustrates a general pattern for autonomous finance: **proposal and criticism should be separate computational roles**. The same mechanism that generates an idea should not be the only mechanism that validates it. Independent deterministic tests, specialist objections and red-team prompts reduce common-mode failure.

In a real tax restructuring, the red-team stage would be far more extensive. It would include jurisdiction-specific anti-avoidance rules, controlled foreign company rules, transfer-pricing documentation, permanent-establishment analysis, beneficial ownership, treaty entitlement, exit taxation, insurance premium taxes, VAT/GST, deferred tax, accounting impacts, sanctions, regulatory approvals, policyholder communications and implementation sequencing. Here, those complexities are represented as categories rather than legal conclusions.

## Infographic — Code Unit 8



```python
# CODE UNIT 8 — Weighted consensus, dissent and deterministic red-team penalties
weights = agents.set_index("agent_id").domain_weight.to_dict()
agent_opinions["weight"] = agent_opinions.agent_id.map(weights)
agent_opinions["weighted_support"] = agent_opinions["weight"] * agent_opinions["confidence"].astype(float)

support = agent_opinions.groupby("preferred_arch_id").agg(
    votes=("agent_id","count"),
    weighted_support=("weighted_support","sum"),
    mean_confidence=("confidence","mean")
).reset_index()

evaluated = shortlist.merge(support,left_on="arch_id",right_on="preferred_arch_id",how="left")
evaluated[["votes","weighted_support","mean_confidence"]] = evaluated[["votes","weighted_support","mean_confidence"]].fillna(0)

def red_team(row):
    flags=[]
    if row["substance"] < .75: flags.append("substance")
    if row["capital_fungibility"] < .68: flags.append("capital_fungibility")
    if row["regulatory_depth"] < .78: flags.append("regulatory_depth")
    if row["withholding_friction"] > .10: flags.append("cross_border_friction")
    if row["tax_friction"] < .15: flags.append("minimum_tax_dependency")
    if row["legal_stability"] < .80: flags.append("legal_stability")
    return flags

evaluated["red_flags"]=evaluated.apply(red_team,axis=1)
evaluated["red_penalty"]=evaluated.red_flags.map(lambda x:.18*len(x))
maxsup=max(evaluated.weighted_support.max(),1)
evaluated["consensus_bonus"]=evaluated.weighted_support/maxsup
evaluated["final_score"]=evaluated.det_score + evaluated.red_penalty - .35*evaluated.consensus_bonus
evaluated=evaluated.sort_values("final_score")
display(evaluated[["arch_id","hq","europe","americas","asia_pacific","votes","weighted_support","red_flags","final_score"]])
```

### Code Unit 9 — Selecting the constrained architecture and designing the operating model

This unit selects the highest-ranked architecture that survives the hard compliance gates and red-team threshold. It then translates a row of scores into a proposed organizational pattern: a listed parent and group headquarters, three regional platforms, retained local regulated carriers where legally required, regional reinsurance coordination, shared-service boundaries and explicit substance requirements.

The key concept is **functional alignment**. If a regional platform is described as the risk owner, it must have the people, board authority, actuarial capability, underwriting governance, claims oversight, capital and systems necessary to own that risk. If a service center performs routine functions, its remuneration should be designed consistently with arm’s-length principles. If a reinsurance center assumes material insurance risk, its capital and decision-making must correspond to that risk. The notebook therefore emits an “architecture contract” describing what must be true for the recommendation to remain valid.

The cell also estimates the change in synthetic structural friction relative to the legacy network. This is not a forecast of tax savings. It is an index combining entity complexity, cross-border friction and administrative duplication. The distinction is intentional: for a modern multinational subject to minimum-tax regimes, the board may obtain more durable value from simplification, capital mobility and governance than from chasing headline tax-rate differentials.

The recommended design is consequently a **hypothesis for professional validation**, not an executable transaction plan. The output includes explicit conditions precedent: current local tax opinions, Pillar Two modeling, transfer-pricing functional analysis, actuarial and solvency work, regulator engagement, securities-law review and board approval. This makes the final answer useful without pretending that synthetic data can replace legal facts.

```python
# CODE UNIT 9 — Select architecture and translate it into an institutional operating model
eligible = evaluated[
    evaluated.hard_feasible &
    (evaluated.red_penalty <= .36)
].copy()
if eligible.empty:
    eligible = evaluated[evaluated.hard_feasible].copy()

winner = eligible.iloc[0]
architecture_contract = {
    "listed_parent_and_group_hq": winner.hq,
    "europe_platform": winner.europe,
    "americas_platform": winner.americas,
    "asia_pacific_platform": winner.asia_pacific,
    "local_carriers":"Retain wherever licensing, policyholder protection, tax, capital or supervisory requirements make them necessary.",
    "reinsurance":"Coordinate regionally; centralize only where risk ownership, capital, governance and regulatory permissions genuinely support it.",
    "shared_services":"Place routine services under documented arm's-length arrangements; keep key risk-taking decisions with the entities that economically own the risks.",
    "substance_rule":"Boards, senior management, underwriting/actuarial authority, capital and systems must align with attributed functions and risks.",
    "tax_rule":"Optimize lawful friction and simplification, not sub-minimum-tax profit parking.",
    "conditions_precedent":[
        "jurisdiction-specific legal and tax opinions",
        "full OECD/GloBE Pillar Two model using verified accounting data",
        "transfer-pricing functional and financial-transactions analysis",
        "actuarial, solvency and capital fungibility analysis",
        "supervisory engagement and required approvals",
        "UK listing, shareholder and corporate-governance review",
        "implementation tax, exit tax, stamp duty, VAT/GST and policyholder impact analysis"
    ]
}
legacy_index = legacy_complexity + len(entities)*1.5
target_index = 4 + len(entities)*0.75 + 12  # parent + regional platforms + retained local complexity proxy
simplification = 1-target_index/legacy_index

print("SELECTED SYNTHETIC ARCHITECTURE")
print(json.dumps(architecture_contract,indent=2))
print("\nSynthetic structural-complexity reduction:",f"{simplification:.1%}")
print("Deterministic score:",round(float(winner.det_score),4),
      "Final score:",round(float(winner.final_score),4),
      "Red flags:",winner.red_flags)
```

### Code Unit 10 — Board report, provenance and references-ready export

The final computational unit assembles the experiment into a board-style report. It records the synthetic company profile, selected architecture, quantitative scores, swarm consensus, dissent, red-team flags, required safeguards and implementation phases. It also creates machine-readable CSV and JSON artifacts that can be downloaded from Colab and inspected independently of the notebook.

The implementation roadmap is staged. Phase 1 is evidence: legal-entity inventory, tax attributes, regulatory licenses, capital, contracts and reinsurance. Phase 2 is design validation: jurisdiction-specific legal and tax opinions, Pillar Two calculations, transfer-pricing functional analysis and supervisory engagement. Phase 3 is sequencing: establish or strengthen regional platforms, migrate functions and people, amend reinsurance, obtain approvals, and only then transfer contracts or entities where feasible. Phase 4 is stabilization: monitor effective tax rates, capital, solvency, service quality, claims performance, transfer-pricing outcomes and regulatory commitments.

Provenance is part of the deliverable. The report records the random seed, model identifier, whether live LLM mode was used, the number of agents and the candidate universe. A real system would add prompt hashes, source-document versions and human approvals.

The notebook closes by emphasizing that a sophisticated swarm should make the decision process **more inspectable**, not less. The value of the 50 agents lies in surfacing specialist conflicts and hidden constraints. The final recommendation is accepted only after deterministic validation. This is a useful model for financial institutions adopting autonomous systems: AI can broaden the range of hypotheses and accelerate multidisciplinary analysis, while governance, evidence and accountable human judgment remain the mechanisms that authorize real-world action.
