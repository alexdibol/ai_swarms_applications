## Research objective

This notebook asks a deliberately unconventional question: can the conceptual architecture of **quantum teleportation** inspire a coherent and testable theory of financial risk management?

The analogy must be handled with discipline. Financial risk is not a quantum state, a portfolio is not a qubit, and this notebook does not claim that quantum teleportation literally transfers market risk from one institution to another. Quantum teleportation is instead used as a **formal source of architectural ideas**. In the standard protocol, an unknown quantum state is not physically transported as an object. Rather, a previously shared entangled resource, a local joint measurement, a classical message, and a conditional reconstruction operation allow the informational state to be reproduced at another location. The original state is not cloned. The protocol is constrained by information, resources, locality, and the requirement for a corrective operation.

Those structural ideas are remarkably suggestive for risk management. Conventional finance often describes risk as something that is “transferred”: an institution hedges an exposure with a derivative, purchases insurance, securitizes a portfolio, novates a contract, or offsets a position. Yet the language of transfer can conceal an important fact. The underlying economic uncertainty has not disappeared. What changes is the **distribution, representation, conditional ownership, observability, and response architecture** associated with that uncertainty. Hedging one exposure may create basis risk, counterparty risk, liquidity risk, collateral requirements, model risk, or nonlinear residual exposure elsewhere.

The central hypothesis explored here is therefore:

> **Risk should be managed as a state that can be encoded, decomposed, conditionally reconstructed, and redistributed across a network, but cannot be costlessly copied, erased, or transferred without residual information and resource requirements.**

We call the resulting research program **Teleportation-Inspired Risk Management (TIRM)**. The name describes the analogy, not a claim of quantum implementation.

The mathematical starting point is a risk state

$
\mathcal{R}_t = (\mathbf{x}_t,\Sigma_t,\mathbf{\ell}_t,\mathbf{c}_t,\mathbf{m}_t,\mathcal{N}_t),
$

where \(\mathbf{x}_t\) represents exposures, \(\Sigma_t\) the covariance or scenario structure, \(\mathbf{\ell}_t\) liquidity conditions, \(\mathbf{c}_t\) counterparty and collateral states, \(\mathbf{m}_t\) model uncertainty, and \(\mathcal{N}_t\) the financial network. A risk-management intervention does not simply subtract a scalar quantity called “risk.” It transforms this state.

The teleportation analogy suggests decomposing that transformation into four operations. First comes **resource preparation**: hedges, collateral, capital, insurance capacity, contingent liquidity, or contractual relationships must exist before stress. Second comes **measurement**: the institution observes an imperfect representation of its state through prices, Greeks, scenarios, stress tests, positions, and risk models. Third comes **communication**: information and contractual instructions must move through governance and market infrastructure. Fourth comes **conditional reconstruction**: trades, margin transfers, capital actions, or contractual payoffs reconstruct a desired exposure profile elsewhere in the system.

This immediately suggests a conservation-like principle. Risk cannot be made to vanish merely because a local metric falls. A hedge may reduce local variance while increasing liquidity sensitivity or counterparty dependence. We therefore define a generalized residual:

$
\Delta \mathcal{R}
=
\mathcal{R}_{\mathrm{post}}
-
\mathcal{T}(\mathcal{R}_{\mathrm{pre}}),
$

where \(\mathcal{T}\) is the intended risk transformation. The norm of this residual measures **teleportation error**: basis risk, slippage, latency, model error, counterparty failure, collateral friction, and unobserved state variables.

A second important analogy concerns the no-cloning theorem. In finance, a complete risk state cannot generally be duplicated at zero cost and with perfect fidelity. Replication requires instruments, balance sheet, market depth, information, and model assumptions. The notebook explores a financial **no-free-cloning principle**: any attempt to reproduce a contingent payoff or risk state with high fidelity consumes scarce resources and introduces implementation error or contingent obligations.

A third analogy concerns entanglement. The notebook does not equate statistical correlation with quantum entanglement. Instead, “entanglement” is used as a deliberately restricted label for **pre-existing contractual or balance-sheet coupling** that makes conditional redistribution possible. A derivative counterparty, clearing arrangement, collateral agreement, insurance contract, or committed liquidity facility creates a state-dependent connection between institutions. Such coupling is a resource in normal conditions and a channel of systemic propagation in stress.

The notebook ultimately asks the swarm to determine whether these ideas can be organized into a theory with explicit axioms. Candidate axioms include conservation of residual risk, costly reconstruction, dependence on pre-positioned capacity, measurement incompleteness, communication latency, nonlocal redistribution through contractual networks, and the impossibility of judging system safety from local risk reduction alone.

A theory is useful only if it generates propositions that could be wrong. The notebook therefore requires empirical implications. Examples include whether hedge effectiveness deteriorates predictably with liquidity and communication latency; whether reductions in local market risk are accompanied by measurable increases in counterparty or funding dependence; whether institutions with richer pre-positioned contingent resources reconstruct target exposures more faithfully during shocks; and whether network concentration predicts the gap between local and system-wide risk reduction.

The ambition is to demonstrate how advanced autonomous research systems can be used not only to solve a known equation, but to participate in the disciplined construction of a new theoretical framework while preserving the distinction between metaphor, mathematics, and empirical science.

## Quantum Teleportation in 10 Easy-to-Understand Ideas

- **1. Teleportation does not mean transporting an object from one place to another.**  
  In quantum physics, teleportation means transferring the **quantum state — the information describing a system —** from one location to another. The physical particle itself does not travel between the two locations.

- **2. Think of it as transferring a state rather than transferring an asset.**  
  For a financial practitioner, imagine that we are interested not in moving a security itself, but in transferring the complete information required to reproduce a particular state somewhere else.

- **3. Teleportation begins with two quantum systems that are entangled.**  
  Entanglement creates an unusual relationship between two particles: their states are linked in a way that cannot be described simply as two independent objects. They form, in an important sense, a **joint system**.

- **4. The two entangled particles can then be separated.**  
  One can be located with Alice and the other with Bob — the conventional names used in explanations of quantum communication. Distance does not eliminate the quantum relationship created between them.

- **5. Alice also possesses the quantum state that she wants to transmit.**  
  She therefore has two things: the unknown state to be transferred and her half of the entangled pair. Bob possesses the other half of the entangled pair.

- **6. Alice performs a special joint measurement.**  
  She measures the state she wants to transmit together with her entangled particle. This operation produces a small amount of ordinary, classical information that can be communicated to Bob.

- **7. Something remarkable happens during this process: the original state disappears.**  
  Quantum teleportation does **not** create a second identical copy. The original quantum state is destroyed as part of the measurement process. The information is therefore transferred rather than duplicated.

- **8. Alice must send Bob the measurement result through an ordinary communication channel.**  
  This is crucial. Teleportation does not allow information to travel instantaneously or faster than light. Bob cannot reconstruct the state until Alice's classical message reaches him.

- **9. Bob uses Alice's message to transform his particle.**  
  Once Bob receives the classical information, he performs the appropriate operation on his half of the entangled pair. His particle then acquires the quantum state that originally existed on Alice's side.

- **10. The deeper lesson is about information, dependence, and state transformation.**  
  Teleportation shows that information can be redistributed across a system through a combination of **shared dependence, local intervention, measurement, communication, and reconstruction**. For our risk-management experiment, these ideas are valuable not because financial risks literally obey quantum mechanics, but because they inspire a different way of thinking about how risk states may be distributed, transmitted, transformed, observed, and reconstructed across an interconnected financial system.

## From Quantum Teleportation to Financial Risk: 10 Intuitive Connections

The analogy must be used carefully: **financial markets are not quantum systems**. We are borrowing the architecture of quantum teleportation as an intellectual device for asking different questions about risk.

- **1. Quantum state → Financial risk state**  
  In teleportation, what matters is not the physical particle but its **state**. In finance, we can similarly distinguish an asset from its risk state. A portfolio is not simply a collection of securities; at any moment it has a state defined by exposures, correlations, liquidity, leverage, volatility, funding conditions, counterparty dependencies, and market regime.

- **2. Entanglement → Interdependence**  
  Financial institutions and markets are deeply interconnected. A bank, hedge fund, clearinghouse, insurer, and sovereign may appear to be separate entities while sharing common exposures and dependencies. The analogy with entanglement encourages us to model **relationships as part of the risk state itself**, rather than treating every institution independently.

- **3. Separation does not imply independence → Geographic and institutional distance can be misleading**  
  Two institutions may operate in different countries and hold apparently different portfolios yet remain strongly connected through common collateral, funding markets, counterparties, currencies, or underlying factors. Physical or legal separation therefore does not necessarily imply economic independence.

- **4. Teleportation transfers a state → Risk can migrate without the original asset moving**  
  Financial risk frequently changes location without the underlying asset changing hands. Derivatives, guarantees, insurance, reinsurance, securitization, collateral agreements, and hedging can redistribute exposures across institutions. The important question becomes: **Where does the risk state ultimately reside?**

- **5. Measurement changes what we know about the system → Risk measurement is part of risk management**  
  Teleportation requires measurement. Financial institutions similarly observe their systems through VaR, stress tests, sensitivities, liquidity measures, margin requirements, scenario analysis, and other diagnostics. These measurements are not merely descriptive: once reported, they can trigger limits, margin calls, hedging, deleveraging, or regulatory intervention and therefore affect subsequent behavior.

- **6. Classical information is still necessary → Markets require communication channels**  
  Quantum teleportation cannot be completed through entanglement alone; ordinary information must also be transmitted. Likewise, financial systems depend upon prices, margin calls, disclosures, ratings, regulatory reports, collateral notices, trading signals, and counterparty information. **Connections transmit exposure; information determines how participants respond to it.**

- **7. Reconstruction occurs somewhere else → Exposure can reappear in another part of the system**  
  Hedging a risk does not necessarily eliminate that risk from the financial system. It may reconstruct it elsewhere. A bank buying protection reduces one exposure while creating an exposure for the protection seller. An insurer can transfer risk to a reinsurer, which may subsequently transfer part of it to capital markets. The relevant question is therefore not merely *"Did we reduce our risk?"* but **"Where did the risk go?"**

- **8. No cloning → Risk transfer should not automatically be interpreted as risk disappearance**  
  Quantum mechanics prevents the arbitrary copying of an unknown quantum state. The financial analogy is different but useful: accounting representations can make the same economic risk appear in several places, while contractual transfers can make it appear to disappear from one balance sheet. A rigorous framework should distinguish **risk creation, risk transfer, risk transformation, risk sharing, and genuine risk extinction**.

- **9. Protocol matters → The architecture of risk transfer matters**  
  Teleportation succeeds only because a precise sequence connects entanglement, measurement, communication, and reconstruction. Financial risk transfer is similarly dependent on architecture. Collateral rules, netting agreements, clearing arrangements, capital requirements, liquidity constraints, legal enforceability, and settlement mechanisms determine whether a hedge actually behaves as expected under stress.

- **10. The system becomes the natural unit of analysis → From portfolio risk to network risk**  
  The deepest lesson for our experiment is methodological. Traditional risk management frequently asks: **"What risks does this institution hold?"** The teleportation-inspired perspective asks a broader question: **"How is the risk state distributed across the system, how can it migrate, what information reveals that migration, and where will it be reconstructed after participants respond?"** This moves the analysis from isolated balance sheets toward **dynamic networks of interconnected risk states**.

## CELL 2 — Build a toy financial world in which risk can be transformed but not erased

Before asking agents to theorize, the notebook creates a simple numerical environment. A portfolio has market-factor exposures, covariance risk, liquidity sensitivity, and counterparty dependence. A hedge transforms the exposure vector, but its effectiveness is degraded by basis error, liquidity friction, execution latency, and counterparty reliability.

The purpose is not to claim that this toy model is the theory. It gives the swarm's later propositions something concrete against which to be illustrated. A local variance measure can fall while a broader system-risk functional remains unchanged or even increases because the intervention creates other dependencies.

This distinction is central to the teleportation analogy. The financial state is transformed and redistributed; it is not annihilated.

```python
# @title
# CELL 2 — Build a toy financial world in which risk can be transformed but not erased
def generalized_risk(x, Sigma, liquidity_burden=0.0, counterparty_burden=0.0,
                     model_burden=0.0, w=(1.0,0.35,0.45,0.20)):
    market = float(np.sqrt(max(x @ Sigma @ x, 0.0)))
    total = (w[0]*market + w[1]*liquidity_burden +
             w[2]*counterparty_burden + w[3]*model_burden)
    return {"market":market, "generalized":total}

Sigma = np.array([
    [0.040, 0.018, 0.010],
    [0.018, 0.090, 0.015],
    [0.010, 0.015, 0.025],
])
x0 = np.array([1.0, 0.45, -0.20])

pre = generalized_risk(x0, Sigma)
print("Initial exposure state:", x0)
print("Initial risk:", pre)

def risk_reconstruction(x, hedge_vector, hedge_ratio,
                        basis_error, liquidity, latency, cp_reliability):
    intended = x - hedge_ratio*hedge_vector
    implementation_loss = basis_error + 0.25*(1-liquidity) + 0.20*latency
    realized_ratio = hedge_ratio * max(0.0, 1-implementation_loss)
    realized = x - realized_ratio*hedge_vector
    liquidity_burden = hedge_ratio*(1-liquidity)
    counterparty_burden = hedge_ratio*(1-cp_reliability)
    residual = realized - intended
    fidelity = math.exp(-np.linalg.norm(residual))
    return intended, realized, residual, fidelity, liquidity_burden, counterparty_burden

h = np.array([0.85,0.25,-0.05])
demo = risk_reconstruction(x0,h,0.8,0.05,0.8,0.1,0.97)
print("Illustrative reconstruction fidelity:", round(demo[3],4))
```

### Pydantic Object vs. JSON Schema

To rephrase your question: "What is the difference between an *instance of a Pydantic model* (a 'Pydantic object') and a *JSON Schema*?"

Here's the distinction:

*   **Pydantic Object (Instance of a Pydantic Model):**
    *   This is a **live Python object** created from a Pydantic model definition (like an `AgentContract` instance).
    *   It **holds actual data** that has been validated against the rules defined in its corresponding Pydantic model.
    *   It exists **in your Python program's memory** and allows you to interact with the structured data using Python methods and attributes.
    *   **Example:** If `AgentContract` is the blueprint, a specific `AgentContract(family='quantum', variant=1, ...)` is a Pydantic object. It's the filled-out form.

*   **JSON Schema:**
    *   This is a **standardized, language-agnostic format** (written in JSON itself) for **describing the structure, types, and constraints of JSON data**.
    *   It's like a **blueprint or contract for JSON documents**. It doesn't contain any actual data; it just defines what the data *should* look like.
    *   It's used for **validation and documentation** of JSON data, allowing other systems or applications to understand and verify the expected format of incoming JSON.
    *   **Pydantic's connection:** Pydantic models can **automatically generate a JSON Schema** that accurately describes the structure they enforce. This allows you to communicate the expected JSON format to external systems.

**In essence:**

*   A **Pydantic object** is the **validated data itself**, living as a Python object.
*   A **JSON Schema** is the **formal description of the data's structure**, typically used for communication and validation between different systems (and often generated *from* a Pydantic model).

### BaseModel as the Parent of Pydantic Models

When you see `class MyModel(BaseModel):` in Pydantic, it means `MyModel` is inheriting from `BaseModel`.

*   **`BaseModel`** is the foundational class provided by the Pydantic library.
    *   It's not the parent of a 'Pydantic object' directly, but rather the parent class for all Pydantic *models*. A 'Pydantic object' is an *instance* of one of these Pydantic models (e.g., an instance of `AgentContract`).

*   **What `BaseModel` provides:** By inheriting from `BaseModel`, your custom Pydantic models (like `AgentContract`, `TestableProposition`, `FinancialMapping` in our notebook) automatically gain a lot of powerful functionality:
    1.  **Data Validation:** `BaseModel` enforces the type hints you define for each field. If incoming data doesn't match the expected type, Pydantic raises a validation error.
    2.  **Data Parsing:** It automatically parses incoming data (e.g., from JSON or Python dictionaries) into the correct Python types.
    3.  **Serialization:** It allows you to easily convert a Pydantic model instance back into standard Python data types (like dictionaries) or JSON strings.
    4.  **Immutability (Optional):** With `model_config = ConfigDict(extra="forbid")` or `frozen=True`, `BaseModel` helps create immutable data structures.
    5.  **Schema Generation:** It can automatically generate a JSON Schema from your model definition, which is useful for documentation and interoperability.
    6.  **Convenience Methods:** It provides methods like `model_dump()`, `model_dump_json()`, `model_validate()`, `model_construct()`, etc., for working with the data.

In essence, `BaseModel` is the magic behind Pydantic that transforms simple Python classes with type hints into robust data validation and serialization tools.

## CELL 4 — Execute the swarm and collect validated theory fragments

This unit has been hardened for the current **Claude Sonnet 5** API rather than carrying forward calling conventions from older Claude generations. The notebook no longer asks the model to emit free-form text that is subsequently rescued with regular expressions. Each agent is called through Anthropic's native structured-output parser against a Pydantic contract, so malformed JSON and missing or wrongly typed fields fail validation instead of silently contaminating the research synthesis.

The call deliberately omits legacy sampling parameters such as `temperature`. Transient rate-limit, timeout, connection, and server failures receive bounded retries with backoff; authentication, permission, and bad-request errors fail immediately. Concurrency is intentionally conservative to reduce synchronized rate-limit failures across fifty agents. Every successful call records attempts, request ID, token usage, latency, and stop reason, while every failed call retains its error type and attempt history.

A minimum swarm-completeness threshold prevents the deterministic layer from presenting a thin or damaged set of responses as a fifty-agent result. Crucially, failed structured outputs are **not repaired with regex or guessed JSON**: they are retried cleanly or recorded as failures.

```python
# @title
# ============================================================
# CELL 4 — EXECUTE, VALIDATE, AND FINALIZE THE 50-AGENT
# TELEPORTATION-INSPIRED RISK MANAGEMENT (TIRM) SWARM
#
# FULLY REPAIRED VERSION
#
# IMPORTANT DESIGN:
#
#   A. If valid first-pass results ALREADY EXIST in memory,
#      this cell REUSES them and makes ZERO API calls.
#
#   B. If no prior results exist, the cell:
#        1. performs ONE preflight call;
#        2. if successful, executes the remaining 49 agents;
#        3. uses LOW effort throughout the first pass.
#
#   C. Agent identity belongs to the experimental controller.
#      Claude's reported family / variant are preserved for
#      provenance but NEVER cause rejection.
#
#   D. Scientific sufficiency:
#        >= 50% validated agents
#        AND
#        100% intellectual-family representation.
#
#   E. If the first pass satisfies the scientific gate,
#      NO recovery API calls are made.
#
#   F. Recovery is available only when scientifically necessary,
#      is family-aware, MEDIUM effort, and capped at 10 calls.
#
#   G. Scientific disagreement is valid research data.
#      Computational / schema failure is missing data.
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

from collections import Counter, defaultdict
from concurrent.futures import ThreadPoolExecutor, as_completed
from datetime import datetime, timezone
import time
import json

import pandas as pd


# ============================================================
# 1. CONFIGURATION
# ============================================================

FIRST_PASS_EFFORT = "low"
RECOVERY_EFFORT = "medium"

MIN_AGENT_COVERAGE = 0.50
MIN_FAMILY_COVERAGE = 1.00

PREFERRED_AGENTS_PER_FAMILY = 2

ENABLE_RECOVERY = True
MAX_RECOVERY_CALLS = 10

FIRST_PASS_WORKERS = min(MAX_WORKERS, 5)

N_FAMILIES = len(FAMILIES)


# ============================================================
# 2. EXPERIMENTAL INTEGRITY
# ============================================================

assert MODEL == "claude-sonnet-5", (
    f"Expected claude-sonnet-5, found {MODEL}."
)

assert N_AGENTS == 50, (
    f"Expected 50 agents, found {N_AGENTS}."
)

assert len(prompts) == N_AGENTS, (
    f"Expected {N_AGENTS} prompts, found {len(prompts)}."
)

assert N_FAMILIES == 10, (
    f"Expected 10 intellectual families, found {N_FAMILIES}."
)


# ============================================================
# 3. HELPERS
# ============================================================

def agent_name(spec):
    return f"{spec['family']}-{int(spec['variant'])}"


def safe_usage(msg, field):
    usage = getattr(msg, "usage", None)

    if usage is None:
        return None

    return getattr(usage, field, None)


# ============================================================
# 4. FAILURE TAXONOMY
# ============================================================

def classify_error(exc):

    message = str(exc).lower()

    if (
        "credit balance is too low" in message
        or "purchase credits" in message
        or "billing" in message
    ):
        return "insufficient_credit"

    if isinstance(exc, RateLimitError):
        return "rate_limit"

    if isinstance(exc, APITimeoutError):
        return "timeout"

    if isinstance(exc, APIConnectionError):
        return "connection_error"

    if isinstance(exc, InternalServerError):
        return "server_error"

    if isinstance(exc, AuthenticationError):
        return "authentication_error"

    if isinstance(exc, PermissionDeniedError):
        return "permission_error"

    if isinstance(exc, BadRequestError):
        return "bad_request"

    if "parsed_output" in message:
        return "no_parsed_output"

    if (
        "validation error" in message
        or "validationerror" in message
        or "validation" in message
    ):
        return "schema_validation"

    if isinstance(exc, ValueError):
        return "value_error"

    return type(exc).__name__


# ============================================================
# 5. CORE AGENT EXECUTION
#
# Claude produces research content.
# The notebook owns experimental identity.
# ============================================================

def execute_agent(
    spec,
    effort=FIRST_PASS_EFFORT,
    phase="first_pass"
):

    started = time.time()

    aid = agent_name(spec)

    canonical_family = spec["family"]
    canonical_variant = int(spec["variant"])

    try:

        msg = client.messages.parse(

            model=MODEL,

            max_tokens=2500,

            output_config={
                "effort": effort
            },

            system=(
                "You are one specialist research agent inside a "
                "50-agent interdisciplinary research swarm studying "
                "Teleportation-Inspired Risk Management (TIRM). "

                "Perform the assigned specialist research task with "
                "scientific precision. Skepticism and disagreement "
                "are welcome. "

                "Financial risk is NOT literally a quantum state. "
                "Statistical dependence is NOT quantum entanglement. "
                "Do NOT claim physical teleportation of financial risk. "

                "Quantum teleportation is used only as a disciplined "
                "source of conceptual architecture. "

                "Every borrowed concept must map to an identifiable "
                "financial mechanism. "

                "Generate falsifiable financial propositions and "
                "explicitly identify where the analogy breaks. "

                "Return a complete object conforming exactly to the "
                "requested structured-output schema."
            ),

            messages=[
                {
                    "role": "user",
                    "content": spec["prompt"]
                }
            ],

            output_format=AgentContract,
        )

        parsed = getattr(
            msg,
            "parsed_output",
            None
        )

        if parsed is None:
            raise ValueError(
                "Claude returned no parsed_output."
            )

        # ----------------------------------------------------
        # Preserve Claude's own labels for provenance.
        # ----------------------------------------------------

        reported_family = getattr(
            parsed,
            "family",
            None
        )

        reported_variant = getattr(
            parsed,
            "variant",
            None
        )

        identity_normalized = (
            reported_family != canonical_family
            or str(reported_variant) != str(canonical_variant)
        )

        # ----------------------------------------------------
        # Canonical experimental identity comes from notebook.
        # ----------------------------------------------------

        parsed.family = canonical_family
        parsed.variant = canonical_variant

        obj = parsed.model_dump(
            mode="json"
        )

        raw = parsed.model_dump_json()

        elapsed = time.time() - started

        return {

            "ok": True,

            "agent_id": aid,

            "family": canonical_family,

            "variant": canonical_variant,

            "reported_family": reported_family,

            "reported_variant": reported_variant,

            "identity_normalized": identity_normalized,

            "phase": phase,

            "effort": effort,

            "obj": obj,

            "raw": raw,

            "seconds": elapsed,

            "request_id": getattr(
                msg,
                "_request_id",
                None
            ),

            "stop_reason": getattr(
                msg,
                "stop_reason",
                None
            ),

            "input_tokens": safe_usage(
                msg,
                "input_tokens"
            ),

            "output_tokens": safe_usage(
                msg,
                "output_tokens"
            ),

            "failure_type": None,

            "error": None,
        }

    except Exception as exc:

        elapsed = time.time() - started

        return {

            "ok": False,

            "agent_id": aid,

            "family": canonical_family,

            "variant": canonical_variant,

            "reported_family": None,

            "reported_variant": None,

            "identity_normalized": False,

            "phase": phase,

            "effort": effort,

            "obj": None,

            "raw": None,

            "seconds": elapsed,

            "request_id": None,

            "stop_reason": None,

            "input_tokens": None,

            "output_tokens": None,

            "failure_type": classify_error(exc),

            "error": repr(exc),
        }


# ============================================================
# 6. POPULATION DIAGNOSTICS
#
# IMPORTANT:
# The canonical key is now consistently "gate_pass".
# This repairs the previous pass_gate / gate_pass bug.
# ============================================================

def diagnose_population(successful_results):

    valid = [
        r
        for r in successful_results
        if r.get("ok", False)
    ]

    family_counts = Counter(
        r["family"]
        for r in valid
    )

    represented_families = [
        family
        for family in FAMILIES.keys()
        if family_counts.get(family, 0) > 0
    ]

    missing_families = [
        family
        for family in FAMILIES.keys()
        if family_counts.get(family, 0) == 0
    ]

    thin_families = [
        family
        for family in FAMILIES.keys()
        if (
            family_counts.get(family, 0) > 0
            and
            family_counts.get(family, 0)
            < PREFERRED_AGENTS_PER_FAMILY
        )
    ]

    agent_coverage = (
        len(valid)
        /
        N_AGENTS
    )

    family_coverage = (
        len(represented_families)
        /
        N_FAMILIES
    )

    gate_pass = (
        agent_coverage >= MIN_AGENT_COVERAGE
        and
        family_coverage >= MIN_FAMILY_COVERAGE
    )

    return {

        "successful_agents":
            len(valid),

        "agent_coverage":
            agent_coverage,

        "family_counts":
            dict(family_counts),

        "represented_families":
            represented_families,

        "missing_families":
            missing_families,

        "thin_families":
            thin_families,

        "family_coverage":
            family_coverage,

        "gate_pass":
            gate_pass,
    }


# ============================================================
# 7. STRUCTURAL FAILURE DETECTION
# ============================================================

STRUCTURAL_FAILURES = {
    "insufficient_credit",
    "authentication_error",
    "permission_error",
    "bad_request",
}


def structural_failure_share(results):

    failures = [
        r
        for r in results
        if not r.get("ok", False)
    ]

    if not failures:
        return 0.0

    structural = sum(
        r.get("failure_type")
        in STRUCTURAL_FAILURES
        for r in failures
    )

    return structural / len(failures)


# ============================================================
# 8. DETECT WHETHER THE SWARM HAS ALREADY RUN
#
# THIS IS THE CRITICAL COST-PROTECTION FEATURE.
#
# In your current runtime we expect:
#
#     first_pass_results      -> 50 records
#     successful_by_agent     -> 46 valid contracts
#
# If those exist, NO API CALLS ARE MADE.
# ============================================================

existing_first_pass = globals().get(
    "first_pass_results",
    None
)

existing_successful = globals().get(
    "successful_by_agent",
    None
)


REUSE_EXISTING_RESULTS = (

    isinstance(existing_first_pass, list)

    and

    len(existing_first_pass) == N_AGENTS

    and

    isinstance(existing_successful, dict)

    and

    len(existing_successful) > 0
)


# ============================================================
# 9. HEADER
# ============================================================

print("=" * 88)

print(
    "TELEPORTATION-INSPIRED RISK MANAGEMENT (TIRM)"
)

print(
    "50-AGENT INTERDISCIPLINARY RESEARCH SWARM"
)

print("=" * 88)

print()


# ============================================================
# 10A. REUSE EXISTING RESULTS
#
# YOUR CURRENT NOTEBOOK SHOULD ENTER HERE.
#
# ZERO API CALLS.
# ============================================================

if REUSE_EXISTING_RESULTS:

    print(
        "✓ EXISTING FIRST-PASS SWARM DETECTED IN MEMORY"
    )

    print()

    print(
        f"Execution records found : "
        f"{len(existing_first_pass)}"
    )

    print(
        f"Valid contracts found   : "
        f"{len(existing_successful)}"
    )

    print()

    print(
        "NO ANTHROPIC API CALLS WILL BE MADE."
    )

    print(
        "The existing research population will be finalized."
    )

    print()

    first_pass_results = existing_first_pass

    successful_by_agent = existing_successful

    swarm_execution_mode = (
        "REUSED_EXISTING_RESULTS"
    )

    swarm_seconds = 0.0


# ============================================================
# 10B. OTHERWISE EXECUTE THE SWARM
#
# This branch runs ONLY if the existing 50-agent results are
# not present in memory.
# ============================================================

else:

    print(
        "No complete prior first-pass population was found."
    )

    print()

    print(
        "A new swarm execution is required."
    )

    print()

    print("=" * 88)

    print(
        "PREFLIGHT — ONE LOW-EFFORT PRODUCTION AGENT"
    )

    print("=" * 88)

    print()

    preflight_spec = prompts[0]

    preflight_result = execute_agent(
        preflight_spec,
        effort=FIRST_PASS_EFFORT,
        phase="preflight"
    )

    if not preflight_result["ok"]:

        print(
            "✗ PREFLIGHT FAILED"
        )

        print()

        print(
            f"Agent        : "
            f"{preflight_result['agent_id']}"
        )

        print(
            f"Failure type : "
            f"{preflight_result['failure_type']}"
        )

        print()

        print(
            "COMPLETE ERROR"
        )

        print("-" * 88)

        print(
            preflight_result["error"]
        )

        print("-" * 88)

        print()

        print(
            "THE 50-AGENT SWARM HAS NOT BEEN LAUNCHED."
        )

        raise RuntimeError(
            "Cell 4 preflight failed."
        )

    print(
        "✓ PREFLIGHT PASSED"
    )

    print()

    print(
        f"Agent                 : "
        f"{preflight_result['agent_id']}"
    )

    print(
        f"Canonical family      : "
        f"{preflight_result['family']}"
    )

    print(
        f"Model-reported family : "
        f"{preflight_result['reported_family']}"
    )

    print(
        f"Identity normalized   : "
        f"{preflight_result['identity_normalized']}"
    )

    print()

    print(
        "Launching remaining 49 LOW-effort agents."
    )

    swarm_started = time.time()

    first_pass_results = [
        preflight_result
    ]

    remaining_specs = prompts[1:]

    completed = 1
    successes = 1
    failures = 0

    print()
    print("=" * 88)

    print(
        "STAGE 1 — LOW-EFFORT FIRST PASS"
    )

    print("=" * 88)

    with ThreadPoolExecutor(
        max_workers=FIRST_PASS_WORKERS
    ) as executor:

        future_to_spec = {

            executor.submit(
                execute_agent,
                spec,
                FIRST_PASS_EFFORT,
                "first_pass"
            ): spec

            for spec in remaining_specs
        }

        for future in as_completed(
            future_to_spec
        ):

            spec = future_to_spec[future]

            try:

                result = future.result()

            except Exception as exc:

                result = {

                    "ok": False,

                    "agent_id":
                        agent_name(spec),

                    "family":
                        spec["family"],

                    "variant":
                        int(spec["variant"]),

                    "reported_family":
                        None,

                    "reported_variant":
                        None,

                    "identity_normalized":
                        False,

                    "phase":
                        "first_pass",

                    "effort":
                        FIRST_PASS_EFFORT,

                    "obj":
                        None,

                    "raw":
                        None,

                    "seconds":
                        None,

                    "request_id":
                        None,

                    "stop_reason":
                        None,

                    "input_tokens":
                        None,

                    "output_tokens":
                        None,

                    "failure_type":
                        "executor_error",

                    "error":
                        repr(exc),
                }

            first_pass_results.append(
                result
            )

            completed += 1

            if result["ok"]:

                successes += 1
                symbol = "✓"

            else:

                failures += 1
                symbol = "✗"

            elapsed = (
                result["seconds"]
                if result["seconds"] is not None
                else 0.0
            )

            print(

                f"{symbol} "
                f"{completed:02d}/{N_AGENTS} | "

                f"{result['agent_id']:<32} | "

                f"{elapsed:6.1f}s | "

                f"success={successes:02d} | "

                f"failed={failures:02d}",

                flush=True
            )

    successful_by_agent = {

        r["agent_id"]: r

        for r in first_pass_results

        if r["ok"]
    }

    swarm_seconds = (
        time.time()
        -
        swarm_started
    )

    swarm_execution_mode = (
        "NEW_API_EXECUTION"
    )


# ============================================================
# 11. VALIDATE THE AVAILABLE FIRST-PASS POPULATION
# ============================================================

first_pass_failures = [

    r

    for r in first_pass_results

    if not r.get("ok", False)
]


first_diag = diagnose_population(

    list(
        successful_by_agent.values()
    )
)


print()
print("=" * 88)

print(
    "FIRST-PASS RESEARCH POPULATION"
)

print("=" * 88)

print()

print(
    f"Validated contracts : "
    f"{first_diag['successful_agents']}/{N_AGENTS}"
)

print(
    f"Agent coverage      : "
    f"{first_diag['agent_coverage']:.1%}"
)

print(
    f"Families represented: "
    f"{len(first_diag['represented_families'])}/{N_FAMILIES}"
)

print(
    f"Family coverage     : "
    f"{first_diag['family_coverage']:.1%}"
)


# ============================================================
# 12. FAMILY REPRESENTATION
# ============================================================

print()
print(
    "FAMILY REPRESENTATION"
)

print("-" * 88)


for family in FAMILIES.keys():

    count = first_diag[
        "family_counts"
    ].get(
        family,
        0
    )

    if count == 0:

        marker = "MISSING"

    elif (
        count
        < PREFERRED_AGENTS_PER_FAMILY
    ):

        marker = "THIN"

    else:

        marker = "OK"

    print(

        f"{family:<28} "
        f"{count}/5   "
        f"{marker}"
    )


# ============================================================
# 13. IDENTITY NORMALIZATION
# ============================================================

normalized_first_pass = [

    r

    for r in successful_by_agent.values()

    if r.get(
        "identity_normalized",
        False
    )
]


print()
print(
    "IDENTITY NORMALIZATION"
)

print("-" * 88)

print(
    f"Contracts normalized : "
    f"{len(normalized_first_pass)}"
)

print(
    "These are VALID contracts; only orchestration metadata "
    "was canonicalized."
)


# ============================================================
# 14. FAILURE TAXONOMY
# ============================================================

if first_pass_failures:

    print()
    print(
        "FIRST-PASS FAILURE TYPES"
    )

    print("-" * 88)

    failure_counts = Counter(

        r.get(
            "failure_type",
            "unknown"
        )

        for r in first_pass_failures
    )

    for failure_type, count in (
        failure_counts.most_common()
    ):

        print(
            f"{failure_type:<32} "
            f"{count:>3}"
        )


# ============================================================
# 15. SCIENTIFIC SUFFICIENCY GATE
#
# FIXED:
# We consistently use first_diag["gate_pass"].
# ============================================================

print()
print("=" * 88)

print(
    "SCIENTIFIC SUFFICIENCY CHECK"
)

print("=" * 88)

print()


if first_diag["gate_pass"]:

    print(
        "✓ FIRST-PASS SCIENTIFIC GATE PASSED"
    )

    print()

    print(
        f"{first_diag['successful_agents']} validated "
        f"contracts provide "
        f"{first_diag['agent_coverage']:.1%} agent coverage."
    )

    print(
        f"All {N_FAMILIES} intellectual families are represented."
    )

    print()

    print(
        "NO RECOVERY API CALLS ARE SCIENTIFICALLY NECESSARY."
    )


else:

    print(
        "⚠ FIRST-PASS SCIENTIFIC GATE NOT YET SATISFIED"
    )


# ============================================================
# 16. RECOVERY DECISION
#
# IMPORTANT:
# With your current 46/50, 10/10 result,
# this entire recovery branch is SKIPPED.
# ============================================================

recovery_results = []

recovery_calls_used = 0


if first_diag["gate_pass"]:

    print()
    print("=" * 88)

    print(
        "RECOVERY NOT REQUIRED"
    )

    print("=" * 88)

    print()

    print(
        "The existing first-pass research population already "
        "satisfies both numerical and disciplinary requirements."
    )

    print()

    print(
        "Failed schema observations are retained in the audit "
        "trail rather than purchased again."
    )


elif not ENABLE_RECOVERY:

    print()

    print(
        "Recovery disabled by configuration."
    )


else:

    structural_share = (
        structural_failure_share(
            first_pass_results
        )
    )

    if (
        first_pass_failures
        and
        structural_share >= 0.50
    ):

        print()
        print("=" * 88)

        print(
            "RECOVERY BLOCKED — STRUCTURAL FAILURE PATTERN"
        )

        print("=" * 88)

        print()

        print(
            f"{structural_share:.1%} of failures are "
            "structural API failures."
        )

        print(
            "No additional API calls will be purchased."
        )

    else:

        print()
        print("=" * 88)

        print(
            "STAGE 2 — TARGETED MEDIUM-EFFORT RECOVERY"
        )

        print("=" * 88)

        print()

        print(
            f"Maximum recovery budget: "
            f"{MAX_RECOVERY_CALLS} calls."
        )

        spec_by_agent = {

            agent_name(spec):
                spec

            for spec in prompts
        }

        remaining_recovery_specs = [

            spec_by_agent[
                result["agent_id"]
            ]

            for result
            in first_pass_failures

            if result["agent_id"]
            in spec_by_agent
        ]

        def recovery_priority(spec):

            current_diag = diagnose_population(

                list(
                    successful_by_agent.values()
                )
            )

            family_count = current_diag[
                "family_counts"
            ].get(
                spec["family"],
                0
            )

            if family_count == 0:
                category = 0

            elif (
                family_count
                < PREFERRED_AGENTS_PER_FAMILY
            ):
                category = 1

            else:
                category = 2

            return (

                category,

                family_count,

                list(
                    FAMILIES.keys()
                ).index(
                    spec["family"]
                ),

                int(
                    spec["variant"]
                )
            )

        while (

            remaining_recovery_specs

            and

            recovery_calls_used
            < MAX_RECOVERY_CALLS
        ):

            current_diag = diagnose_population(

                list(
                    successful_by_agent.values()
                )
            )

            if current_diag["gate_pass"]:
                break

            remaining_recovery_specs.sort(
                key=recovery_priority
            )

            spec = (
                remaining_recovery_specs.pop(0)
            )

            recovery_calls_used += 1

            print(

                f"Recovery "
                f"{recovery_calls_used:02d}/"
                f"{MAX_RECOVERY_CALLS} | "

                f"{agent_name(spec):<32}",

                end=" | ",

                flush=True
            )

            result = execute_agent(

                spec,

                effort=RECOVERY_EFFORT,

                phase="recovery"
            )

            recovery_results.append(
                result
            )

            if result["ok"]:

                successful_by_agent[
                    result["agent_id"]
                ] = result

                print(
                    f"✓ success "
                    f"({result['seconds']:.1f}s)"
                )

            else:

                print(
                    f"✗ "
                    f"{result['failure_type']} "
                    f"({result['seconds']:.1f}s)"
                )

                if (
                    result["failure_type"]
                    in STRUCTURAL_FAILURES
                ):

                    print()

                    print(
                        "Structural failure encountered."
                    )

                    print(
                        "Recovery stopped to protect budget."
                    )

                    break


# ============================================================
# 17. FINAL VALIDATED POPULATION
# ============================================================

family_order = {

    family: index

    for index, family

    in enumerate(
        FAMILIES.keys()
    )
}


ok = sorted(

    list(
        successful_by_agent.values()
    ),

    key=lambda r: (

        family_order.get(
            r["family"],
            999
        ),

        int(
            r["variant"]
        )
    )
)


final_diag = diagnose_population(
    ok
)


# ============================================================
# 18. FINAL THREE-STATE RESEARCH GATE
# ============================================================

if final_diag["gate_pass"]:

    RESEARCH_GATE = "PASS"
    SWARM_READY = True

elif (
    final_diag["agent_coverage"]
    >= MIN_AGENT_COVERAGE
):

    RESEARCH_GATE = (
        "PASS_WITH_CAUTION"
    )

    SWARM_READY = False

else:

    RESEARCH_GATE = "FAIL"
    SWARM_READY = False


# ============================================================
# 19. COMPLETE ATTEMPT HISTORY
# ============================================================

all_attempts = (

    list(
        first_pass_results
    )

    +

    list(
        recovery_results
    )
)


attempt_log = pd.DataFrame(

    [

        {

            "agent_id":
                r.get("agent_id"),

            "family":
                r.get("family"),

            "variant":
                r.get("variant"),

            "reported_family":
                r.get("reported_family"),

            "reported_variant":
                r.get("reported_variant"),

            "identity_normalized":
                r.get(
                    "identity_normalized",
                    False
                ),

            "phase":
                r.get("phase"),

            "effort":
                r.get("effort"),

            "success":
                r.get("ok"),

            "seconds":
                r.get("seconds"),

            "input_tokens":
                r.get("input_tokens"),

            "output_tokens":
                r.get("output_tokens"),

            "failure_type":
                r.get("failure_type"),

            "error":
                r.get("error"),
        }

        for r in all_attempts
    ]
)


# ============================================================
# 20. FINAL FAILURE TABLE
# ============================================================

successful_ids = set(
    successful_by_agent.keys()
)


bad = []


for spec in prompts:

    aid = agent_name(spec)

    if aid in successful_ids:
        continue

    history = [

        r

        for r in all_attempts

        if r.get("agent_id") == aid
    ]

    if history:

        last = history[-1]

        bad.append(
            {

                "agent_id":
                    aid,

                "family":
                    spec["family"],

                "variant":
                    int(spec["variant"]),

                "attempts":
                    len(history),

                "last_phase":
                    last.get("phase"),

                "last_effort":
                    last.get("effort"),

                "failure_type":
                    last.get("failure_type"),

                "error":
                    last.get("error"),
            }
        )

    else:

        bad.append(
            {

                "agent_id":
                    aid,

                "family":
                    spec["family"],

                "variant":
                    int(spec["variant"]),

                "attempts":
                    0,

                "last_phase":
                    None,

                "last_effort":
                    None,

                "failure_type":
                    "not_executed",

                "error":
                    "No execution record.",
            }
        )


failure_table = pd.DataFrame(
    bad
)


# ============================================================
# 21. SUCCESSFUL-AGENT PROVENANCE
# ============================================================

provenance = pd.DataFrame(

    [

        {

            "agent_id":
                r.get("agent_id"),

            "family":
                r.get("family"),

            "variant":
                r.get("variant"),

            "reported_family":
                r.get("reported_family"),

            "reported_variant":
                r.get("reported_variant"),

            "identity_normalized":
                r.get(
                    "identity_normalized",
                    False
                ),

            "phase":
                r.get("phase"),

            "effort":
                r.get("effort"),

            "seconds":
                r.get("seconds"),

            "request_id":
                r.get("request_id"),

            "stop_reason":
                r.get("stop_reason"),

            "input_tokens":
                r.get("input_tokens"),

            "output_tokens":
                r.get("output_tokens"),
        }

        for r in ok
    ]
)


# ============================================================
# 22. COMPUTATIONAL ACCOUNTING
# ============================================================

total_api_calls_recorded = len(
    all_attempts
)


total_input_tokens = sum(

    r.get(
        "input_tokens"
    ) or 0

    for r in all_attempts
)


total_output_tokens = sum(

    r.get(
        "output_tokens"
    ) or 0

    for r in all_attempts
)


low_effort_successes = sum(

    1

    for r in ok

    if r.get("effort")
    == FIRST_PASS_EFFORT
)


medium_effort_successes = sum(

    1

    for r in ok

    if r.get("effort")
    == RECOVERY_EFFORT
)


normalized_successes = sum(

    bool(
        r.get(
            "identity_normalized",
            False
        )
    )

    for r in ok
)


schema_validation_failures = sum(

    r.get("failure_type")
    == "schema_validation"

    for r in all_attempts

    if not r.get("ok", False)
)


# ============================================================
# 23. FINAL REPORT
# ============================================================

print()
print("=" * 88)

print(
    "FINAL TIRM SWARM REPORT"
)

print("=" * 88)


print()
print(
    "EXECUTION MODE"
)

print("-" * 88)

print(
    f"{swarm_execution_mode}"
)


if (
    swarm_execution_mode
    == "REUSED_EXISTING_RESULTS"
):

    print(
        "API calls made during THIS execution of Cell 4: 0"
    )


print()
print(
    "RESEARCH POPULATION"
)

print("-" * 88)

print(
    f"Commissioned agents       : "
    f"{N_AGENTS}"
)

print(
    f"Validated agents          : "
    f"{len(ok)}/{N_AGENTS}"
)

print(
    f"Agent coverage            : "
    f"{final_diag['agent_coverage']:.1%}"
)

print(
    f"Families represented      : "
    f"{len(final_diag['represented_families'])}/{N_FAMILIES}"
)

print(
    f"Family coverage           : "
    f"{final_diag['family_coverage']:.1%}"
)


# ============================================================
# 24. FINAL FAMILY REPRESENTATION
# ============================================================

print()
print(
    "FINAL FAMILY REPRESENTATION"
)

print("-" * 88)


for family in FAMILIES.keys():

    count = final_diag[
        "family_counts"
    ].get(
        family,
        0
    )

    if count == 0:

        marker = "MISSING"

    elif (
        count
        < PREFERRED_AGENTS_PER_FAMILY
    ):

        marker = "THIN"

    else:

        marker = "OK"

    print(

        f"{family:<28} "
        f"{count}/5   "
        f"{marker}"
    )


# ============================================================
# 25. COMPUTATIONAL ECONOMY
# ============================================================

print()
print(
    "COMPUTATIONAL ECONOMY"
)

print("-" * 88)

print(
    f"LOW-effort successes      : "
    f"{low_effort_successes}"
)

print(
    f"MEDIUM-effort recoveries  : "
    f"{medium_effort_successes}"
)

print(
    f"Recovery calls used       : "
    f"{recovery_calls_used}"
)

print(
    f"Historical API attempts   : "
    f"{total_api_calls_recorded}"
)

print(
    f"Input tokens recorded     : "
    f"{total_input_tokens:,}"
)

print(
    f"Output tokens recorded    : "
    f"{total_output_tokens:,}"
)


# ============================================================
# 26. IDENTITY PROVENANCE
# ============================================================

print()
print(
    "IDENTITY PROVENANCE"
)

print("-" * 88)

print(
    f"Validated contracts       : "
    f"{len(ok)}"
)

print(
    f"Model labels normalized   : "
    f"{normalized_successes}"
)

print(
    "Identity normalization is administrative "
    "canonicalization, not research failure."
)


# ============================================================
# 27. FAILURE REPORT
# ============================================================

if len(bad) > 0:

    print()
    print(
        "UNRECOVERED OBSERVATIONS"
    )

    print("-" * 88)

    final_failure_counts = Counter(

        item.get(
            "failure_type",
            "unknown"
        )

        for item in bad
    )

    for failure_type, count in (
        final_failure_counts.most_common()
    ):

        print(

            f"{failure_type:<32} "
            f"{count:>3}"
        )

    print()

    print(
        "These are missing computational observations, "
        "not scientific disagreements."
    )


# ============================================================
# 28. FINAL QUALITY GATE
# ============================================================

print()
print("=" * 88)


if RESEARCH_GATE == "PASS":

    print(
        "✓ RESEARCH QUALITY GATE: PASS"
    )

    print()

    print(
        f"The experiment contains {len(ok)} validated "
        f"research contracts from {N_AGENTS} commissioned agents."
    )

    print(
        f"Agent coverage is "
        f"{final_diag['agent_coverage']:.1%}."
    )

    print(
        f"Family coverage is "
        f"{final_diag['family_coverage']:.1%}."
    )

    print()

    print(
        "All ten intellectual traditions are represented."
    )

    print()

    print(
        "The swarm is scientifically sufficient for "
        "cross-disciplinary convergence analysis."
    )

    print()

    print(
        "CELL 5 MAY PROCEED."
    )


elif RESEARCH_GATE == "PASS_WITH_CAUTION":

    print(
        "⚠ RESEARCH QUALITY GATE: PASS WITH CAUTION"
    )

    print()

    print(
        "Numerical coverage is sufficient, but one or more "
        "intellectual families are missing."
    )


else:

    print(
        "✗ RESEARCH QUALITY GATE: FAIL"
    )

    print()

    print(
        "The validated population is below the required "
        "scientific coverage threshold."
    )


print()

print(
    f"SWARM_READY = {SWARM_READY}"
)

print("=" * 88)


# ============================================================
# 29. MACHINE-READABLE STATUS
# ============================================================

swarm_status = {

    "experiment":
        "Teleportation-Inspired Risk Management",

    "model":
        MODEL,

    "execution_mode":
        swarm_execution_mode,

    "commissioned_agents":
        N_AGENTS,

    "validated_agents":
        len(ok),

    "failed_agents":
        len(bad),

    "agent_coverage":
        final_diag[
            "agent_coverage"
        ],

    "intellectual_families":
        N_FAMILIES,

    "represented_families":
        len(
            final_diag[
                "represented_families"
            ]
        ),

    "family_coverage":
        final_diag[
            "family_coverage"
        ],

    "family_counts":
        final_diag[
            "family_counts"
        ],

    "missing_families":
        final_diag[
            "missing_families"
        ],

    "thin_families":
        final_diag[
            "thin_families"
        ],

    "research_gate":
        RESEARCH_GATE,

    "swarm_ready":
        SWARM_READY,

    "minimum_agent_coverage":
        MIN_AGENT_COVERAGE,

    "minimum_family_coverage":
        MIN_FAMILY_COVERAGE,

    "preferred_agents_per_family":
        PREFERRED_AGENTS_PER_FAMILY,

    "first_pass_effort":
        FIRST_PASS_EFFORT,

    "recovery_effort":
        RECOVERY_EFFORT,

    "maximum_recovery_calls":
        MAX_RECOVERY_CALLS,

    "recovery_calls_used":
        recovery_calls_used,

    "low_effort_successes":
        low_effort_successes,

    "medium_effort_recoveries":
        medium_effort_successes,

    "identity_normalizations":
        normalized_successes,

    "schema_validation_failures":
        schema_validation_failures,

    "historical_api_attempts":
        total_api_calls_recorded,

    "input_tokens_recorded":
        total_input_tokens,

    "output_tokens_recorded":
        total_output_tokens,

    "timestamp_utc":
        datetime.now(
            timezone.utc
        ).isoformat(),
}


print()
print(
    "SWARM STATUS"
)

print("-" * 88)

print(

    json.dumps(

        swarm_status,

        indent=2,

        ensure_ascii=False
    )
)
```

In Cell 4, the word **“failed” has a very specific technical meaning**. It does not mean that an agent disagreed with the theory, produced a negative result, or concluded that Teleportation-Inspired Risk Management is wrong. An agent can be deeply skeptical of the proposed theory and still be classified as completely successful.

A **successful agent** is simply an agent for which the entire computational and validation process was completed correctly. The agent received its assigned research prompt, Claude Sonnet 5 produced a response, that response satisfied the required structured-output architecture, Pydantic successfully validated it, and the identity of the responding agent corresponded to the intellectual family and variant that had been requested. At that point, the notebook possesses a valid research contract that can be passed to the subsequent analytical stages.

A **failed agent**, by contrast, is one for which that computational process did not produce a usable validated research contract. Failure could occur because the API repeatedly timed out, a rate limit persisted beyond the permitted retries, there was an authentication or permission problem, Claude did not return a valid structured output, the response could not satisfy the required schema, or the returned family and variant did not correspond to the agent that had actually been requested.

The distinction is important because **failure says nothing about the intellectual position taken by the agent**. Suppose, for example, that one of the agents belonging to the adversarial-falsification family concludes that a central proposition of TIRM is conceptually weak, empirically untestable, or adds no explanatory power relative to conventional financial risk theory. If that conclusion is returned correctly within the required research contract, that agent is classified as successful. Indeed, such a result could be particularly valuable scientifically because the purpose of the swarm is not to manufacture agreement but to subject the proposed theory to heterogeneous intellectual scrutiny.

This means that Cell 4 and Cell 5 deal with two fundamentally different questions.

**Cell 4 is concerned with computational validity.** It asks whether each of the fifty research assignments produced a complete, valid, machine-readable research contract. A successful result represents usable research evidence. A failed result represents missing evidence because the computational process did not successfully deliver the required research object.

**Cell 5 is concerned with scientific content.** Once the valid research contracts have been collected, the notebook can ask what those agents actually concluded. Some may support a particular principle, some may reject it, some may qualify it, some may identify boundary conditions, and others may propose alternative formulations. Those differences are precisely what the research design is intended to capture.

Consequently, an important distinction runs through the entire notebook:

**Computational failure is missing data; scientific disagreement is valuable data.**

The eighty-percent threshold in Cell 4 should therefore also be interpreted in this way. It does not require eighty percent of the agents to agree with TIRM. It requires at least eighty percent of the fifty commissioned research agents to return valid research contracts. Once that minimum research population exists, Cell 5 can investigate how much agreement, disagreement, convergence, criticism, and cross-disciplinary support actually exists within it.

This distinction is especially important for the architecture of the fifty-agent swarm. We deliberately created heterogeneous intellectual traditions and included an adversarial-falsification family because the objective is not consensus by construction. The objective is to discover which ideas survive serious examination from substantially different intellectual perspectives.

Thus, a skeptical agent that successfully argues against TIRM is a **successful agent**. An agent whose API call, structured output, or validation process fails is a **failed agent**.

That separation preserves the scientific integrity of the experiment: **the infrastructure determines whether an observation exists; the subsequent research analysis determines what that observation means.**

## CELL 5 — Validate contracts and measure cross-disciplinary conceptual convergence

The raw responses are normalized into a canonical vocabulary. Candidate principles are mapped into categories such as conservation, reconstruction cost, pre-positioned capacity, measurement incompleteness, communication latency, network redistribution, fidelity loss, no-free-cloning, and local-versus-systemic divergence.

The notebook counts both agent-level and family-level support. Family support is more informative because it asks whether an idea survives translation across disciplines. A principle appearing independently in quantum information, derivatives, networks, and control theory has a different status from an idea repeated by five variants of one family.

This unit produces the conceptual architecture from which the candidate theory will be synthesized.

```python
# @title
## ============================================================
# CELL 5 — VALIDATE CONTRACTS AND MEASURE CROSS-DISCIPLINARY
# CONCEPTUAL CONVERGENCE
#
# TIRM — Teleportation-Inspired Risk Management
#
# PURPOSE
# -------
# Transform the validated 50-agent swarm population produced
# by Cell 4 into a deterministic cross-disciplinary convergence
# analysis.
#
# IMPORTANT:
#   • ZERO API CALLS
#   • Uses ONLY the validated contracts already stored in `ok`
#   • Does NOT attempt to recover failed agents
#   • Does NOT use the obsolete MIN_SUCCESS_RATE rule
#   • Respects the scientific gate established in Cell 4
#
# THREE LEVELS OF CONVERGENCE
# ---------------------------
# 1. Agent support:
#       fraction of validated agents supporting a principle
#
# 2. Family breadth:
#       fraction of intellectual families in which at least
#       one validated agent supports the principle
#
# 3. Equal-family support:
#       average within-family support, giving each intellectual
#       family equal weight regardless of whether 4 or 5 agents
#       survived validation.
#
# This distinction matters because the validated population is
# slightly unbalanced across families.
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. VERIFY UPSTREAM SCIENTIFIC STATE
# ============================================================

required_objects = [
    "ok",
    "SWARM_READY",
    "RESEARCH_GATE",
    "swarm_status",
    "FAMILIES",
    "PRINCIPLES",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Cell 5 cannot proceed because the following "
        f"Cell 4 objects are missing: {missing_objects}"
    )


if not SWARM_READY:
    raise RuntimeError(
        "Cell 4 did not certify the swarm as ready for synthesis."
    )


if RESEARCH_GATE != "PASS":
    raise RuntimeError(
        f"Expected RESEARCH_GATE='PASS', found {RESEARCH_GATE!r}."
    )


if not isinstance(ok, list) or len(ok) == 0:
    raise RuntimeError(
        "The validated swarm population `ok` is empty or invalid."
    )


print("=" * 92)
print("CELL 5 — CROSS-DISCIPLINARY CONCEPTUAL CONVERGENCE")
print("Teleportation-Inspired Risk Management (TIRM)")
print("=" * 92)

print()
print("✓ Cell 4 research population detected")
print(f"✓ Validated contracts available : {len(ok)}")
print(f"✓ Research gate                 : {RESEARCH_GATE}")
print("✓ API calls in Cell 5           : 0")


# ============================================================
# 2. NORMALIZE THE VALIDATED CONTRACTS
#
# The API/Pydantic layer already validated structure.
# Here we normalize the objects for deterministic analysis.
# ============================================================

def normalize_contract(r):

    if not r.get("ok", False):
        raise ValueError(
            f"Non-successful result entered validated population: "
            f"{r.get('agent_id')}"
        )

    o = r.get("obj")

    if not isinstance(o, dict):
        raise ValueError(
            f"Agent {r.get('agent_id')} has no valid contract object."
        )

    required_contract_fields = [
        "supported_principles",
        "financial_mapping",
        "candidate_axioms",
        "mathematical_objects",
        "testable_propositions",
        "analogy_breaks",
        "novel_contribution",
        "confidence",
    ]

    missing = [
        field
        for field in required_contract_fields
        if field not in o
    ]

    if missing:
        raise ValueError(
            f"Agent {r.get('agent_id')} is missing contract fields: "
            f"{missing}"
        )

    # Deduplicate principle votes inside an individual agent.
    # One agent can vote at most once for each principle.
    supported = list(
        dict.fromkeys(
            o["supported_principles"]
        )
    )

    invalid_principles = sorted(
        set(supported) - set(PRINCIPLES)
    )

    if invalid_principles:
        raise ValueError(
            f"Agent {r.get('agent_id')} contains unsupported "
            f"principles: {invalid_principles}"
        )

    return {

        "agent_id":
            r.get("agent_id"),

        "family":
            r["family"],

        "variant":
            int(r["variant"]),

        "supported_principles":
            supported,

        "mapping":
            o["financial_mapping"],

        "axioms":
            o["candidate_axioms"],

        "math":
            o["mathematical_objects"],

        "propositions":
            o["testable_propositions"],

        "analogy_breaks":
            o["analogy_breaks"],

        "novel_contribution":
            o["novel_contribution"],

        "confidence":
            float(o["confidence"]),

        "seconds":
            r.get("seconds"),

        "phase":
            r.get("phase"),

        "effort":
            r.get("effort"),

        "request_id":
            r.get("request_id"),

        "input_tokens":
            r.get("input_tokens"),

        "output_tokens":
            r.get("output_tokens"),

        "identity_normalized":
            bool(
                r.get(
                    "identity_normalized",
                    False
                )
            ),
    }


normalized_contracts = [
    normalize_contract(r)
    for r in ok
]


contracts = pd.DataFrame(
    normalized_contracts
)


# ============================================================
# 3. VERIFY POPULATION REPRESENTATION
# ============================================================

N_VALID = len(contracts)

represented_families = (
    contracts["family"]
    .nunique()
)

family_sizes = (
    contracts
    .groupby("family")
    .size()
    .reindex(
        list(FAMILIES.keys()),
        fill_value=0
    )
)


if represented_families != len(FAMILIES):
    raise RuntimeError(
        "Cell 5 expected all intellectual families to be "
        "represented, but the validated population is incomplete."
    )


print()
print("-" * 92)
print("VALIDATED RESEARCH POPULATION")
print("-" * 92)

print(
    f"Commissioned agents             : "
    f"{N_AGENTS}"
)

print(
    f"Validated contracts             : "
    f"{N_VALID}"
)

print(
    f"Validated-population coverage   : "
    f"{N_VALID / N_AGENTS:.1%}"
)

print(
    f"Intellectual families           : "
    f"{represented_families}/{len(FAMILIES)}"
)

print()

for family, n in family_sizes.items():

    print(
        f"{family:<30} "
        f"{int(n)}/5"
    )


# ============================================================
# 4. BUILD AGENT × PRINCIPLE INCIDENCE MATRIX
#
# Each cell is:
#
#       1 = agent supports principle
#       0 = agent does not list principle as supported
#
# ============================================================

agent_principle_matrix = pd.DataFrame(
    0,
    index=contracts["agent_id"],
    columns=PRINCIPLES,
    dtype=int
)


for _, row in contracts.iterrows():

    for principle in row["supported_principles"]:

        agent_principle_matrix.loc[
            row["agent_id"],
            principle
        ] = 1


agent_metadata = (
    contracts[
        [
            "agent_id",
            "family",
            "variant",
            "confidence"
        ]
    ]
    .set_index("agent_id")
)


# ============================================================
# 5. BUILD LONG-FORM VOTE TABLE
# ============================================================

vote_rows = []


for _, row in contracts.iterrows():

    for principle in row["supported_principles"]:

        vote_rows.append(
            {
                "agent_id":
                    row["agent_id"],

                "family":
                    row["family"],

                "variant":
                    row["variant"],

                "principle":
                    principle,

                "confidence":
                    row["confidence"],
            }
        )


votes = pd.DataFrame(
    vote_rows
)


if votes.empty:
    raise RuntimeError(
        "Validated agents returned no supported principles."
    )


# ============================================================
# 6. FAMILY × PRINCIPLE SUPPORT SHARE
#
# This is more informative than simply asking whether ANY
# member of a family supported a principle.
#
# Example:
#
#     control_theory:
#         4 of 5 agents support principle X
#         -> family support share = 0.80
#
# ============================================================

family_principle_share = pd.DataFrame(
    0.0,
    index=list(FAMILIES.keys()),
    columns=PRINCIPLES
)


for family in FAMILIES.keys():

    family_agents = contracts.loc[
        contracts["family"] == family,
        "agent_id"
    ].tolist()

    if len(family_agents) == 0:
        continue

    family_principle_share.loc[family] = (
        agent_principle_matrix
        .loc[family_agents]
        .mean(axis=0)
    )


# ============================================================
# 7. FAMILY × PRINCIPLE PRESENCE MATRIX
#
# Breadth measure:
#
#       1 = at least one validated agent in family supports it
#       0 = no validated agent in family supports it
#
# ============================================================

family_principle_presence = (
    family_principle_share > 0
).astype(int)


# ============================================================
# 8. COMPUTE THREE CONVERGENCE MEASURES
# ============================================================

summary_rows = []


for principle in PRINCIPLES:

    supporting_agents = int(
        agent_principle_matrix[
            principle
        ].sum()
    )

    agent_support = (
        supporting_agents
        /
        N_VALID
    )

    supporting_families = int(
        family_principle_presence[
            principle
        ].sum()
    )

    family_breadth = (
        supporting_families
        /
        len(FAMILIES)
    )

    # Every family gets exactly the same total weight.
    equal_family_support = float(
        family_principle_share[
            principle
        ].mean()
    )

    # Mean confidence only among agents explicitly supporting
    # the principle. This is descriptive, not a voting weight.
    supporter_ids = (
        agent_principle_matrix.index[
            agent_principle_matrix[
                principle
            ] == 1
        ]
    )

    if len(supporter_ids) > 0:

        mean_supporter_confidence = float(
            agent_metadata
            .loc[
                supporter_ids,
                "confidence"
            ]
            .mean()
        )

    else:

        mean_supporter_confidence = np.nan

    summary_rows.append(
        {
            "principle":
                principle,

            "supporting_agents":
                supporting_agents,

            "validated_agents":
                N_VALID,

            "agent_support":
                agent_support,

            "supporting_families":
                supporting_families,

            "total_families":
                len(FAMILIES),

            "family_breadth":
                family_breadth,

            "equal_family_support":
                equal_family_support,

            "mean_supporter_confidence":
                mean_supporter_confidence,
        }
    )


principle_summary = pd.DataFrame(
    summary_rows
).set_index(
    "principle"
)


principle_summary = principle_summary.sort_values(
    [
        "family_breadth",
        "equal_family_support",
        "agent_support"
    ],
    ascending=False
)


# ============================================================
# 9. DEFINE A TRANSPARENT CANDIDATE THEORETICAL CORE
#
# We distinguish:
#
# BROAD CORE
#   >= 70% of intellectual families represented
#
# STRONG CORE
#   >= 70% family breadth
#   AND
#   >= 50% equal-family support
#
# UNIVERSAL-BREADTH PRINCIPLE
#   at least one supporter in every family
#
# These are analytical labels, not hidden model judgments.
# ============================================================

BROAD_FAMILY_THRESHOLD = 0.70
STRONG_EQUAL_FAMILY_THRESHOLD = 0.50


broad_core = (
    principle_summary[
        principle_summary[
            "family_breadth"
        ]
        >= BROAD_FAMILY_THRESHOLD
    ]
    .index
    .tolist()
)


strong_core = (
    principle_summary[
        (
            principle_summary[
                "family_breadth"
            ]
            >= BROAD_FAMILY_THRESHOLD
        )
        &
        (
            principle_summary[
                "equal_family_support"
            ]
            >= STRONG_EQUAL_FAMILY_THRESHOLD
        )
    ]
    .index
    .tolist()
)


universal_breadth = (
    principle_summary[
        principle_summary[
            "family_breadth"
        ] == 1.0
    ]
    .index
    .tolist()
)


# ============================================================
# 10. MISSING-DATA ROBUSTNESS BOUNDS
#
# We commissioned 50 agents but validated N_VALID.
#
# For each principle:
#
# LOWER BOUND:
#     assume every failed observation would NOT support it.
#
# UPPER BOUND:
#     assume every failed observation WOULD support it.
#
# This prevents the four schema failures from being silently
# treated as either agreement or disagreement.
# ============================================================

N_MISSING = N_AGENTS - N_VALID


principle_summary[
    "commissioned_support_lower_bound"
] = (
    principle_summary[
        "supporting_agents"
    ]
    /
    N_AGENTS
)


principle_summary[
    "commissioned_support_upper_bound"
] = (
    (
        principle_summary[
            "supporting_agents"
        ]
        +
        N_MISSING
    )
    /
    N_AGENTS
)


# ============================================================
# 11. DISPLAY CONVERGENCE RESULTS
# ============================================================

print()
print("=" * 92)
print("CROSS-DISCIPLINARY CONVERGENCE")
print("=" * 92)

print()

print(
    "BROAD CORE "
    f"(family breadth ≥ {BROAD_FAMILY_THRESHOLD:.0%})"
)

print("-" * 92)

if broad_core:

    for principle in broad_core:
        print(f"  • {principle}")

else:
    print("  None")


print()

print(
    "STRONG CORE "
    f"(family breadth ≥ {BROAD_FAMILY_THRESHOLD:.0%} "
    f"and equal-family support ≥ "
    f"{STRONG_EQUAL_FAMILY_THRESHOLD:.0%})"
)

print("-" * 92)

if strong_core:

    for principle in strong_core:
        print(f"  • {principle}")

else:
    print("  None")


print()

print(
    "UNIVERSAL-BREADTH PRINCIPLES "
    "(represented in all 10 families)"
)

print("-" * 92)

if universal_breadth:

    for principle in universal_breadth:
        print(f"  • {principle}")

else:
    print("  None")


# ============================================================
# 12. PRINCIPLE SUMMARY TABLE
# ============================================================

display_columns = [
    "supporting_agents",
    "agent_support",
    "supporting_families",
    "family_breadth",
    "equal_family_support",
    "mean_supporter_confidence",
    "commissioned_support_lower_bound",
    "commissioned_support_upper_bound",
]


print()
print("=" * 92)
print("PRINCIPLE-LEVEL CONVERGENCE TABLE")
print("=" * 92)


display(
    principle_summary[
        display_columns
    ].style.format(
        {
            "agent_support":
                "{:.1%}",

            "family_breadth":
                "{:.1%}",

            "equal_family_support":
                "{:.1%}",

            "mean_supporter_confidence":
                "{:.3f}",

            "commissioned_support_lower_bound":
                "{:.1%}",

            "commissioned_support_upper_bound":
                "{:.1%}",
        }
    )
)


# ============================================================
# 13. FAMILY × PRINCIPLE DEPTH MATRIX
# ============================================================

print()
print("=" * 92)
print("FAMILY × PRINCIPLE SUPPORT SHARE")
print("=" * 92)

print(
    "Each cell is the fraction of validated agents within that "
    "family supporting the principle."
)

display(
    family_principle_share.style.format(
        "{:.0%}"
    )
)


# ============================================================
# 14. FAMILY × PRINCIPLE BREADTH MATRIX
# ============================================================

print()
print("=" * 92)
print("FAMILY × PRINCIPLE PRESENCE")
print("=" * 92)

print(
    "1 = at least one validated agent in the family supports "
    "the principle."
)

display(
    family_principle_presence
)


# ============================================================
# 15. VISUALIZATION — FAMILY BREADTH
# ============================================================

plot_data = (
    principle_summary[
        "family_breadth"
    ]
    .sort_values()
)


plt.figure(
    figsize=(10, 6)
)

plot_data.plot(
    kind="barh"
)

plt.xlabel(
    "Share of independent intellectual families"
)

plt.ylabel(
    "Candidate TIRM principle"
)

plt.title(
    "TIRM — Cross-Disciplinary Breadth of Candidate Principles"
)

plt.xlim(
    0,
    1.05
)

plt.tight_layout()

plt.show()


# ============================================================
# 16. VISUALIZATION — EQUAL-FAMILY SUPPORT
# ============================================================

plot_data_equal = (
    principle_summary[
        "equal_family_support"
    ]
    .sort_values()
)


plt.figure(
    figsize=(10, 6)
)

plot_data_equal.plot(
    kind="barh"
)

plt.xlabel(
    "Equal-family weighted support"
)

plt.ylabel(
    "Candidate TIRM principle"
)

plt.title(
    "TIRM — Depth of Cross-Disciplinary Convergence"
)

plt.xlim(
    0,
    1.05
)

plt.tight_layout()

plt.show()


# ============================================================
# 17. VISUALIZATION — FAMILY × PRINCIPLE HEATMAP
#
# Matplotlib only; no seaborn dependency.
# ============================================================

fig, ax = plt.subplots(
    figsize=(13, 7)
)

heat = ax.imshow(
    family_principle_share.values,
    aspect="auto",
    vmin=0,
    vmax=1
)

ax.set_xticks(
    np.arange(
        len(
            family_principle_share.columns
        )
    )
)

ax.set_xticklabels(
    family_principle_share.columns,
    rotation=45,
    ha="right"
)

ax.set_yticks(
    np.arange(
        len(
            family_principle_share.index
        )
    )
)

ax.set_yticklabels(
    family_principle_share.index
)

ax.set_title(
    "TIRM — Family × Principle Convergence Matrix"
)

ax.set_xlabel(
    "Candidate principle"
)

ax.set_ylabel(
    "Intellectual family"
)

cbar = fig.colorbar(
    heat,
    ax=ax
)

cbar.set_label(
    "Within-family support share"
)

plt.tight_layout()

plt.show()


# ============================================================
# 18. PRESERVE FREE-TEXT RESEARCH MATERIAL FOR LATER SYNTHESIS
#
# We deliberately DO NOT force free-text axioms, mathematical
# objects, propositions, or analogy breaks into artificial
# categories in this cell.
#
# They are preserved intact for later synthesis.
# ============================================================

axiom_corpus = []

mathematical_object_corpus = []

proposition_corpus = []

analogy_break_corpus = []

mapping_corpus = []

novel_contribution_corpus = []


for _, row in contracts.iterrows():

    for text in row["axioms"]:

        axiom_corpus.append(
            {
                "agent_id":
                    row["agent_id"],

                "family":
                    row["family"],

                "variant":
                    row["variant"],

                "text":
                    text,
            }
        )

    for text in row["math"]:

        mathematical_object_corpus.append(
            {
                "agent_id":
                    row["agent_id"],

                "family":
                    row["family"],

                "variant":
                    row["variant"],

                "text":
                    text,
            }
        )

    for proposition in row["propositions"]:

        proposition_corpus.append(
            {
                "agent_id":
                    row["agent_id"],

                "family":
                    row["family"],

                "variant":
                    row["variant"],

                "claim":
                    proposition.get(
                        "claim"
                    ),

                "observable":
                    proposition.get(
                        "observable"
                    ),

                "rejection_condition":
                    proposition.get(
                        "rejection_condition"
                    ),
            }
        )

    for text in row["analogy_breaks"]:

        analogy_break_corpus.append(
            {
                "agent_id":
                    row["agent_id"],

                "family":
                    row["family"],

                "variant":
                    row["variant"],

                "text":
                    text,
            }
        )

    mapping_corpus.append(
        {
            "agent_id":
                row["agent_id"],

            "family":
                row["family"],

            "variant":
                row["variant"],

            **row["mapping"],
        }
    )

    novel_contribution_corpus.append(
        {
            "agent_id":
                row["agent_id"],

            "family":
                row["family"],

            "variant":
                row["variant"],

            "text":
                row["novel_contribution"],
        }
    )


axioms_df = pd.DataFrame(
    axiom_corpus
)

mathematical_objects_df = pd.DataFrame(
    mathematical_object_corpus
)

propositions_df = pd.DataFrame(
    proposition_corpus
)

analogy_breaks_df = pd.DataFrame(
    analogy_break_corpus
)

mappings_df = pd.DataFrame(
    mapping_corpus
)

novel_contributions_df = pd.DataFrame(
    novel_contribution_corpus
)


# ============================================================
# 19. CELL 5 MACHINE-READABLE OUTPUT
# ============================================================

convergence_status = {

    "validated_agents":
        N_VALID,

    "commissioned_agents":
        N_AGENTS,

    "missing_observations":
        N_MISSING,

    "represented_families":
        represented_families,

    "total_families":
        len(FAMILIES),

    "broad_family_threshold":
        BROAD_FAMILY_THRESHOLD,

    "strong_equal_family_threshold":
        STRONG_EQUAL_FAMILY_THRESHOLD,

    "broad_core":
        broad_core,

    "strong_core":
        strong_core,

    "universal_breadth":
        universal_breadth,

    "api_calls_in_cell_5":
        0,
}


# ============================================================
# 20. FINAL CELL 5 REPORT
# ============================================================

print()
print("=" * 92)
print("CELL 5 COMPLETE")
print("=" * 92)

print()

print(
    f"Validated research contracts analyzed : "
    f"{N_VALID}"
)

print(
    f"Missing computational observations    : "
    f"{N_MISSING}"
)

print(
    f"Intellectual families represented     : "
    f"{represented_families}/{len(FAMILIES)}"
)

print(
    f"Candidate principles evaluated        : "
    f"{len(PRINCIPLES)}"
)

print(
    f"Broad-core principles                 : "
    f"{len(broad_core)}"
)

print(
    f"Strong-core principles                : "
    f"{len(strong_core)}"
)

print(
    f"Universal-breadth principles          : "
    f"{len(universal_breadth)}"
)

print()

print(
    "API calls made by Cell 5              : 0"
)

print()

print(
    "✓ Agent-level convergence measured."
)

print(
    "✓ Family-level breadth measured."
)

print(
    "✓ Equal-family convergence measured."
)

print(
    "✓ Missing observations bounded rather than imputed."
)

print(
    "✓ Free-text theoretical material preserved for synthesis."
)

print()

print(
    "CELL 5 READY FOR THEORETICAL SYNTHESIS."
)

print("=" * 92)
```

## CELL 6 — Derive candidate axioms of Teleportation-Inspired Risk Management

The sixth unit converts consensus into explicit theoretical statements. A deterministic set of candidate axioms is combined with the swarm's strongest propositions. The notebook defines a generalized risk state and a transformation operator representing a risk-management intervention.

The core distinction is between intended transformation and realized reconstruction. Their difference is the residual or teleportation error. Fidelity is defined as a decreasing function of that error. Capacity, information quality, latency, liquidity, and counterparty reliability enter as determinants of reconstruction quality.

The cell prints a compact candidate theory. It is intentionally provisional: axioms are hypotheses about useful structure, not laws of nature.

```python
# @title
# ============================================================
# CELL 6 — DERIVE THE CANDIDATE AXIOMATIC SYSTEM OF
# TELEPORTATION-INSPIRED RISK MANAGEMENT (TIRM)
#
# PURPOSE
# -------
# Convert the cross-disciplinary convergence evidence produced
# in Cell 5 into a transparent candidate axiomatic system.
#
# IMPORTANT:
#   • ZERO API CALLS
#   • Uses ONLY validated Cell 4 contracts and Cell 5 analysis
#   • Does NOT treat swarm agreement as mathematical proof
#   • Separates:
#         1. canonical theoretical formulation,
#         2. empirical swarm support,
#         3. strength of cross-disciplinary convergence,
#         4. formal mathematical interpretation.
#
# Cell 5 asks:
#     "Which principles survive across disciplines?"
#
# Cell 6 asks:
#     "What coherent axiomatic theory can be constructed
#      from those surviving principles?"
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. VERIFY UPSTREAM STATE
# ============================================================

required_objects = [
    "SWARM_READY",
    "RESEARCH_GATE",
    "principle_summary",
    "broad_core",
    "strong_core",
    "universal_breadth",
    "family_principle_share",
    "contracts",
    "axioms_df",
    "analogy_breaks_df",
    "propositions_df",
    "PRINCIPLES",
    "FAMILIES",
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:
    raise RuntimeError(
        "Cell 6 cannot proceed because the following "
        f"upstream objects are missing: {missing_objects}"
    )

if not SWARM_READY:
    raise RuntimeError(
        "The swarm has not passed the Cell 4 scientific gate."
    )

if RESEARCH_GATE != "PASS":
    raise RuntimeError(
        f"Expected RESEARCH_GATE='PASS', found {RESEARCH_GATE!r}."
    )


print("=" * 96)
print("CELL 6 — CANDIDATE AXIOMATIC SYSTEM")
print("TELEPORTATION-INSPIRED RISK MANAGEMENT (TIRM)")
print("=" * 96)

print()
print("✓ Cell 4 validated research population available")
print("✓ Cell 5 convergence analysis available")
print("✓ No additional model inference required")
print("✓ API calls in Cell 6: 0")


# ============================================================
# 2. CANONICAL TIRM AXIOMS
#
# These are the notebook's canonical theoretical formulations.
#
# Their inclusion in the candidate theory is NOT automatic.
# Cell 5 convergence evidence determines their empirical status
# within the research swarm.
# ============================================================

CANONICAL_AXIOMS = {

    "residual_risk_conservation":
        "A local reduction in one measured risk channel does not "
        "imply destruction of the underlying economic uncertainty; "
        "residual exposure must be located in retained, transformed, "
        "or redistributed channels.",

    "costly_reconstruction":
        "Reconstructing a target risk state requires scarce "
        "instruments, balance-sheet capacity, liquidity, collateral, "
        "information, or governance resources.",

    "prepositioned_capacity":
        "Risk transformation under stress depends on resources and "
        "contractual channels established before the stress event.",

    "measurement_incompleteness":
        "The true economic risk state is only partially observed "
        "through models, prices, positions, scenarios, accounting "
        "representations, and other measurement systems.",

    "communication_latency":
        "The fidelity of a risk intervention deteriorates when "
        "measurement, authorization, contractual communication, or "
        "execution is delayed relative to changes in the underlying "
        "risk state.",

    "network_redistribution":
        "Risk interventions redistribute contingent claims and "
        "dependencies through contractual, market, collateral, and "
        "balance-sheet networks.",

    "fidelity_loss":
        "Realized post-intervention exposures differ from intended "
        "exposures because of basis risk, liquidity effects, model "
        "error, execution error, counterparty risk, timing, and other "
        "implementation frictions.",

    "no_free_cloning":
        "A contingent risk state cannot be replicated with arbitrary "
        "fidelity at zero economic cost and without assumptions about "
        "instruments, information, liquidity, counterparties, and "
        "implementation capacity.",

    "local_systemic_divergence":
        "An intervention can improve a local risk metric while "
        "leaving system-wide risk unchanged, redistributing it, "
        "or increasing it elsewhere in the financial network.",

    "conditional_correction":
        "Effective reconstruction of a target risk state requires "
        "state-contingent corrective actions following measurement, "
        "information transmission, and changes in market conditions."
}


# Verify exact correspondence with the controlled vocabulary.

missing_canonical = (
    set(PRINCIPLES)
    -
    set(CANONICAL_AXIOMS)
)

extra_canonical = (
    set(CANONICAL_AXIOMS)
    -
    set(PRINCIPLES)
)

if missing_canonical or extra_canonical:

    raise RuntimeError(
        "Canonical axiom vocabulary does not match PRINCIPLES.\n"
        f"Missing: {sorted(missing_canonical)}\n"
        f"Extra: {sorted(extra_canonical)}"
    )


# ============================================================
# 3. FORMAL TIRM STATE ARCHITECTURE
#
# TIRM treats financial risk management as a sequence of:
#
#     state
#       ↓
#     measurement
#       ↓
#     communication
#       ↓
#     conditional intervention
#       ↓
#     reconstructed state
#       ↓
#     residual error
#
# This is an architectural analogy to quantum teleportation,
# NOT a claim that financial systems obey quantum mechanics.
# ============================================================

TIRM_FORMALISM = {

    "risk_state":
        "R_t = (x_t, Sigma_t, l_t, c_t, m_t, N_t)",

    "intended_transformation":
        "R*_t = T(R_t | Q_t)",

    "measurement":
        "y_t = M_t(R_t) + eta_t",

    "message":
        "z_t = G_t(y_t, I_t)",

    "realized_reconstruction":
        "R'_t = C_t(y_t, z_t, Q_t, S_t)",

    "residual_error":
        "epsilon_t = R'_t - R*_t",

    "weighted_error":
        "E_t = ||W_t epsilon_t||",

    "fidelity":
        "F_t = exp(-E_t),    0 < F_t <= 1",
}


# Interpretation of symbols.

TIRM_SYMBOLS = {

    "R_t":
        "economic risk state at time t",

    "x_t":
        "direct financial exposures",

    "Sigma_t":
        "covariance, dependence, or scenario structure",

    "l_t":
        "liquidity and funding state",

    "c_t":
        "counterparty and collateral state",

    "m_t":
        "model and measurement uncertainty",

    "N_t":
        "financial-network configuration",

    "Q_t":
        "pre-positioned transformation resources: instruments, "
        "capital, collateral, liquidity, contracts, and operational "
        "capacity",

    "M_t":
        "measurement operator",

    "eta_t":
        "measurement error or omitted-state component",

    "y_t":
        "observed representation of the underlying risk state",

    "I_t":
        "institutional information available at decision time",

    "G_t":
        "communication and governance mechanism",

    "z_t":
        "decision-relevant message or instruction",

    "S_t":
        "market and institutional state at execution",

    "C_t":
        "conditional reconstruction/intervention operator",

    "R*_t":
        "intended post-intervention risk state",

    "R'_t":
        "realized post-intervention risk state",

    "epsilon_t":
        "reconstruction error",

    "W_t":
        "economic weighting matrix for residual errors",

    "F_t":
        "reconstruction fidelity",
}


# ============================================================
# 4. CLASSIFY EACH PRINCIPLE USING CELL 5 EVIDENCE
#
# Four evidence classes:
#
# UNIVERSAL_BREADTH
#     appears in all intellectual families
#
# STRONG_CORE
#     >= 70% family breadth AND
#     >= 50% equal-family support
#
# BROAD_CORE
#     >= 70% family breadth
#
# EMERGING
#     below broad-core threshold
#
# The hierarchy is mutually exclusive.
# ============================================================

def convergence_class(principle):

    if principle in universal_breadth:
        return "UNIVERSAL_BREADTH"

    if principle in strong_core:
        return "STRONG_CORE"

    if principle in broad_core:
        return "BROAD_CORE"

    return "EMERGING"


# ============================================================
# 5. CONSTRUCT AXIOM EVIDENCE TABLE
# ============================================================

axiom_rows = []


for principle in PRINCIPLES:

    evidence = principle_summary.loc[
        principle
    ]

    axiom_rows.append(
        {
            "principle":
                principle,

            "canonical_axiom":
                CANONICAL_AXIOMS[
                    principle
                ],

            "convergence_class":
                convergence_class(
                    principle
                ),

            "supporting_agents":
                int(
                    evidence[
                        "supporting_agents"
                    ]
                ),

            "agent_support":
                float(
                    evidence[
                        "agent_support"
                    ]
                ),

            "supporting_families":
                int(
                    evidence[
                        "supporting_families"
                    ]
                ),

            "family_breadth":
                float(
                    evidence[
                        "family_breadth"
                    ]
                ),

            "equal_family_support":
                float(
                    evidence[
                        "equal_family_support"
                    ]
                ),

            "mean_supporter_confidence":
                float(
                    evidence[
                        "mean_supporter_confidence"
                    ]
                )
                if pd.notna(
                    evidence[
                        "mean_supporter_confidence"
                    ]
                )
                else np.nan,

            "support_lower_bound":
                float(
                    evidence[
                        "commissioned_support_lower_bound"
                    ]
                ),

            "support_upper_bound":
                float(
                    evidence[
                        "commissioned_support_upper_bound"
                    ]
                ),
        }
    )


axiom_evidence = pd.DataFrame(
    axiom_rows
).set_index(
    "principle"
)


# Order by strength of cross-disciplinary evidence.

class_order = {
    "UNIVERSAL_BREADTH": 0,
    "STRONG_CORE": 1,
    "BROAD_CORE": 2,
    "EMERGING": 3,
}


axiom_evidence[
    "_class_order"
] = (
    axiom_evidence[
        "convergence_class"
    ]
    .map(
        class_order
    )
)


axiom_evidence = (
    axiom_evidence
    .sort_values(
        [
            "_class_order",
            "family_breadth",
            "equal_family_support",
            "agent_support",
        ],
        ascending=[
            True,
            False,
            False,
            False,
        ]
    )
    .drop(
        columns="_class_order"
    )
)


# ============================================================
# 6. DEFINE THE CANDIDATE AXIOMATIC SYSTEM
#
# PRIMARY AXIOMS:
#     strong-core OR universal-breadth principles
#
# SECONDARY PRINCIPLES:
#     broad interdisciplinary support but insufficient depth
#     for primary-axiom status
#
# EMERGING PRINCIPLES:
#     theoretically interesting but not sufficiently convergent
#     across the swarm to enter the present core.
# ============================================================

PRIMARY_AXIOMS = [
    principle
    for principle
    in axiom_evidence.index
    if axiom_evidence.loc[
        principle,
        "convergence_class"
    ]
    in {
        "UNIVERSAL_BREADTH",
        "STRONG_CORE",
    }
]


SECONDARY_PRINCIPLES = [
    principle
    for principle
    in axiom_evidence.index
    if axiom_evidence.loc[
        principle,
        "convergence_class"
    ]
    == "BROAD_CORE"
]


EMERGING_PRINCIPLES = [
    principle
    for principle
    in axiom_evidence.index
    if axiom_evidence.loc[
        principle,
        "convergence_class"
    ]
    == "EMERGING"
]


# ============================================================
# 7. DISPLAY THE FORMAL ARCHITECTURE
# ============================================================

print()
print("=" * 96)
print("I. FORMAL TIRM ARCHITECTURE")
print("=" * 96)

print()

print(
    "Risk state:"
)

print(
    "    R_t = (x_t, Sigma_t, l_t, c_t, m_t, N_t)"
)

print()

print(
    "Intended transformation:"
)

print(
    "    R*_t = T(R_t | Q_t)"
)

print()

print(
    "Imperfect measurement:"
)

print(
    "    y_t = M_t(R_t) + eta_t"
)

print()

print(
    "Communication / governance:"
)

print(
    "    z_t = G_t(y_t, I_t)"
)

print()

print(
    "Realized reconstruction:"
)

print(
    "    R'_t = C_t(y_t, z_t, Q_t, S_t)"
)

print()

print(
    "Residual reconstruction error:"
)

print(
    "    epsilon_t = R'_t - R*_t"
)

print()

print(
    "Weighted reconstruction error:"
)

print(
    "    E_t = ||W_t epsilon_t||"
)

print()

print(
    "Reconstruction fidelity:"
)

print(
    "    F_t = exp(-E_t),     0 < F_t <= 1"
)


# ============================================================
# 8. DISPLAY PRIMARY CANDIDATE AXIOMS
# ============================================================

print()
print("=" * 96)
print("II. PRIMARY CANDIDATE AXIOMS")
print("=" * 96)

print()

if PRIMARY_AXIOMS:

    for number, principle in enumerate(
        PRIMARY_AXIOMS,
        start=1
    ):

        evidence = axiom_evidence.loc[
            principle
        ]

        print(
            f"AXIOM {number}: "
            f"{principle.upper()}"
        )

        print(
            "-" * 96
        )

        print(
            CANONICAL_AXIOMS[
                principle
            ]
        )

        print()

        print(
            "Cross-disciplinary evidence:"
        )

        print(
            f"    Supporting agents     : "
            f"{int(evidence['supporting_agents'])}/"
            f"{len(contracts)} "
            f"({evidence['agent_support']:.1%})"
        )

        print(
            f"    Supporting families   : "
            f"{int(evidence['supporting_families'])}/"
            f"{len(FAMILIES)} "
            f"({evidence['family_breadth']:.1%})"
        )

        print(
            f"    Equal-family support  : "
            f"{evidence['equal_family_support']:.1%}"
        )

        print(
            f"    Evidence class        : "
            f"{evidence['convergence_class']}"
        )

        print()


else:

    print(
        "No principles satisfy the current primary-axiom "
        "threshold."
    )


# ============================================================
# 9. DISPLAY SECONDARY AND EMERGING PRINCIPLES
# ============================================================

print()
print("=" * 96)
print("III. SECONDARY THEORETICAL PRINCIPLES")
print("=" * 96)

print()

if SECONDARY_PRINCIPLES:

    for principle in SECONDARY_PRINCIPLES:

        evidence = axiom_evidence.loc[
            principle
        ]

        print(
            f"• {principle}"
        )

        print(
            f"  Family breadth: "
            f"{evidence['family_breadth']:.1%}"
            f" | Equal-family support: "
            f"{evidence['equal_family_support']:.1%}"
        )

        print(
            f"  {CANONICAL_AXIOMS[principle]}"
        )

        print()

else:

    print(
        "None."
    )


print()
print("=" * 96)
print("IV. EMERGING / NON-CORE PRINCIPLES")
print("=" * 96)

print()

if EMERGING_PRINCIPLES:

    for principle in EMERGING_PRINCIPLES:

        evidence = axiom_evidence.loc[
            principle
        ]

        print(
            f"• {principle}"
        )

        print(
            f"  Family breadth: "
            f"{evidence['family_breadth']:.1%}"
            f" | Equal-family support: "
            f"{evidence['equal_family_support']:.1%}"
        )

        print()

else:

    print(
        "None."
    )


# ============================================================
# 10. DISPLAY AXIOM EVIDENCE TABLE
# ============================================================

print()
print("=" * 96)
print("V. AXIOM EVIDENCE TABLE")
print("=" * 96)


display_columns = [
    "convergence_class",
    "supporting_agents",
    "agent_support",
    "supporting_families",
    "family_breadth",
    "equal_family_support",
    "mean_supporter_confidence",
    "support_lower_bound",
    "support_upper_bound",
]


display(
    axiom_evidence[
        display_columns
    ].style.format(
        {
            "agent_support":
                "{:.1%}",

            "family_breadth":
                "{:.1%}",

            "equal_family_support":
                "{:.1%}",

            "mean_supporter_confidence":
                "{:.3f}",

            "support_lower_bound":
                "{:.1%}",

            "support_upper_bound":
                "{:.1%}",
        }
    )
)


# ============================================================
# 11. FAMILY SUPPORT STRUCTURE OF PRIMARY AXIOMS
# ============================================================

if PRIMARY_AXIOMS:

    primary_family_matrix = (
        family_principle_share[
            PRIMARY_AXIOMS
        ]
        .copy()
    )

else:

    primary_family_matrix = pd.DataFrame(
        index=list(
            FAMILIES.keys()
        )
    )


print()
print("=" * 96)
print("VI. FAMILY SUPPORT STRUCTURE OF PRIMARY AXIOMS")
print("=" * 96)

if not primary_family_matrix.empty:

    display(
        primary_family_matrix.style.format(
            "{:.0%}"
        )
    )

else:

    print(
        "No primary axioms available for family-level display."
    )


# ============================================================
# 12. VISUALIZE AXIOMATIC SUPPORT
#
# We deliberately use equal-family support on the horizontal
# axis because each intellectual tradition should receive the
# same aggregate weight.
# ============================================================

if len(axiom_evidence) > 0:

    plot_data = (
        axiom_evidence[
            "equal_family_support"
        ]
        .sort_values()
    )

    plt.figure(
        figsize=(11, 7)
    )

    plot_data.plot(
        kind="barh"
    )

    plt.axvline(
        x=0.50,
        linestyle="--",
        linewidth=1
    )

    plt.xlabel(
        "Equal-family weighted support"
    )

    plt.ylabel(
        "Candidate TIRM principle"
    )

    plt.title(
        "TIRM — Cross-Disciplinary Support for Candidate Axioms"
    )

    plt.xlim(
        0,
        1.05
    )

    plt.tight_layout()

    plt.show()


# ============================================================
# 13. AXIOMATIC ARCHITECTURE
#
# This converts the principles into a conceptual causal chain.
#
# It is NOT an empirical causal estimate. It is the proposed
# theoretical architecture of TIRM.
# ============================================================

TIRM_ARCHITECTURE = {

    "A. STATE OBSERVATION": [
        "measurement_incompleteness",
    ],

    "B. PRE-POSITIONED TRANSFORMATION CAPACITY": [
        "prepositioned_capacity",
        "costly_reconstruction",
        "no_free_cloning",
    ],

    "C. INFORMATION AND CONTROL": [
        "communication_latency",
        "conditional_correction",
    ],

    "D. RECONSTRUCTION QUALITY": [
        "fidelity_loss",
    ],

    "E. RISK REDISTRIBUTION": [
        "residual_risk_conservation",
        "network_redistribution",
    ],

    "F. SYSTEMIC CONSEQUENCES": [
        "local_systemic_divergence",
    ],
}


architecture_rows = []


for stage, principles in TIRM_ARCHITECTURE.items():

    for principle in principles:

        architecture_rows.append(
            {
                "stage":
                    stage,

                "principle":
                    principle,

                "candidate_status":
                    (
                        "PRIMARY AXIOM"
                        if principle
                        in PRIMARY_AXIOMS

                        else
                        "SECONDARY PRINCIPLE"
                        if principle
                        in SECONDARY_PRINCIPLES

                        else
                        "EMERGING PRINCIPLE"
                    ),

                "family_breadth":
                    axiom_evidence.loc[
                        principle,
                        "family_breadth"
                    ],

                "equal_family_support":
                    axiom_evidence.loc[
                        principle,
                        "equal_family_support"
                    ],
            }
        )


axiomatic_architecture = pd.DataFrame(
    architecture_rows
)


print()
print("=" * 96)
print("VII. TIRM AXIOMATIC ARCHITECTURE")
print("=" * 96)

print()

for stage, principles in TIRM_ARCHITECTURE.items():

    print(
        stage
    )

    print(
        "-" * 96
    )

    for principle in principles:

        status = (
            "PRIMARY"
            if principle
            in PRIMARY_AXIOMS

            else
            "SECONDARY"
            if principle
            in SECONDARY_PRINCIPLES

            else
            "EMERGING"
        )

        print(
            f"    {principle:<32} "
            f"[{status}]"
        )

    print()


# ============================================================
# 14. THE CENTRAL TIRM PROPOSITION
#
# This is a synthesis statement of the architecture, not an
# additional swarm vote.
# ============================================================

TIRM_CENTRAL_PROPOSITION = (
    "Financial risk management should be modeled not as the "
    "deletion of a scalar quantity of risk, but as the "
    "measurement, transformation, communication, redistribution, "
    "and conditional reconstruction of a multidimensional economic "
    "risk state across instruments, institutions, and networks. "
    "The quality of that reconstruction depends on pre-positioned "
    "resources, information, timing, market conditions, and "
    "implementation capacity; therefore local risk reduction does "
    "not by itself establish system-wide risk reduction."
)


print()
print("=" * 96)
print("VIII. CENTRAL TIRM PROPOSITION")
print("=" * 96)

print()

print(
    TIRM_CENTRAL_PROPOSITION
)


# ============================================================
# 15. EPISTEMIC BOUNDARY
#
# Essential to the theory:
# the analogy must not be allowed to become a claim of literal
# quantum behavior in financial systems.
# ============================================================

TIRM_EPISTEMIC_BOUNDARY = [
    "Financial risk is not literally a quantum state.",
    "Financial dependence is not quantum entanglement.",
    "Risk is not physically teleported.",
    "The no-free-cloning principle is an economic analogy, not "
    "an invocation of the quantum no-cloning theorem as a law "
    "of finance.",
    "Reconstruction fidelity is a financial modeling construct, "
    "not quantum-state fidelity.",
    "The value of the analogy depends entirely on whether the "
    "resulting financial propositions are coherent, measurable, "
    "and falsifiable.",
]


print()
print("=" * 96)
print("IX. EPISTEMIC BOUNDARY")
print("=" * 96)

print()

for item in TIRM_EPISTEMIC_BOUNDARY:

    print(
        f"• {item}"
    )


# ============================================================
# 16. MACHINE-READABLE THEORY OBJECT
# ============================================================

TIRM_THEORY = {

    "name":
        "Teleportation-Inspired Risk Management",

    "abbreviation":
        "TIRM",

    "risk_state":
        TIRM_FORMALISM[
            "risk_state"
        ],

    "formalism":
        TIRM_FORMALISM,

    "symbols":
        TIRM_SYMBOLS,

    "primary_axioms":
        {
            principle:
                CANONICAL_AXIOMS[
                    principle
                ]

            for principle
            in PRIMARY_AXIOMS
        },

    "secondary_principles":
        {
            principle:
                CANONICAL_AXIOMS[
                    principle
                ]

            for principle
            in SECONDARY_PRINCIPLES
        },

    "emerging_principles":
        {
            principle:
                CANONICAL_AXIOMS[
                    principle
                ]

            for principle
            in EMERGING_PRINCIPLES
        },

    "architecture":
        TIRM_ARCHITECTURE,

    "central_proposition":
        TIRM_CENTRAL_PROPOSITION,

    "epistemic_boundary":
        TIRM_EPISTEMIC_BOUNDARY,

    "validated_agents":
        len(contracts),

    "represented_families":
        contracts[
            "family"
        ].nunique(),

    "api_calls_in_cell_6":
        0,
}


# ============================================================
# 17. FINAL REPORT
# ============================================================

print()
print("=" * 96)
print("CELL 6 COMPLETE — CANDIDATE TIRM AXIOMATIC SYSTEM")
print("=" * 96)

print()

print(
    f"Validated contracts used        : "
    f"{len(contracts)}"
)

print(
    f"Intellectual families used      : "
    f"{contracts['family'].nunique()}"
)

print(
    f"Primary candidate axioms        : "
    f"{len(PRIMARY_AXIOMS)}"
)

print(
    f"Secondary theoretical principles: "
    f"{len(SECONDARY_PRINCIPLES)}"
)

print(
    f"Emerging principles             : "
    f"{len(EMERGING_PRINCIPLES)}"
)

print(
    f"Total controlled principles     : "
    f"{len(PRINCIPLES)}"
)

print()

print(
    "API calls made by Cell 6       : 0"
)

print()

print(
    "✓ Cross-disciplinary evidence attached to every principle."
)

print(
    "✓ Primary axioms distinguished from secondary principles."
)

print(
    "✓ Formal state-transition architecture defined."
)

print(
    "✓ Reconstruction error and fidelity explicitly defined."
)

print(
    "✓ Local and systemic risk distinguished."
)

print(
    "✓ Quantum analogy explicitly bounded."
)

print(
    "✓ Machine-readable TIRM theory object created."
)

print()

print(
    "TIRM CANDIDATE AXIOMATIC SYSTEM READY FOR "
    "MATHEMATICAL AND EMPIRICAL DEVELOPMENT."
)

print("=" * 96)
```

### From Quantum Teleportation to a Richer Theory of Systemic Risk

The purpose of Teleportation-Inspired Risk Management is not to claim that financial markets behave according to quantum mechanics. Nor is the central idea simply one of information transmission. The deeper inspiration comes from the architecture of quantum teleportation itself: **entanglement, observation, intervention, dependence between separated elements, transformation of the original state, and the emergence of a new state elsewhere in the system**.

These ideas provide a vocabulary for constructing richer models of financial risk.

Traditional risk management frequently starts with an institution, identifies an exposure, measures that exposure, and asks how it can be reduced. A bank has an interest-rate risk, a currency exposure, a credit exposure or an equity position; it measures the risk and then hedges, insures, collateralizes, sells or transfers it.

The apparent conclusion is that risk has been reduced.

TIRM argues that this is only the beginning of the analysis.

A financial intervention changes not only the position of the institution taking the action but potentially the state of other participants and, through them, the configuration of the financial system. The important question therefore becomes not simply:

**How much risk did Alice eliminate?**

but:

**What happened to the system when Alice acted?**

This is where the inspiration from teleportation becomes useful.

### Alice and Bob are not simply exchanging information

Suppose Alice is a financial institution holding a substantial exposure. Bob is another institution connected to Alice through a derivative, guarantee, insurance contract, collateral arrangement, funding relationship or some other contingent financial claim.

Alice observes her condition and decides to act. She may hedge a position, buy protection, demand collateral, sell assets, purchase insurance or restructure her balance sheet.

Bob is not merely the recipient of a message that he subsequently “decrypts.” That would reduce the analogy to information theory, and that is not the idea.

**Bob is part of the interconnected state that Alice's action changes.**

If Alice buys protection from Bob, Bob acquires a contingent exposure. If Alice demands additional collateral, Bob may need to obtain liquidity. If Alice sells assets aggressively, prices may fall and affect Bob's balance sheet even if Bob has no direct contract with Alice. If Alice hedges through a central counterparty, exposures can migrate through clearing members, collateral networks and liquidity providers.

The action taken by Alice therefore changes the conditions confronting Bob.

And Bob's reaction may subsequently affect Charlie, Diana and the rest of the network.

This is why **entanglement** is a useful source of inspiration. We are not claiming quantum entanglement in a physical sense. The financial analogue is the existence of economic relationships through which the states of institutions cannot always be understood independently. Contracts, common assets, collateral, leverage, funding markets, guarantees and balance-sheet relationships create forms of interdependence.

Separation does not necessarily imply independence.

### Observation also matters

A second important inspiration comes from the role of observation.

In a simple traditional representation, measurement is treated as something relatively passive: the institution measures risk and then decides what to do.

In a systemic environment, observation can itself influence subsequent behavior.

A margin model observes deterioration and generates a margin call. A risk limit is breached and forces a trading desk to reduce its position. A credit downgrade changes collateral requirements. A stress test changes management behavior. A regulatory capital calculation can induce an institution to alter its portfolio. A market price used to measure risk can trigger additional sales that subsequently change that same market price.

Thus the distinction between **observing the state** and **changing the state** becomes less clean.

Again, this is not because financial measurement literally produces quantum measurement effects. The inspiration is conceptual: **the mechanisms through which a system is observed can become part of the mechanisms through which that system evolves.**

That is an important ingredient for richer models of systemic risk.

### The canonical elements now acquire a concrete meaning

The canonical propositions generated by the swarm can therefore be understood as different dimensions of this richer description of financial intervention.

**Measurement incompleteness** tells us that Alice never observes the complete economic state. Prices, positions, sensitivities, stress tests and models are representations of the system rather than the system itself.

**Pre-positioned capacity** tells us that Alice's ability to change her state depends on relationships and resources that frequently have to exist beforehand: counterparties, collateral agreements, credit lines, capital, liquidity, clearing arrangements and legal contracts.

**Costly reconstruction** tells us that reaching a desired new risk position consumes those resources. Risk transformation is not frictionless.

**Fidelity loss** tells us that the state Alice actually achieves will not necessarily be the state she intended. Basis risk, market movements, liquidity constraints, execution delays, model errors and counterparty behavior create a difference between intention and realization.

**Residual-risk conservation** directs our attention toward what remains after Alice appears to have reduced her risk. The economic uncertainty may have changed form rather than disappeared.

**Network redistribution** tells us to follow that transformation through the system. Some exposure may now reside with Bob; some may appear as collateral demand; some may migrate to a clearing house; some may become liquidity risk; and some may be distributed indirectly through prices and common asset holdings.

**Communication latency** reminds us that the system continues evolving while institutions measure, decide, authorize and execute their responses.

**Conditional correction** tells us that risk management is not necessarily a single transaction. As conditions change, further interventions may be required to maintain the desired state.

**No-free-cloning**, interpreted strictly as an economic analogy rather than a quantum law, tells us that reproducing a desired contingent payoff or risk configuration with high accuracy requires instruments, capital, liquidity, information and assumptions. Replication is neither perfect nor free.

Finally, **local-systemic divergence** captures perhaps the most important result: Alice can become safer while the system becomes no safer—and may under some circumstances become more fragile.

### From risk reduction to state transformation

The deeper insight is therefore that financial risk management should not be modeled simply as the subtraction of risk from an institution.

It should be modeled as a **transformation of an interconnected system**.

A hedge changes exposures.  
A collateral call changes liquidity needs.  
An insurance contract changes the distribution of losses.  
A guarantee creates contingent obligations.  
A securitization changes who bears particular risks.  
An asset sale changes prices and therefore the measured state of other balance sheets.  
A margin call can induce another sale, which changes prices again, which generates further margin calls.

The system evolves because its components are interconnected and because actions taken by one component alter the conditions faced by others.

The question therefore changes fundamentally.

Instead of asking only:

**“How much risk did the institution eliminate?”**

TIRM asks:

**“What new state of the financial system was created by the intervention?”**

That immediately generates a richer sequence of questions.

Where did the original exposure go? What form did it take? Who now bears it? Which institutions became more dependent on one another? What new collateral or liquidity requirements appeared? What new contingent claims were created? How accurately did the intervention produce the intended result? What happened to institutions indirectly connected to the transaction? And what happens when those institutions react?

### The real inspiration from teleportation

This is ultimately why teleportation is useful as an intellectual starting point.

The objective is **not to import quantum physics into finance** and it is not merely to describe financial markets as information-processing systems.

The objective is to use the conceptual richness of teleportation—**entanglement, state dependence, observation, intervention, transformation, conditional effects and reconstruction**—to encourage financial models that are richer than the conventional sequence of measuring an exposure and applying a hedge.

In our financial interpretation, there is no simple Bob waiting at the other end to receive and decode Alice's message.

**Bob is part of the system whose state may change because Alice acted.**

And once Bob's state changes, Bob may act as well.

That action affects others.

The result is a sequence of interconnected transformations:

**Alice acts → Bob is affected → Bob reacts → others are affected → the network changes → the risk state of the system changes again.**

This is the systemic interpretation that the canonical elements and theoretical propositions are beginning to reveal.

The fundamental unit of analysis is therefore no longer the isolated exposure or even the isolated institution. It is the **evolving configuration of interconnected risk states**.

A successful hedge for Alice is consequently not, by itself, evidence of successful risk management for the system.

The relevant question is what Alice's intervention has done to Alice, to Bob, to the relationships connecting them, and ultimately to the stability of the network in which both operate.

That is the central theoretical insight of TIRM:

> **Risk management is not merely the reduction of risk at its point of origin. It is an intervention in an interconnected system that transforms the states, dependencies and possible future actions of its participants. The quality of risk management must therefore be judged by the state of the system that the intervention creates, not solely by the risk metric that disappears from Alice's balance sheet.**

## CELL 7 — Numerical experiment: local risk reduction versus systemic redistribution

A theory of risk management should illuminate something that conventional scalar metrics can obscure. This experiment simulates institutions that hedge a market exposure through counterparties. The hedge reduces the initiating institution's market variance, but creates contingent counterparty and liquidity dependence.

The simulation varies hedge ratio, basis error, counterparty reliability, liquidity, and latency. It then compares local market-risk reduction with a broader generalized risk functional. This makes visible the distinction between successful local reconstruction and genuine system-wide risk reduction.

The experiment also computes a fidelity measure: how closely does the realized post-intervention state match the intended target? High hedge ratios need not imply high fidelity when implementation channels are fragile.

```python
# @title
# CELL 7 — Numerical experiment: local risk reduction versus systemic redistribution
sim=[]
for _ in range(5000):
    hedge_ratio=rng.uniform(0,1.2)
    basis=rng.uniform(0,0.20)
    liquidity=rng.uniform(0.35,1.0)
    latency=rng.uniform(0,0.50)
    cp=rng.uniform(0.80,1.0)

    intended,realized,residual,fidelity,liq_b,cp_b = risk_reconstruction(
        x0,h,hedge_ratio,basis,liquidity,latency,cp
    )
    before=generalized_risk(x0,Sigma)
    after=generalized_risk(realized,Sigma,liq_b,cp_b,np.linalg.norm(residual))

    sim.append({
      "hedge_ratio":hedge_ratio,"basis_error":basis,"liquidity":liquidity,
      "latency":latency,"cp_reliability":cp,"fidelity":fidelity,
      "market_before":before["market"],"market_after":after["market"],
      "generalized_before":before["generalized"],"generalized_after":after["generalized"]
    })

sim=pd.DataFrame(sim)
sim["local_improvement"]=sim.market_before-sim.market_after
sim["system_improvement"]=sim.generalized_before-sim.generalized_after
sim["divergence"]=(sim.local_improvement>0)&(sim.system_improvement<0)

print("Fraction of simulations with local market-risk improvement but generalized-risk deterioration:",
      round(sim.divergence.mean(),4))
display(sim.describe())

plt.figure(figsize=(7,5))
plt.scatter(sim.local_improvement,sim.system_improvement,s=8,alpha=.25)
plt.axhline(0,linewidth=1); plt.axvline(0,linewidth=1)
plt.xlabel("Local market-risk improvement")
plt.ylabel("Generalized system-risk improvement")
plt.title("Risk reduction versus risk redistribution")
plt.tight_layout()
plt.show()

plt.figure(figsize=(7,5))
plt.scatter(sim.liquidity,sim.fidelity,s=8,alpha=.25)
plt.xlabel("Liquidity capacity")
plt.ylabel("Reconstruction fidelity")
plt.title("Implementation capacity and fidelity")
plt.tight_layout()
plt.show()
```

# A Meeting Between Finance and Fifty Scientists  
## What Quantum Teleportation Can Teach Us About How to Think About Risk

Imagine a financial professional walking into a room with fifty scientists.

The financial professional has spent a career thinking about risk. He knows the language of markets, portfolios, hedging, derivatives, collateral, liquidity, counterparties and regulation. He knows how institutions measure exposures and how they try to reduce them. He understands that no serious financial institution can operate without models, limits, stress tests and mechanisms for transferring or mitigating risk.

But he has not come to this meeting looking for another model.

He is not asking the scientists for a better equation for Value at Risk. He is not asking for another volatility estimator, another optimization algorithm or another technique for pricing derivatives.

His question is more fundamental.

He wants to know whether there might be another **way of thinking about what happens when we manage risk**.

And then he makes what initially sounds like a rather strange suggestion:

“Let us think about quantum teleportation.”

At first, the connection seems almost absurd.

Quantum teleportation belongs to quantum physics. Financial risk management belongs to economics, markets and institutions. One deals with quantum states, measurement and entanglement; the other with balance sheets, derivatives, collateral, funding, prices and human organizations.

There is no reason to believe that a bank is a quantum system. There is no reason to describe a derivative contract as quantum entanglement. There is no reason to suppose that financial risk is literally teleported from one institution to another.

Indeed, making any of those claims would weaken the entire exercise.

But that is not the question being asked.

The question is whether an unfamiliar scientific framework can help us see something that our familiar financial framework has encouraged us to overlook.

That distinction is the beginning of the experiment.

---

## The Problem With a Familiar Story

Much of risk management can be described through a remarkably intuitive sequence.

An institution has an exposure.

It observes and measures that exposure.

It decides that the exposure is undesirable.

It intervenes.

It hedges, insures, sells, collateralizes, diversifies or restructures.

It then measures the resulting exposure.

If the number is smaller, the intervention appears to have succeeded.

There is nothing wrong with this sequence. It is extraordinarily useful. Modern finance could hardly function without it.

But perhaps it encourages us to stop too soon.

Suppose Alice is a bank with a large interest-rate exposure. Alice measures the exposure and decides to hedge it. After the hedge, her measured interest-rate risk falls substantially.

The conventional story can end there.

Alice had risk. Alice acted. Alice now has less risk.

Success.

The fifty scientists are asked to look at this familiar sequence from outside finance.

And the intellectual provocation supplied to them is quantum teleportation.

Not its equations.

Not its physics.

Its **architecture**.

Its sequence of events.

Its insistence that the state of one element cannot necessarily be understood without considering relationships already established with other elements.

Its treatment of observation as part of what happens rather than as something entirely outside the system.

Its combination of an initial state, pre-existing relationships, observation, intervention, conditional consequences and a resulting state.

The question becomes:

What happens if we use that unfamiliar sequence to interrogate Alice's hedge?

---

## The First Advice From the Swarm: Do Not Stop

The most important result of the experiment can be expressed in an extraordinarily simple sentence:

**Follow the process further.**

Do not stop when Alice has measured her risk.

Do not stop when Alice has hedged her risk.

Do not stop when Alice's measured exposure has fallen.

Ask what happened next.

Alice did not exist in isolation before she acted. She was already embedded in a financial system.

She had counterparties.

She had collateral agreements.

She depended on funding markets.

She owned assets also held by other institutions.

She operated through exchanges, clearing houses and settlement systems.

She was connected through prices to institutions with which she might have had no direct contractual relationship whatsoever.

Her balance sheet was part of a larger network of balance sheets.

Therefore, when Alice acts, the interesting question is not merely what happens to Alice.

The interesting question is what Alice's action does to the system.

Suppose Alice buys protection from Bob.

Alice's market exposure may decline, but Bob now has a contingent obligation.

Perhaps Bob hedges that obligation with Charlie.

Perhaps the contract requires Bob to post additional collateral when markets move.

Bob therefore needs liquidity.

He sells assets to obtain it.

Those sales affect market prices.

Those prices are used to value Diana's portfolio.

Diana now breaches a risk limit.

She sells as well.

Prices move again.

A clearing house increases margin requirements.

More institutions need liquidity.

Some draw committed credit facilities.

Banks providing those facilities see their own liquidity positions change.

None of this means that Alice's original hedge was necessarily a mistake.

It means something more subtle:

**the hedge was an intervention in a system.**

The reduction in Alice's measured risk was only one part of the event.

---

## Bob Is Not Waiting for a Message

This point is particularly important because it prevents the teleportation analogy from collapsing into a simplistic story about information transmission.

In our framework, Bob is not merely waiting at the other end for Alice to send him a message that he somehow decodes.

That is not the insight.

Bob matters because **Bob's economic state can change as a consequence of Alice's action**.

And if Bob's state changes, Bob may act.

That action can affect Charlie.

Charlie may react.

The resulting price movement can affect Diana.

Diana may respond.

The system continues evolving.

The relevant object is therefore not a message moving from Alice to Bob. It is a **cascade of state changes within an interconnected structure**.

This is where the inspiration from quantum teleportation operates at a higher altitude.

We are not saying:

## The Meaning of “Entanglement” in This Exercise

The word entanglement must be handled carefully.

Financial institutions are not quantum-entangled objects.

But the concept directs our attention toward something economically important: **separation does not necessarily imply independence**.

Two institutions may be legally separate, geographically distant and operationally independent, yet their economic conditions can be tightly connected.

They may be counterparties.

They may own the same assets.

One may guarantee the obligations of another.

They may depend on the same source of short-term funding.

They may post collateral to the same clearing system.

They may be exposed to the same dealer.

One institution's asset may be another institution's liability.

A decline in a common asset price may affect both simultaneously.

The useful question inspired by entanglement is therefore not whether these relationships are physically quantum.

They plainly are not.

The question is whether our model of risk should begin from the recognition that the state of an institution sometimes cannot be adequately understood without the relationships connecting it to the rest of the system.

That is a much more defensible proposition.

And potentially a very useful one.

## Conclusion — From teleportation metaphor to a candidate architecture of risk

The purpose of this notebook has not been to claim that finance is secretly quantum mechanical. It has been to use one of the most conceptually precise protocols in quantum information as a disciplined provocation for rethinking what financial institutions actually do when they say that they have “transferred risk.”

The conventional phrase is useful but incomplete. When a bank hedges an option book, when an asset manager buys protection, when a corporation swaps floating for fixed payments, or when an insurer reinsures a liability, the underlying uncertainty has not disappeared from the economic system. A transformation has occurred. Exposures have been reconfigured; contingent claims have been created; collateral and liquidity obligations may have appeared; information must be measured and communicated; and another institution or market mechanism has become conditionally involved in the outcome.

Quantum teleportation provides an unusually rich architecture for thinking about this process because the protocol itself is not a story of an object simply moving from A to B. It requires a pre-existing resource, a measurement, classical communication, and a conditional correction. Reconstruction occurs subject to strict informational structure. The original state cannot simply be cloned. Translating these architectural ideas into finance produces the candidate framework developed here: **Teleportation-Inspired Risk Management, or TIRM**.

The framework begins with a multidimensional risk state rather than a single scalar measure. Market exposure, covariance structure, liquidity, counterparty conditions, model uncertainty, and network position jointly determine the institution's economic state. A risk-management intervention specifies an intended transformation of that state. The realized post-intervention state will generally differ from the intended one. The difference is the residual reconstruction error.

This perspective changes the central question. Instead of asking only, “How much risk did the hedge remove?”, TIRM asks: **What state was intended, what resources were required to reconstruct it, what information was measured, what communication and execution occurred, where did the associated contingent exposure migrate, and with what fidelity was the target state achieved?**

Several candidate principles follow. Risk reduction in one channel does not establish that economic uncertainty has vanished. High-fidelity reconstruction consumes scarce capacity. Effective intervention during stress depends partly on resources positioned before the shock. Measurement is incomplete. Communication and execution take time. Contractual networks enable redistribution while simultaneously creating dependencies. Replication is not free. And local improvement can diverge from system-wide improvement.

The toy simulation demonstrates the intuition without pretending to prove the theory. An institution can reduce its direct market-factor exposure while increasing liquidity and counterparty burdens. Under sufficiently adverse implementation conditions, the local market-risk metric improves while the generalized risk measure deteriorates. The result illustrates why the location and representation of residual risk matter.

The no-free-cloning analogy is especially productive when stated carefully. Financial replication is commonplace, but perfect replication is neither universal nor costless. Instruments must exist. Markets must possess depth. Models must correctly identify the relevant state variables. Counterparties must perform. Dynamic strategies require trading. Balance sheets and collateral are finite. Thus a financial risk state may be reproducible to useful precision, but fidelity is purchased with resources and remains conditional on assumptions.

The analogy also reveals a dual nature of financial coupling. Pre-existing contracts and institutional relationships are resources because they make rapid conditional reconstruction possible. Yet the same connections can transmit distress. A derivatives network, central counterparty, insurance arrangement, or committed funding line can stabilize one institution while concentrating dependence elsewhere. The architecture that permits risk transformation is therefore also part of the architecture of systemic risk.

This is where TIRM could become more than terminology. It suggests that risk management should be evaluated by **reconstruction fidelity and migration accounting**, not solely by the reduction of a local metric. A hedge with excellent delta reduction but severe liquidity sensitivity may have lower generalized fidelity than its market-risk statistics imply. An institution with substantial contingent resources established before stress may be able to reconstruct target states more reliably than an institution attempting to acquire those resources after markets have already moved.

These statements are hypotheses, not conclusions. The notebook therefore formulates falsifiable propositions. If execution latency has no measurable relationship with realized hedge error, the communication-latency principle weakens. If local market-risk reductions do not systematically migrate into measurable counterparty, funding, liquidity, or basis channels, the redistribution thesis loses force. If pre-positioned capacity does not improve intervention quality during stress, one of the framework's central architectural analogies would be unsupported. If network concentration does not help explain local-systemic divergence, that part of the theory should be discarded.

The most important safeguard is that the analogy has explicit boundaries. Financial exposures are not wavefunctions. Correlated banks are not quantum-entangled particles. There is no physical financial counterpart to the Bell-state measurement, and the proposed residual-risk principle is not a literal physical conservation law. TIRM earns a place in financial theory only if its state-transformation language produces clearer models, better measurements, or empirical predictions that simpler descriptions fail to provide.

The fifty-agent swarm contributes to this process in a particular way. Quantum-information specialists prevent abuse of the source theory. Risk and derivatives specialists translate the architecture into recognizable financial mechanisms. Network theorists expose systemic redistribution. Information and control specialists formalize measurement and reconstruction. Liquidity specialists force the theory into real markets. Governance agents introduce institutional latency and authority. Mathematical agents seek parsimony. Adversarial agents attempt to show that the entire framework is unnecessary. Cross-family agreement therefore has greater significance than simple repetition.

The notebook consequently represents a different use of agentic AI from the preceding closed-form-option experiment. There, the swarm searched for a compact mathematical approximation to a known numerical object. Here, the swarm participates in **theory construction**. It proposes concepts, tests mappings across disciplines, searches for invariants, creates falsifiable statements, and exposes conceptual failure points. Deterministic code then measures convergence and illustrates the resulting mechanisms.

The next research step would be empirical. A serious TIRM program could construct reconstruction-fidelity measures from transaction data, hedge slippage, collateral movements, counterparty exposures, and execution latency. Historical stress periods could reveal whether institutions with stronger pre-positioned capacity achieved better state reconstruction. Network data could test whether apparent local de-risking coincided with concentration of contingent exposure elsewhere. These studies would determine whether teleportation-inspired language has genuine explanatory power.

The theory therefore ends where a useful theory should begin: with propositions capable of being rejected. Quantum teleportation supplies the intellectual spark. The swarm supplies heterogeneous disciplined reasoning. Financial mathematics supplies the state representation. Simulation supplies intuition. Empirical evidence must ultimately decide what survives.
