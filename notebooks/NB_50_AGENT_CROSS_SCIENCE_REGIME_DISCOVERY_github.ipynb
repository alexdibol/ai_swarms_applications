# CROSS-SCIENCE MARKET REGIME DISCOVERY WITH A 50-AGENT SWARM

Financial markets are observed continuously but their governing state is not. Prices, volatility, liquidity, correlations, credit spreads, yield curves, volume, dispersion and cross-asset relationships are visible; the latent condition that makes those signals cohere is not. Traders therefore confront an inverse problem. We see consequences and try to infer the hidden mechanism that generated them. A calm expansion, a momentum-dominated risk-on phase, an inflation shock, a liquidity crisis, a deleveraging episode and an early recovery can produce radically different mappings from the same observable signal to subsequent returns. A strategy that is sensible in one state may be destructive in another. This makes regime identification one of the deepest recurring problems in market science.

The Hidden Markov Model became famous in this setting for a good reason. It gives a disciplined probabilistic language for an unobserved state that evolves through time and emits observable data. Yet the HMM is only one member of a much larger intellectual family. Other sciences have spent decades, and sometimes centuries, solving structurally similar problems: infer an invisible state from noisy measurements; detect phase transitions; identify changes in dynamical systems; distinguish signal from noise; recover hidden sources; characterize metastable states; recognize topology changes; infer a system’s energy landscape; and determine when an apparently stable process has crossed into a different operating domain. The central premise of this notebook is that finance should borrow more aggressively from those neighboring mathematical traditions.

The experiment therefore uses transfer learning in a broad scientific sense. We are not merely transferring neural-network weights from one dataset to another. We transfer abstractions, mathematical objects, diagnostics and inferential habits from other fields into market-regime discovery. Fifty Claude Sonnet 5 agents are divided into ten disciplinary families: statistical physics, dynamical systems, electrical and signal engineering, control theory, information theory, applied mathematics, chemistry and statistical mechanics, astrophysics, network science, and geophysics/seismology. Five agents within each family receive different methodological identities. Their task is not to forecast the market directly. Their task is to ask a more foundational question: if this market-regime problem appeared in my own science, what hidden object would I infer, what observables would I measure, what mathematics would I use, and what empirical signature would convince me that the inferred state is real?

This distinction is essential. A swarm is useful only if diversity is genuine. Fifty agents repeating a financial HMM with different wording would add cost but little knowledge. Here disciplinary prompts deliberately create different priors. A physicist may see phase transitions and order parameters. A signal engineer may see a latent source embedded in multiscale noise. A control theorist may think in terms of state-space observability and filtering. A chemist may recognize metastability and transition rates. An astrophysicist may use change-point detection and weak-signal extraction. A network scientist may define regime through correlation topology rather than through the marginal distribution of returns. A geophysicist may focus on precursors, critical slowing, rupture-like transitions and multiscale event clustering. These are not metaphors for decoration; the notebook asks the agents to translate them into mathematical candidates that can be operationalized.

The architecture separates scientific imagination from empirical discipline. Claude is used for hypothesis generation, model specification, critique and synthesis. Python is used for data, feature construction, deterministic diagnostics, clustering, transition analysis, validation and provenance. The language model is never allowed to manufacture market observations. It receives a compact numerical description of the empirical feature space and proposes models against an explicit schema. This separation matters because a plausible narrative is not evidence of a regime. A useful regime model must produce states that are persistent enough to be meaningful, distinct enough to interpret, stable enough to survive perturbations, and economically consequential enough to change conditional market behavior.

The notebook also avoids the common mistake of forcing a single winning model too early. The swarm first generates a library of cross-science proposals. Those proposals are embedded into a common taxonomy: latent-state representation, transition mechanism, observation mapping, mathematical machinery, diagnostics, falsification tests and expected failure modes. A synthesis stage then looks for recurring structures across disciplines. The objective is an architecture rather than a beauty contest. In complex markets, the best representation may be composite: a probabilistic latent state for persistence, spectral features for multiscale structure, network features for collective behavior, and change-point logic for transitions. The swarm is therefore asked to identify complementarities as well as disagreements.

For the empirical substrate, the notebook downloads a diversified set of liquid market proxies and constructs signals that represent several dimensions of market state: trend, realized volatility, downside asymmetry, cross-sectional dispersion, average correlation, breadth, drawdown, volatility-of-volatility, volume pressure where available, and cross-asset relative behavior. The exact assets are replaceable. What matters is that regime inference is multivariate. A hidden state of the market should not be inferred from the S&P 500 return alone if the claim is that the state governs the market system.

The executable benchmark deliberately remains transparent. It constructs a robustly scaled feature matrix, applies several unsupervised views of the same latent problem, and produces a consensus state representation. The swarm’s theoretical proposals are not blindly converted into arbitrary code generated at runtime. Instead, they inform a controlled architecture built from auditable components. This is a crucial governance choice. Scientific creativity occurs in the proposal layer; empirical claims occur in the reproducible layer. The notebook records prompts, model identifiers, request IDs, agent outputs, consensus diagnostics and validation statistics so that the research path can be inspected later.

Validation is treated as a first-class problem. Regime discovery is unusually vulnerable to beautiful overfitting because there is no directly observed label called “true market regime.” The notebook therefore asks whether inferred states exhibit persistence, whether they differ in subsequent return and risk distributions, whether the partition is stable under feature perturbation, whether a smaller subswarm reaches similar scientific conclusions, and whether transitions concentrate around periods of unusually large changes in the market feature vector. None of these tests proves ontological truth. Together, however, they make the latent-state hypothesis more falsifiable.

The final output is a research architecture for a “Market Regime Observatory”: an ensemble that treats regime as a latent, evolving, multiscale object rather than as a single categorical label. It distinguishes state estimation from transition detection and uncertainty. Instead of saying merely “the market is in regime 3,” the system aims to report the current consensus state, its probability or confidence, the features that make the state distinctive, the degree of disagreement among alternative scientific lenses, and evidence that the system may be approaching a transition.

The ambition is deliberately high but the epistemic claim is modest. This notebook does not announce that the Holy Grail has been solved. It constructs a disciplined experiment for searching a much larger hypothesis space than finance usually explores. Its value lies in asking whether fifty scientifically diverse agents can discover transferable mathematical structures that a finance-only research team might overlook, and whether those structures can be compressed into an empirical architecture that survives transparent tests. If successful, the contribution is not a magical regime label. It is a better research machine for reasoning about hidden market conditions.

### CODE UNIT 2 — Building the multivariate market-state measurement system

CODE UNIT 2 turns raw prices into an observable state vector. This is the empirical heart of the inverse problem: the hidden regime is not observed, so the quality of inference depends on whether the measurements span the dimensions through which regimes reveal themselves. The feature set is intentionally compact enough for interpretation but broad enough to represent trend, volatility, dependence, breadth, drawdown and cross-asset stress.

Daily log returns are computed first. Rolling equity trend and volatility summarize directional and risk conditions. Downside volatility separates harmful variability from symmetric noise. Breadth measures the fraction of risk assets with positive medium-horizon momentum. Average rolling correlation captures the familiar phenomenon that diversification can disappear in stress. Cross-sectional dispersion measures disagreement among assets. Drawdown records path dependence. Changes in VIX and credit-relative performance introduce forward-looking risk pricing and financing conditions. Treasury/equity and gold/equity relative momentum add information about defensive rotation.

Features are winsorized through robust quantile clipping and then transformed with RobustScaler, which is less sensitive to extreme observations than mean-variance standardization. This is important because extreme market events are informative but should not numerically dominate every distance calculation. The code retains both the economically interpretable feature table and the scaled matrix used by algorithms.

The cell also constructs a compact empirical summary that later goes to the swarm. Agents do not receive thousands of daily rows. They receive feature names, moments and correlations, enough to anchor proposals in the actual measurement system while keeping calls inexpensive. This preserves the separation between language-model hypothesis generation and deterministic data analysis.


```python
from sklearn.preprocessing import RobustScaler

rets = np.log(px).diff()
risk_assets = [c for c in ["SPY","QQQ","IWM","HYG","LQD","GLD","DBC","TLT"] if c in px.columns]

F = pd.DataFrame(index=px.index)
F["spy_ret_5d"] = np.log(px["SPY"]).diff(5)
F["spy_trend_63d"] = np.log(px["SPY"]).diff(63)
F["spy_vol_21d"] = rets["SPY"].rolling(21).std() * np.sqrt(252)
F["spy_downvol_63d"] = rets["SPY"].clip(upper=0).rolling(63).std() * np.sqrt(252)
F["spy_drawdown"] = px["SPY"] / px["SPY"].cummax() - 1
F["cross_asset_dispersion"] = rets[risk_assets].rolling(21).std().mean(axis=1)

rolling_corr = rets[risk_assets].rolling(63).corr()
avg_corr = []
dates = []
for dt in px.index:
    try:
        C = rolling_corr.loc[dt]
        a = C.values[np.triu_indices_from(C.values, 1)]
        avg_corr.append(np.nanmean(a)); dates.append(dt)
    except Exception:
        pass
F.loc[dates, "avg_corr_63d"] = avg_corr

mom63 = np.log(px[risk_assets]).diff(63)
F["breadth_63d"] = (mom63 > 0).mean(axis=1)
if "^VIX" in px:
    F["vix_level"] = np.log(px["^VIX"])
    F["vix_change_21d"] = np.log(px["^VIX"]).diff(21)
if "HYG" in px and "LQD" in px:
    F["credit_risk_21d"] = np.log(px["HYG"]/px["LQD"]).diff(21)
if "TLT" in px:
    F["bond_equity_rotation_21d"] = np.log(px["TLT"]/px["SPY"]).diff(21)
if "GLD" in px:
    F["gold_equity_rotation_21d"] = np.log(px["GLD"]/px["SPY"]).diff(21)

F = F.replace([np.inf,-np.inf], np.nan).dropna()
lo, hi = F.quantile(.01), F.quantile(.99)
F_clip = F.clip(lo, hi, axis=1)

scaler = RobustScaler()
X = pd.DataFrame(scaler.fit_transform(F_clip), index=F_clip.index, columns=F_clip.columns)

empirical_summary = {
    "features": list(F.columns),
    "n_observations": int(len(F)),
    "start": str(F.index.min().date()),
    "end": str(F.index.max().date()),
    "feature_medians": F.median().round(4).to_dict(),
    "feature_correlations": F.corr().round(2).to_dict()
}
print("Feature matrix:", X.shape)
display(F.tail())
```

The JSON schema is strict and compact. String lengths are controlled by prompting rather than enormous token budgets. This unit does not yet call Claude. It builds the population and the output contract, making the swarm inspectable before cost is incurred. The resulting roster can be displayed and audited. ### CODE UNIT 4 — Current Claude Sonnet 5 structured-output interface and canary

CODE UNIT 4 is the live Claude interface and is deliberately engineered around the current Sonnet 5 API rather than patterns inherited from older Claude generations. The notebook uses the Messages API with the model string `claude-sonnet-5`, low effort through `output_config`, and native JSON Schema structured output. It does not use temperature, top-p, top-k, assistant prefilling, manual thinking-token budgets, regex extraction or “please return JSON” followed by hopeful parsing.

The function checks `stop_reason` before accepting the response. If the first call ends because the token ceiling was reached, it retries once with a larger output allowance. Only text blocks are concatenated, and the resulting string is parsed with `json.loads`. Request IDs and stop reasons are retained for provenance. A malformed or refused response raises an explicit error rather than being converted into a fabricated fallback opinion.

Most importantly, the code performs a canary call on the first agent. This is the protection against the failure pattern that has affected earlier notebooks: launching dozens of expensive calls with an outdated parameter or response format and discovering the incompatibility only after the swarm is underway. The remaining forty-nine calls run only after the first response has passed schema-shaped parsing and basic identity checks.

The prompt asks each agent to remain inside its scientific discipline and to translate, not merely rename, its mathematics. It also tells the agent that the output is a research hypothesis, not investment advice. The empirical summary is included in compact JSON form so proposals are connected to the actual feature space.


```python
# ============================================================
# CODE CELL 4
# CLAUDE SONNET 5 INTERFACE & CANARY TEST
# ============================================================

import json


# ============================================================
# 1. CLAUDE-COMPATIBLE STRUCTURED OUTPUT SCHEMA
# ============================================================
#
# Array lengths and numerical ranges are intentionally NOT
# encoded here.
#
# They are enforced locally after Claude returns its JSON.
# ============================================================

CLAUDE_AGENT_SCHEMA = {

    "type": "object",

    "properties": {

        "agent_id": {
            "type": "string"
        },

        "family": {
            "type": "string"
        },

        "proposed_name": {
            "type": "string"
        },

        "latent_object": {
            "type": "string"
        },

        "transfer_principle": {
            "type": "string"
        },

        "mathematical_core": {
            "type": "array",
            "items": {
                "type": "string"
            }
        },

        "required_features": {
            "type": "array",
            "items": {
                "type": "string"
            }
        },

        "transition_detector": {
            "type": "string"
        },

        "state_count": {
            "type": "integer"
        },

        "validation_tests": {
            "type": "array",
            "items": {
                "type": "string"
            }
        },

        "failure_modes": {
            "type": "array",
            "items": {
                "type": "string"
            }
        },

        "novelty": {
            "type": "number"
        },

        "plausibility": {
            "type": "number"
        },

        "confidence": {
            "type": "number"
        },

        "rationale": {
            "type": "string"
        }
    },


    "required": [

        "agent_id",

        "family",

        "proposed_name",

        "latent_object",

        "transfer_principle",

        "mathematical_core",

        "required_features",

        "transition_detector",

        "state_count",

        "validation_tests",

        "failure_modes",

        "novelty",

        "plausibility",

        "confidence",

        "rationale"
    ],


    "additionalProperties": False
}


# ============================================================
# 2. SCHEMA PREFLIGHT
# ============================================================

schema_text = json.dumps(
    CLAUDE_AGENT_SCHEMA
)


FORBIDDEN_SCHEMA_TERMS = [

    '"minItems"',

    '"maxItems"',

    '"minimum"',

    '"maximum"'
]


for term in FORBIDDEN_SCHEMA_TERMS:

    assert term not in schema_text, (
        f"Unsupported schema constraint detected: {term}"
    )


print(
    "Claude structured-output schema: CLEAN"
)


# ============================================================
# 3. LOCAL SCIENTIFIC VALIDATION
# ============================================================

def validate_agent_output(
    obj,
    agent
):

    # --------------------------------------------------------
    # Identity
    # --------------------------------------------------------

    if obj["agent_id"] != agent["agent_id"]:

        raise ValueError(
            "Agent ID mismatch."
        )


    if obj["family"] != agent["family"]:

        raise ValueError(
            "Scientific family mismatch."
        )


    # --------------------------------------------------------
    # Array lengths
    # --------------------------------------------------------

    if not (
        2
        <= len(obj["mathematical_core"])
        <= 5
    ):

        raise ValueError(
            "mathematical_core must contain 2–5 items."
        )


    if not (
        2
        <= len(obj["required_features"])
        <= 8
    ):

        raise ValueError(
            "required_features must contain 2–8 items."
        )


    if not (
        2
        <= len(obj["validation_tests"])
        <= 5
    ):

        raise ValueError(
            "validation_tests must contain 2–5 items."
        )


    if not (
        1
        <= len(obj["failure_modes"])
        <= 3
    ):

        raise ValueError(
            "failure_modes must contain 1–3 items."
        )


    # --------------------------------------------------------
    # State count
    # --------------------------------------------------------

    if not (
        2
        <= int(obj["state_count"])
        <= 8
    ):

        raise ValueError(
            "state_count must be between 2 and 8."
        )


    # --------------------------------------------------------
    # Quantitative scores
    # --------------------------------------------------------

    for field in [

        "novelty",

        "plausibility",

        "confidence"

    ]:

        value = float(
            obj[field]
        )


        if not (
            0.0
            <= value
            <= 1.0
        ):

            raise ValueError(
                f"{field} must be between 0 and 1."
            )


    # --------------------------------------------------------
    # Feature validation
    # --------------------------------------------------------

    available_features = set(
        empirical_summary["features"]
    )


    requested_features = set(
        obj["required_features"]
    )


    unknown_features = (
        requested_features
        -
        available_features
    )


    if unknown_features:

        raise ValueError(
            "Agent requested unknown features: "
            f"{sorted(unknown_features)}"
        )


    return True


# ============================================================
# 4. AGENT PROMPT
# ============================================================

def build_agent_prompt(
    agent
):

    return f"""
You are Agent {agent['agent_id']} in a 50-agent
cross-science research swarm.

SCIENTIFIC FAMILY:
{agent['family']}

SPECIALTY:
{agent['specialty']}

RESEARCH QUESTION:

Can mathematical ideas from your scientific specialty reveal
latent financial market regimes that conventional approaches
may fail to identify?

The observed financial system is represented by the empirical
features below.

The true market regime is latent.

Your task is to propose ONE scientifically meaningful,
empirically falsifiable latent-state architecture inspired by
your assigned discipline.

Do not merely rename a Hidden Markov Model.

AVAILABLE MARKET FEATURES:

{json.dumps(empirical_summary, default=str)}

RESPONSE REQUIREMENTS:

mathematical_core:
2–5 concise mathematical ideas.

required_features:
2–8 names drawn ONLY from the available features.

validation_tests:
2–5 empirical falsification tests.

failure_modes:
1–3 important failure modes.

state_count:
integer from 2 through 8.

novelty:
number between 0 and 1.

plausibility:
number between 0 and 1.

confidence:
number between 0 and 1.

rationale:
approximately 60–120 words.

IMPORTANT:

Be concise.

Do not write an essay.

Use the structured output schema.

This is exploratory educational research,
not investment advice.
"""


# ============================================================
# 5. CLAUDE CALL
# ============================================================

def call_agent(
    agent,
    max_tokens=None
):

    if max_tokens is None:

        max_tokens = INITIAL_MAX_TOKENS


    response = client.messages.create(

        model=MODEL,

        max_tokens=max_tokens,

        messages=[

            {

                "role": "user",

                "content":
                    build_agent_prompt(agent)
            }

        ],

        output_config={

            "effort":
                EFFORT,

            "format": {

                "type":
                    "json_schema",

                "schema":
                    CLAUDE_AGENT_SCHEMA
            }
        }
    )


    # ========================================================
    # 6. ADAPTIVE TOKEN ESCALATION
    # ========================================================

    if response.stop_reason == "max_tokens":

        if max_tokens < RETRY_MAX_TOKENS:

            print(
                f"       ↳ {agent['agent_id']} "
                f"truncated at {max_tokens} tokens."
            )

            print(
                f"       ↳ Retrying with "
                f"{RETRY_MAX_TOKENS}..."
            )


            return call_agent(

                agent,

                max_tokens=
                    RETRY_MAX_TOKENS
            )


        elif max_tokens < FINAL_MAX_TOKENS:

            print(
                f"       ↳ {agent['agent_id']} "
                f"still truncated."
            )

            print(
                f"       ↳ Final retry with "
                f"{FINAL_MAX_TOKENS}..."
            )


            return call_agent(

                agent,

                max_tokens=
                    FINAL_MAX_TOKENS
            )


        else:

            raise RuntimeError(

                f"{agent['agent_id']} could not complete "
                f"structured output within "
                f"{FINAL_MAX_TOKENS} tokens."
            )


    # ========================================================
    # 7. EXTRACT TEXT BLOCK
    # ========================================================

    text_blocks = [

        block.text

        for block in response.content

        if getattr(
            block,
            "type",
            None
        ) == "text"
    ]


    if not text_blocks:

        raise RuntimeError(

            f"{agent['agent_id']} returned no text block. "
            f"stop_reason={response.stop_reason}"
        )


    raw_json = "".join(
        text_blocks
    )


    # ========================================================
    # 8. PARSE STRUCTURED OUTPUT
    # ========================================================

    try:

        result = json.loads(
            raw_json
        )

    except json.JSONDecodeError as exc:

        raise RuntimeError(

            f"{agent['agent_id']} returned invalid JSON."

        ) from exc


    # ========================================================
    # 9. LOCAL VALIDATION
    # ========================================================

    validate_agent_output(

        result,

        agent
    )


    # ========================================================
    # 10. PROVENANCE
    # ========================================================

    result["_request_id"] = getattr(

        response,

        "_request_id",

        None
    )


    result["_stop_reason"] = (
        response.stop_reason
    )


    result["_token_ceiling_used"] = (
        max_tokens
    )


    return result


# ============================================================
# 11. ONE-AGENT CANARY
# ============================================================

print("\n" + "=" * 72)

print(
    "CLAUDE SONNET 5 — ONE-AGENT CANARY"
)

print("=" * 72)


print(
    f"\nModel      : {MODEL}"
)

print(
    f"Effort     : {EFFORT}"
)

print(
    f"Agent      : {agents[0]['agent_id']}"
)

print(
    f"Family     : {agents[0]['family']}"
)

print(
    f"Specialty  : {agents[0]['specialty']}"
)


print(
    "\nRunning canary...\n"
)


canary = call_agent(
    agents[0]
)


# ============================================================
# 12. CANARY REPORT
# ============================================================

print("=" * 72)

print(
    "CANARY PASSED"
)

print("=" * 72)


print(
    "\nClaude Sonnet 5           : OK"
)

print(
    "Low reasoning effort      : OK"
)

print(
    "Structured JSON           : OK"
)

print(
    "Scientific validation     : OK"
)

print(
    "Feature validation        : OK"
)

print(
    "Adaptive token handling   : OK"
)


print(
    f"\nToken ceiling actually used: "
    f"{canary['_token_ceiling_used']}"
)


print(
    "\nREADY FOR THE FULL 50-AGENT SWARM"
)


display(

    pd.DataFrame(
        [canary]
    ).T
)
```

Parallelism could reduce elapsed time, but sequential execution is easier to debug, gentler on rate limits, and more transparent in an educational notebook. Progress is printed after each agent, and every successful response is checkpointed to disk immediately. If Colab disconnects later, the intellectual output already obtained is not lost.

The cell then constructs two views of the swarm. The first is a proposal table containing discipline, model name, latent object, transition detector, proposed state count, novelty, plausibility and confidence. The second is a vocabulary analysis over the mathematical-core fields. This is not intended as sophisticated natural-language processing. It is a transparent first map of which mathematical ideas recur across otherwise independent disciplines.

A useful swarm should show both convergence and disagreement. Convergence across unrelated fields can be evidence that a mathematical structure is broadly appropriate to the inverse problem. Disagreement is equally valuable because it reveals dimensions that a single model may omit. For example, several families may agree that persistence is important while disagreeing about whether transitions should be represented as probabilistic switching, topological rupture, spectral change or energy-barrier crossing.

The output is saved as JSON and CSV. No agent is silently discarded because its proposal seems unusual. The next synthesis stage receives the complete corpus and is asked to construct an architecture that preserves complementary ideas while avoiding redundant complexity.

### CODE UNIT 6 — Cross-disciplinary synthesis into one research architecture

CODE UNIT 6 asks a separate synthesis call to act as a scientific architect rather than as a fifty-first domain specialist. Its input is the complete proposal corpus, compressed enough to fit comfortably in context. The synthesis contract requires the model to identify cross-disciplinary motifs, preserve meaningful dissent, and specify a concrete hybrid architecture with four layers: representation, state inference, transition detection and uncertainty/validation.

This stage is intentionally not a vote. Majority rule would reward familiar ideas and could erase the rare transfer that makes the exercise worthwhile. Instead, the synthesizer must explain why selected components are complementary and which ideas remain competing hypotheses. It must also produce an executable blueprint using only a controlled menu of empirical components available in the notebook: robust scaling, PCA, Gaussian mixtures, K-means, spectral clustering, rolling transition intensity and consensus voting. This constraint prevents the language model from inventing a library or algorithm that the subsequent code cannot reproduce.

The scientific architecture is therefore a negotiated boundary between open-ended theory generation and reproducible implementation. The swarm can propose topological data analysis, free-energy landscapes or Bayesian change points, but the first empirical prototype translates recurring concepts into transparent approximations. Later research can implement the more specialized methods one by one and test whether they improve out-of-sample behavior.

The output also specifies a target state count. That number is treated as a hypothesis rather than an unquestionable truth; the next unit compares nearby values using separation and persistence diagnostics. The synthesis is stored with request provenance so that later revisions can be compared.


```python
# ============================================================
# CODE CELL 6
# SWARM SYNTHESIS — FROM 50 PROPOSALS TO ONE ARCHITECTURE
# CLAUDE SONNET 5 / LOW EFFORT / STRUCTURED OUTPUT
# ============================================================

import json
import pandas as pd


# ============================================================
# 1. PRECONDITIONS
# ============================================================

if "proposals" not in globals():
    raise RuntimeError(
        "The variable 'proposals' does not exist. "
        "Run Cell 5 first."
    )

if len(proposals) != 50:
    raise RuntimeError(
        f"Cell 6 requires the complete 50-agent swarm. "
        f"Currently available: {len(proposals)}/50."
    )

print("=" * 72)
print("CELL 6 — SWARM SYNTHESIS")
print("=" * 72)

print(f"\nProposals available : {len(proposals)}")
print(f"Model               : {MODEL}")
print(f"Effort              : {EFFORT}")


# ============================================================
# 2. CLAUDE-COMPATIBLE SYNTHESIS SCHEMA
# ============================================================
#
# IMPORTANT:
#
# Do NOT place unsupported constraints such as:
#
#     minItems: 3
#     maxItems: 8
#     minimum: 2
#     maximum: 8
#
# inside the raw Anthropic schema.
#
# Those constraints are enforced locally after generation.
# ============================================================

SYNTH_SCHEMA = {

    "type": "object",

    "properties": {

        "architecture_name": {
            "type": "string"
        },

        "core_motifs": {
            "type": "array",
            "items": {
                "type": "string"
            }
        },

        "minority_ideas_to_preserve": {
            "type": "array",
            "items": {
                "type": "string"
            }
        },

        "representation_layer": {
            "type": "string"
        },

        "state_inference_layer": {
            "type": "string"
        },

        "transition_layer": {
            "type": "string"
        },

        "uncertainty_layer": {
            "type": "string"
        },

        "target_states": {
            "type": "integer"
        },

        "empirical_components": {

            "type": "array",

            "items": {

                "type": "string",

                "enum": [
                    "PCA",
                    "GaussianMixture",
                    "KMeans",
                    "SpectralClustering",
                    "TransitionIntensity",
                    "ConsensusVoting"
                ]
            }
        },

        "falsification_criteria": {
            "type": "array",
            "items": {
                "type": "string"
            }
        },

        "research_rationale": {
            "type": "string"
        }
    },

    "required": [

        "architecture_name",

        "core_motifs",

        "minority_ideas_to_preserve",

        "representation_layer",

        "state_inference_layer",

        "transition_layer",

        "uncertainty_layer",

        "target_states",

        "empirical_components",

        "falsification_criteria",

        "research_rationale"
    ],

    "additionalProperties": False
}


# ============================================================
# 3. SCHEMA PREFLIGHT
# ============================================================

schema_text = json.dumps(
    SYNTH_SCHEMA
)

FORBIDDEN_SCHEMA_TERMS = [
    '"minItems"',
    '"maxItems"',
    '"minimum"',
    '"maximum"'
]

for term in FORBIDDEN_SCHEMA_TERMS:

    if term in schema_text:

        raise RuntimeError(
            f"Unsupported schema term detected: {term}"
        )

print("\nStructured-output schema: CLEAN")


# ============================================================
# 4. COMPACT THE 50 AGENT PROPOSALS
# ============================================================
#
# Cell 6 does not need provenance fields or long metadata.
# We retain only the scientific content required for synthesis.
# ============================================================

KEEP_FIELDS = [

    "agent_id",

    "family",

    "proposed_name",

    "latent_object",

    "transfer_principle",

    "mathematical_core",

    "transition_detector",

    "state_count",

    "validation_tests",

    "failure_modes",

    "novelty",

    "plausibility",

    "confidence"
]


compact = []


for p in proposals:

    compact.append({

        field: p[field]

        for field in KEEP_FIELDS

        if field in p
    })


assert len(compact) == 50


# ============================================================
# 5. SCIENTIFIC FAMILY SUMMARY
# ============================================================
#
# This gives Claude a second view of the swarm:
# individual proposals + aggregate disciplinary structure.
# ============================================================

family_summary_for_prompt = (

    proposal_df

    .groupby("family")[

        [
            "novelty",
            "plausibility",
            "confidence",
            "state_count"
        ]
    ]

    .mean()

    .round(3)

    .reset_index()

    .to_dict(
        orient="records"
    )
)


# ============================================================
# 6. LOCAL VALIDATOR
# ============================================================

ALLOWED_COMPONENTS = {

    "PCA",

    "GaussianMixture",

    "KMeans",

    "SpectralClustering",

    "TransitionIntensity",

    "ConsensusVoting"
}


def validate_synthesis(obj):

    # --------------------------------------------------------
    # Core motifs
    # --------------------------------------------------------

    if not (
        3 <= len(obj["core_motifs"]) <= 8
    ):

        raise ValueError(
            "core_motifs must contain 3–8 items."
        )


    # --------------------------------------------------------
    # Minority ideas
    # --------------------------------------------------------

    if not (
        2
        <= len(obj["minority_ideas_to_preserve"])
        <= 6
    ):

        raise ValueError(
            "minority_ideas_to_preserve "
            "must contain 2–6 items."
        )


    # --------------------------------------------------------
    # Number of latent states
    # --------------------------------------------------------

    K_candidate = int(
        obj["target_states"]
    )


    if not (
        2 <= K_candidate <= 8
    ):

        raise ValueError(
            "target_states must be between 2 and 8."
        )


    # --------------------------------------------------------
    # Empirical components
    # --------------------------------------------------------

    components = (
        obj["empirical_components"]
    )


    if not (
        3 <= len(components) <= 6
    ):

        raise ValueError(
            "empirical_components "
            "must contain 3–6 items."
        )


    unknown = (
        set(components)
        -
        ALLOWED_COMPONENTS
    )


    if unknown:

        raise ValueError(
            "Unsupported empirical components: "
            f"{sorted(unknown)}"
        )


    # --------------------------------------------------------
    # Falsification criteria
    # --------------------------------------------------------

    if not (
        3
        <= len(obj["falsification_criteria"])
        <= 7
    ):

        raise ValueError(
            "falsification_criteria "
            "must contain 3–7 items."
        )


    # --------------------------------------------------------
    # Required textual layers
    # --------------------------------------------------------

    text_fields = [

        "architecture_name",

        "representation_layer",

        "state_inference_layer",

        "transition_layer",

        "uncertainty_layer",

        "research_rationale"
    ]


    for field in text_fields:

        if not str(
            obj[field]
        ).strip():

            raise ValueError(
                f"{field} cannot be empty."
            )


    return True


# ============================================================
# 7. SYNTHESIS PROMPT
# ============================================================

system_prompt = """
You are the scientific architect of a 50-agent
cross-science research swarm.

Fifty independent scientific agents have proposed ways of
understanding latent financial market regimes using ideas
transferred from statistical physics, dynamical systems,
signal engineering, control theory, information theory,
applied mathematics, chemistry, astrophysics, network
science, and geophysics.

Your task is SYNTHESIS, not voting.

Look for mathematical structures that recur independently
across disciplines.

Preserve valuable minority ideas when they add something
different from the dominant consensus.

Prefer a parsimonious architecture.

The final architecture must be:

1. mathematically interpretable;
2. empirically implementable;
3. falsifiable;
4. auditable;
5. robust enough to compare competing latent-state methods.

Do not invent empirical algorithms outside the explicitly
allowed list.

Keep the response concise.

Extended deliberation is unnecessary. Respond directly once
you have identified the major cross-disciplinary structure.

This is educational quantitative research and not
investment advice.
"""


user_prompt = f"""
SYNTHESIZE THE 50-AGENT SWARM.

The objective is to construct ONE empirical architecture for
discovering latent market regimes.

------------------------------------------------------------
AGGREGATE DISCIPLINARY SUMMARY
------------------------------------------------------------

{json.dumps(
    family_summary_for_prompt,
    ensure_ascii=False
)}

------------------------------------------------------------
50 SCIENTIFIC PROPOSALS
------------------------------------------------------------

{json.dumps(
    compact,
    ensure_ascii=False,
    separators=(",", ":")
)}

------------------------------------------------------------
DESIGN REQUIREMENTS
------------------------------------------------------------

CORE MOTIFS

Identify 3–8 mathematical motifs that recur across otherwise
different scientific disciplines.

MINORITY IDEAS

Preserve 2–6 scientifically interesting ideas that may not
dominate the swarm but add genuinely distinct information.

REPRESENTATION LAYER

Explain how the high-dimensional observable market system
should be represented before latent-state inference.

STATE-INFERENCE LAYER

Explain how hidden market states should be inferred.

TRANSITION LAYER

Explain how movement between latent states should be
identified independently of simply changing cluster labels.

UNCERTAINTY LAYER

Explain how disagreement across different clustering
architectures should be interpreted.

TARGET STATES

Choose an initial number of latent regimes between 2 and 8.

The number is an empirical starting hypothesis, not an
ontological claim about the true number of market regimes.

EMPIRICAL COMPONENTS

Select 3–6 components ONLY from:

PCA
GaussianMixture
KMeans
SpectralClustering
TransitionIntensity
ConsensusVoting

FALSIFICATION

Provide 3–7 empirical conditions under which the proposed
architecture should be rejected or substantially revised.

RESEARCH RATIONALE

Explain the synthesis in approximately 100–180 words.

Return only the required structured output.
"""


# ============================================================
# 8. SYNTHESIS API CALL WITH CONTROLLED RETRIES
# ============================================================

SYNTH_INITIAL_TOKENS = 3000

SYNTH_RETRY_TOKENS = 5000

SYNTH_FINAL_TOKENS = 8000


def call_synthesis(
    max_tokens=SYNTH_INITIAL_TOKENS
):

    print(
        f"\nCalling synthesis architect "
        f"(max_tokens={max_tokens}, effort={EFFORT})..."
    )


    response = client.messages.create(

        model=MODEL,

        max_tokens=max_tokens,

        system=system_prompt,

        messages=[

            {
                "role": "user",
                "content": user_prompt
            }

        ],

        output_config={

            "effort":
                EFFORT,

            "format": {

                "type":
                    "json_schema",

                "schema":
                    SYNTH_SCHEMA
            }
        }
    )


    # --------------------------------------------------------
    # Truncation handling
    # --------------------------------------------------------

    if response.stop_reason == "max_tokens":

        if max_tokens < SYNTH_RETRY_TOKENS:

            print(
                "Synthesis truncated. "
                "Retrying with 5000 tokens..."
            )

            return call_synthesis(
                SYNTH_RETRY_TOKENS
            )


        elif max_tokens < SYNTH_FINAL_TOKENS:

            print(
                "Synthesis still truncated. "
                "Final retry with 8000 tokens..."
            )

            return call_synthesis(
                SYNTH_FINAL_TOKENS
            )


        raise RuntimeError(
            "Synthesis could not complete within "
            f"{SYNTH_FINAL_TOKENS} tokens."
        )


    # --------------------------------------------------------
    # Refusal / abnormal termination
    # --------------------------------------------------------

    if response.stop_reason == "refusal":

        raise RuntimeError(
            "Claude returned a refusal for the synthesis."
        )


    # --------------------------------------------------------
    # Extract TEXT blocks by type
    # --------------------------------------------------------

    text_blocks = [

        block.text

        for block in response.content

        if getattr(
            block,
            "type",
            None
        ) == "text"
    ]


    if not text_blocks:

        raise RuntimeError(
            "Synthesis returned no text block. "
            f"stop_reason={response.stop_reason}"
        )


    raw_text = "".join(
        text_blocks
    )


    # --------------------------------------------------------
    # Parse structured JSON
    # --------------------------------------------------------

    try:

        obj = json.loads(
            raw_text
        )

    except json.JSONDecodeError as exc:

        print(
            "\nRAW RESPONSE PREVIEW:\n"
        )

        print(
            raw_text[:2000]
        )

        raise RuntimeError(
            "Synthesis response was not valid JSON."
        ) from exc


    # --------------------------------------------------------
    # Local scientific validation
    # --------------------------------------------------------

    validate_synthesis(
        obj
    )


    # --------------------------------------------------------
    # Provenance
    # --------------------------------------------------------

    obj["_request_id"] = getattr(
        response,
        "_request_id",
        None
    )


    obj["_stop_reason"] = (
        response.stop_reason
    )


    obj["_token_ceiling_used"] = (
        max_tokens
    )


    return obj


# ============================================================
# 9. RUN THE SYNTHESIS
# ============================================================

synthesis = call_synthesis()


# ============================================================
# 10. SAVE SYNTHESIS
# ============================================================

synthesis_path = (
    OUT /
    "swarm_synthesis.json"
)


synthesis_path.write_text(

    json.dumps(
        synthesis,
        indent=2,
        ensure_ascii=False
    ),

    encoding="utf-8"
)


# ============================================================
# 11. SYNTHESIS REPORT
# ============================================================

print("\n" + "=" * 72)

print(
    "SWARM SYNTHESIS COMPLETE"
)

print("=" * 72)


print(
    f"\nArchitecture : "
    f"{synthesis['architecture_name']}"
)

print(
    f"Target states: "
    f"{synthesis['target_states']}"
)

print(
    f"Token ceiling used: "
    f"{synthesis['_token_ceiling_used']}"
)


print(
    "\nEMPIRICAL COMPONENTS"
)

for component in synthesis[
    "empirical_components"
]:

    print(
        f"  • {component}"
    )


print(
    "\nCORE CROSS-SCIENCE MOTIFS"
)

for motif in synthesis[
    "core_motifs"
]:

    print(
        f"  • {motif}"
    )


print(
    "\nMINORITY IDEAS PRESERVED"
)

for idea in synthesis[
    "minority_ideas_to_preserve"
]:

    print(
        f"  • {idea}"
    )


print(
    "\nARCHITECTURE"
)

print(
    "\nRepresentation layer:"
)

print(
    synthesis[
        "representation_layer"
    ]
)


print(
    "\nState-inference layer:"
)

print(
    synthesis[
        "state_inference_layer"
    ]
)


print(
    "\nTransition layer:"
)

print(
    synthesis[
        "transition_layer"
    ]
)


print(
    "\nUncertainty layer:"
)

print(
    synthesis[
        "uncertainty_layer"
    ]
)


print(
    "\nFALSIFICATION CRITERIA"
)

for criterion in synthesis[
    "falsification_criteria"
]:

    print(
        f"  • {criterion}"
    )


print(
    "\nRESEARCH RATIONALE"
)

print(
    synthesis[
        "research_rationale"
    ]
)


print(
    "\nSaved to:"
)

print(
    synthesis_path
)


print("\n" + "=" * 72)

print(
    "CELL 6 COMPLETE — READY FOR EMPIRICAL REGIME INFERENCE"
)

print("=" * 72)
```

### CODE UNIT 7 — Executable ensemble for latent-state inference

CODE UNIT 7 converts the swarm’s conceptual synthesis into a controlled ensemble of empirical regime views. The purpose is not to pretend that PCA, Gaussian mixtures, K-means and spectral clustering literally implement every sophisticated proposal. They serve as transparent representatives of several recurring scientific ideas: low-dimensional order parameters, probabilistic metastable states, geometric basins and graph-like neighborhood structure.

PCA extracts dominant collective coordinates from the market feature system. In physics language these can be interpreted as candidate order parameters; in signal processing they are dominant modes; in control language they are a reduced state representation. The first components are then passed to multiple unsupervised estimators. Gaussian mixtures provide soft probabilistic membership, K-means supplies a geometric partition, and spectral clustering offers a nonlinear neighborhood view. The target number of states is taken from the swarm synthesis but is checked against nearby alternatives.

Because cluster labels are arbitrary, the code aligns model labels to a Gaussian-mixture reference using the Hungarian assignment algorithm. Consensus is then the modal aligned label, while disagreement is the fraction of models that do not support that label. This gives a simple but important uncertainty measure: the system can say not only which state is most supported but how much the scientific views disagree.

The state count is not chosen by a single score. For each candidate K, the code records silhouette separation and average state persistence. Extremely fragmented partitions may separate points geometrically yet fail the economic meaning of a regime. The selected K therefore remains inspectable, and the diagnostics are exported for later challenge.


```python
# ============================================================
# CODE CELL 7
# EMPIRICAL LATENT-STATE INFERENCE
# PCA + GMM + KMEANS + SPECTRAL + CONSENSUS
# ============================================================

# ------------------------------------------------------------
# IMPORTANT
# ------------------------------------------------------------
#
# Cell 1 deliberately did NOT upgrade or reinstall NumPy,
# SciPy or scikit-learn.
#
# We import the scientific tools here, where they are needed,
# using the versions already provided by the Colab runtime.
# ------------------------------------------------------------

from sklearn.decomposition import PCA
from sklearn.mixture import GaussianMixture
from sklearn.cluster import KMeans, SpectralClustering
from sklearn.metrics import silhouette_score
from scipy.optimize import linear_sum_assignment

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. PRECONDITIONS
# ============================================================

if "X" not in globals():

    raise RuntimeError(
        "Standardized feature matrix X is missing. "
        "Run Cell 2 first."
    )


if "synthesis" not in globals():

    raise RuntimeError(
        "Swarm synthesis is missing. "
        "Run Cell 6 first."
    )


if len(X) < 250:

    raise RuntimeError(
        "Too few observations for reliable "
        "latent-state inference."
    )


print("=" * 72)

print(
    "CELL 7 — EMPIRICAL LATENT-STATE INFERENCE"
)

print("=" * 72)


print(
    f"\nObservations        : {len(X):,}"
)

print(
    f"Observed features   : {X.shape[1]}"
)

print(
    f"Architecture        : "
    f"{synthesis['architecture_name']}"
)


# ============================================================
# 2. TARGET NUMBER OF REGIMES
# ============================================================
#
# Cell 6 provides the swarm's initial hypothesis for K.
#
# We do NOT treat K as an unquestionable truth.
# We test nearby alternatives below.
# ============================================================

K = int(
    synthesis["target_states"]
)


if not (
    2 <= K <= 8
):

    raise ValueError(
        f"Invalid target state count from Cell 6: {K}"
    )


print(
    f"Target regimes (K) : {K}"
)


# ============================================================
# 3. PCA REPRESENTATION LAYER
# ============================================================
#
# The observable market system may contain correlated and
# redundant features.
#
# PCA provides a lower-dimensional representation before
# competing state-inference algorithms are applied.
# ============================================================

MAX_PCS = min(
    6,
    X.shape[1],
    len(X)
)


pca_full = PCA(
    n_components=MAX_PCS
)


Z_full = pca_full.fit_transform(
    X.values
)


pc_columns = [

    f"PC{i+1}"

    for i in range(
        MAX_PCS
    )
]


Zdf = pd.DataFrame(

    Z_full,

    index=X.index,

    columns=pc_columns
)


# ------------------------------------------------------------
# Clustering representation
# ------------------------------------------------------------

N_CLUSTER_PCS = min(
    4,
    MAX_PCS
)


Z = Zdf.iloc[
    :,
    :N_CLUSTER_PCS
].values


explained = (
    pca_full
    .explained_variance_ratio_
)


cumulative_explained = (
    np.cumsum(
        explained
    )
)


print(
    f"\nPCA dimensions used for clustering: "
    f"{N_CLUSTER_PCS}"
)

print(
    f"Cumulative variance explained: "
    f"{cumulative_explained[N_CLUSTER_PCS - 1]:.3f}"
)


# ============================================================
# 4. PCA DIAGNOSTIC TABLE
# ============================================================

pca_diagnostic = pd.DataFrame({

    "component":
        pc_columns,

    "variance_explained":
        explained,

    "cumulative_variance":
        cumulative_explained
})


display(
    pca_diagnostic.round(4)
)


# ============================================================
# 5. REGIME PERSISTENCE FUNCTION
# ============================================================

def regime_persistence(labels):

    labels = np.asarray(
        labels
    )


    if len(labels) < 2:

        return np.nan


    return float(
        np.mean(
            labels[1:]
            ==
            labels[:-1]
        )
    )


# ============================================================
# 6. TEST NEARBY VALUES OF K
# ============================================================
#
# The swarm proposes K.
#
# We inspect nearby state counts rather than blindly accepting
# a single number.
#
# Diagnostics:
#
#     GMM BIC
#     silhouette score
#     persistence
#
# ============================================================

k_min = max(
    2,
    K - 1
)


k_max = min(
    8,
    K + 1
)


k_grid = list(
    range(
        k_min,
        k_max + 1
    )
)


diagnostic_rows = []


for k_test in k_grid:

    gmm_test = GaussianMixture(

        n_components=k_test,

        covariance_type="full",

        reg_covar=1e-5,

        n_init=5,

        max_iter=500,

        random_state=SEED
    )


    labels_test = (
        gmm_test
        .fit_predict(Z)
    )


    if len(
        np.unique(
            labels_test
        )
    ) > 1:

        sil = silhouette_score(
            Z,
            labels_test
        )

    else:

        sil = np.nan


    diagnostic_rows.append({

        "K":
            k_test,

        "GMM_BIC":
            gmm_test.bic(Z),

        "silhouette":
            sil,

        "persistence":
            regime_persistence(
                labels_test
            )
    })


k_diagnostics = pd.DataFrame(
    diagnostic_rows
)


print(
    "\nNEARBY STATE-COUNT DIAGNOSTICS"
)


display(
    k_diagnostics.round(4)
)


# ============================================================
# 7. GAUSSIAN MIXTURE MODEL
# ============================================================

gmm = GaussianMixture(

    n_components=K,

    covariance_type="full",

    reg_covar=1e-5,

    n_init=10,

    max_iter=750,

    random_state=SEED
)


gmm_labels = (
    gmm
    .fit_predict(Z)
)


gmm_probabilities = (
    gmm
    .predict_proba(Z)
)


gmm_confidence = (
    gmm_probabilities.max(
        axis=1
    )
)


if len(
    np.unique(
        gmm_labels
    )
) > 1:

    gmm_silhouette = (
        silhouette_score(
            Z,
            gmm_labels
        )
    )

else:

    gmm_silhouette = np.nan


print(
    f"\nGMM silhouette      : "
    f"{gmm_silhouette:.3f}"
)

print(
    f"GMM persistence     : "
    f"{regime_persistence(gmm_labels):.3f}"
)

print(
    f"GMM mean confidence : "
    f"{np.mean(gmm_confidence):.3f}"
)


# ============================================================
# 8. K-MEANS
# ============================================================

kmeans = KMeans(

    n_clusters=K,

    n_init=20,

    max_iter=500,

    random_state=SEED
)


kmeans_labels = (
    kmeans
    .fit_predict(Z)
)


if len(
    np.unique(
        kmeans_labels
    )
) > 1:

    kmeans_silhouette = (
        silhouette_score(
            Z,
            kmeans_labels
        )
    )

else:

    kmeans_silhouette = np.nan


print(
    f"\nKMeans silhouette   : "
    f"{kmeans_silhouette:.3f}"
)

print(
    f"KMeans persistence  : "
    f"{regime_persistence(kmeans_labels):.3f}"
)


# ============================================================
# 9. SPECTRAL CLUSTERING
# ============================================================
#
# Spectral clustering becomes expensive on very long samples.
#
# Therefore it is applied to the most recent observations.
#
# Earlier observations are deliberately left missing for this
# model rather than inventing spectral labels.
# ============================================================

SPECTRAL_MAX_OBS = min(
    2500,
    len(Z)
)


spectral_start = (
    len(Z)
    -
    SPECTRAL_MAX_OBS
)


Z_spectral = Z[
    spectral_start:
]


spectral = SpectralClustering(

    n_clusters=K,

    affinity="nearest_neighbors",

    n_neighbors=min(
        20,
        max(
            5,
            len(Z_spectral) - 1
        )
    ),

    assign_labels="kmeans",

    n_init=10,

    random_state=SEED
)


spectral_recent_labels = (
    spectral
    .fit_predict(
        Z_spectral
    )
)


print(
    f"\nSpectral observations: "
    f"{len(Z_spectral):,}"
)

print(
    f"Spectral persistence : "
    f"{regime_persistence(spectral_recent_labels):.3f}"
)


# ============================================================
# 10. LABEL ALIGNMENT
# ============================================================
#
# Cluster labels are arbitrary.
#
# GMM state 0 does not automatically mean the same thing as
# KMeans state 0.
#
# We therefore align competing methods to the GMM reference
# using maximum-overlap assignment.
# ============================================================

def align_to_reference(
    reference,
    candidate,
    K
):

    reference = np.asarray(
        reference,
        dtype=int
    )


    candidate = np.asarray(
        candidate,
        dtype=int
    )


    if len(reference) != len(candidate):

        raise ValueError(
            "Reference and candidate label arrays "
            "must have equal length."
        )


    overlap = np.zeros(
        (K, K),
        dtype=int
    )


    for ref_state in range(K):

        for cand_state in range(K):

            overlap[
                ref_state,
                cand_state
            ] = np.sum(

                (reference == ref_state)
                &
                (candidate == cand_state)
            )


    row_ind, col_ind = (
        linear_sum_assignment(
            -overlap
        )
    )


    mapping = {

        int(cand):
            int(ref)

        for ref, cand in zip(
            row_ind,
            col_ind
        )
    }


    aligned = np.array([

        mapping.get(
            int(label),
            int(label)
        )

        for label in candidate

    ])


    return (
        aligned,
        mapping,
        overlap
    )


# ============================================================
# 11. ALIGN K-MEANS TO GMM
# ============================================================

kmeans_aligned, kmeans_mapping, kmeans_overlap = (
    align_to_reference(

        gmm_labels,

        kmeans_labels,

        K
    )
)


# ============================================================
# 12. ALIGN SPECTRAL TO GMM
# ============================================================
#
# Only the overlapping recent sample is used.
# ============================================================

gmm_recent = (
    gmm_labels[
        spectral_start:
    ]
)


spectral_aligned_recent, spectral_mapping, spectral_overlap = (
    align_to_reference(

        gmm_recent,

        spectral_recent_labels,

        K
    )
)


# ============================================================
# 13. LABEL DATAFRAME
# ============================================================

labdf = pd.DataFrame(

    index=X.index
)


labdf[
    "GMM"
] = gmm_labels


labdf[
    "KMeans"
] = kmeans_aligned


labdf[
    "Spectral"
] = np.nan


labdf.loc[

    X.index[
        spectral_start:
    ],

    "Spectral"

] = spectral_aligned_recent


# ============================================================
# 14. CONSENSUS FUNCTION
# ============================================================
#
# Consensus is calculated from whatever models are available
# at each observation.
#
# Earlier observations:
#
#     GMM + KMeans
#
# Recent observations:
#
#     GMM + KMeans + Spectral
#
# Disagreement is:
#
#     1 - fraction voting for winning state
#
# ============================================================

def consensus_vote(row):

    votes = (
        row
        .dropna()
        .astype(int)
        .tolist()
    )


    if len(votes) == 0:

        return pd.Series({

            "regime":
                np.nan,

            "disagreement":
                np.nan,

            "models_available":
                0
        })


    counts = pd.Series(
        votes
    ).value_counts()


    winner = int(
        counts.index[0]
    )


    winner_votes = int(
        counts.iloc[0]
    )


    disagreement = (
        1.0
        -
        winner_votes
        /
        len(votes)
    )


    return pd.Series({

        "regime":
            winner,

        "disagreement":
            disagreement,

        "models_available":
            len(votes)
    })


consensus = labdf.apply(

    consensus_vote,

    axis=1
)


labdf[
    "regime"
] = consensus[
    "regime"
].astype(int)


labdf[
    "disagreement"
] = consensus[
    "disagreement"
]


labdf[
    "models_available"
] = consensus[
    "models_available"
].astype(int)


# ============================================================
# 15. CONSENSUS PERSISTENCE
# ============================================================

consensus_persistence = (
    regime_persistence(
        labdf[
            "regime"
        ].values
    )
)


mean_disagreement = float(

    labdf[
        "disagreement"
    ].mean()
)


print(
    f"\nConsensus persistence : "
    f"{consensus_persistence:.3f}"
)

print(
    f"Mean disagreement      : "
    f"{mean_disagreement:.3f}"
)


# ============================================================
# 16. BUILD MASTER STATE DATAFRAME
# ============================================================

state = pd.concat(

    [

        F.loc[
            X.index
        ],

        Zdf,

        labdf

    ],

    axis=1
)


state[
    "gmm_confidence"
] = gmm_confidence


# ============================================================
# 17. REGIME FREQUENCIES
# ============================================================

regime_frequency = (

    state[
        "regime"
    ]

    .value_counts(
        normalize=True
    )

    .sort_index()

    .rename(
        "fraction"
    )

    .to_frame()
)


regime_frequency[
    "percent"
] = (

    100
    *
    regime_frequency[
        "fraction"
    ]
)


print(
    "\nCONSENSUS REGIME FREQUENCIES"
)


display(
    regime_frequency.round(3)
)


# ============================================================
# 18. FEATURE PROFILES BY REGIME
# ============================================================

feature_profiles_cell7 = (

    state

    .groupby(
        "regime"
    )[
        list(
            X.columns
        )
    ]

    .median()
)


print(
    "\nMEDIAN FEATURE PROFILE BY REGIME"
)


display(
    feature_profiles_cell7.round(3)
)


# ============================================================
# 19. METHOD AGREEMENT MATRIX
# ============================================================

agreement = pd.DataFrame(

    index=[
        "GMM",
        "KMeans",
        "Spectral"
    ],

    columns=[
        "GMM",
        "KMeans",
        "Spectral"
    ],

    dtype=float
)


methods = [
    "GMM",
    "KMeans",
    "Spectral"
]


for m1 in methods:

    for m2 in methods:

        valid = (
            labdf[m1].notna()
            &
            labdf[m2].notna()
        )


        if valid.sum() == 0:

            agreement.loc[
                m1,
                m2
            ] = np.nan

        else:

            agreement.loc[
                m1,
                m2
            ] = np.mean(

                labdf.loc[
                    valid,
                    m1
                ].astype(int).values

                ==

                labdf.loc[
                    valid,
                    m2
                ].astype(int).values
            )


print(
    "\nALIGNED METHOD AGREEMENT"
)


display(
    agreement.round(3)
)


# ============================================================
# 20. VISUAL DIAGNOSTIC:
# PCA STATE SPACE
# ============================================================

plt.figure(
    figsize=(10, 6)
)


scatter = plt.scatter(

    Zdf["PC1"],

    Zdf["PC2"],

    c=state["regime"],

    s=8,

    alpha=0.55
)


plt.xlabel(
    "Principal Component 1"
)

plt.ylabel(
    "Principal Component 2"
)

plt.title(
    "Consensus Latent Market States in PCA Space"
)

plt.grid(
    alpha=0.2
)

plt.show()


# ============================================================
# 21. VISUAL DIAGNOSTIC:
# REGIME THROUGH TIME
# ============================================================

plt.figure(
    figsize=(14, 4)
)


plt.scatter(

    state.index,

    state["regime"],

    c=state["regime"],

    s=8
)


plt.yticks(
    range(K)
)

plt.xlabel(
    "Date"
)

plt.ylabel(
    "Consensus regime"
)

plt.title(
    "Inferred Latent Market Regime Through Time"
)

plt.grid(
    alpha=0.2
)

plt.show()


# ============================================================
# 22. SAVE CELL 7 OUTPUTS
# ============================================================

state.to_csv(

    OUT /
    "latent_state_history.csv"
)


k_diagnostics.to_csv(

    OUT /
    "state_count_diagnostics.csv",

    index=False
)


regime_frequency.to_csv(

    OUT /
    "regime_frequency.csv"
)


feature_profiles_cell7.to_csv(

    OUT /
    "regime_feature_profiles_cell7.csv"
)


agreement.to_csv(

    OUT /
    "method_agreement.csv"
)


# ============================================================
# 23. CELL 7 SUMMARY OBJECT
# ============================================================

cell7_report = {

    "architecture":
        synthesis[
            "architecture_name"
        ],

    "target_states":
        int(K),

    "observations":
        int(len(state)),

    "observed_features":
        int(X.shape[1]),

    "pca_components_computed":
        int(MAX_PCS),

    "pca_components_clustered":
        int(N_CLUSTER_PCS),

    "pca_variance_explained":
        float(
            cumulative_explained[
                N_CLUSTER_PCS - 1
            ]
        ),

    "gmm_silhouette":
        float(
            gmm_silhouette
        ),

    "gmm_persistence":
        float(
            regime_persistence(
                gmm_labels
            )
        ),

    "kmeans_silhouette":
        float(
            kmeans_silhouette
        ),

    "kmeans_persistence":
        float(
            regime_persistence(
                kmeans_labels
            )
        ),

    "consensus_persistence":
        float(
            consensus_persistence
        ),

    "mean_disagreement":
        float(
            mean_disagreement
        ),

    "spectral_observations":
        int(
            SPECTRAL_MAX_OBS
        )
}


(
    OUT /
    "cell7_regime_inference_report.json"
).write_text(

    json.dumps(
        cell7_report,
        indent=2
    ),

    encoding="utf-8"
)


# ============================================================
# 24. FINAL REPORT
# ============================================================

print("\n" + "=" * 72)

print(
    "CELL 7 COMPLETE"
)

print("=" * 72)


print(
    f"\nLatent states inferred       : {K}"
)

print(
    f"Observations classified      : {len(state):,}"
)

print(
    f"GMM silhouette               : {gmm_silhouette:.3f}"
)

print(
    f"Consensus persistence        : {consensus_persistence:.3f}"
)

print(
    f"Mean model disagreement      : {mean_disagreement:.3f}"
)

print(
    f"Spectral sample              : {SPECTRAL_MAX_OBS:,}"
)


print(
    "\nThe market now has an empirical latent-state history."
)

print(
    "Cell 8 can characterize those states and study "
    "their transition dynamics."
)

print("=" * 72)
```

### CODE UNIT 8 — Regime interpretation, transition pressure and observatory

CODE UNIT 8 turns anonymous cluster numbers into economically interpretable state profiles and separates stable-state inference from transition pressure. A cluster label has no intrinsic meaning: regime 0 is not “good” or “bad.” The code therefore computes the median standardized feature vector within each consensus state and uses those profiles to characterize the regimes. This allows the researcher to see whether a state is associated with high volatility, deep drawdown, strong breadth, defensive rotation, correlation spikes or other observable structures.

Transition intensity is constructed independently from the labels. It measures the rolling magnitude of changes in the robustly scaled feature vector. This is a deliberately simple approximation to several cross-science ideas: velocity in state space, stress accumulation, distributional change, precursor activity and movement toward a new basin. A high transition-intensity reading does not automatically force a state switch; it says that the observed system is moving unusually rapidly. This distinction is important because transition detection and state classification are different inferential tasks.

The cell also estimates an empirical transition matrix for the consensus labels and average dwell times. These quantities connect naturally to Markov and kinetic-state interpretations while remaining model-agnostic. A regime with one-day average dwell time is probably a cluster rather than a meaningful market condition. A persistent regime with distinctive features and nontrivial transition structure is more interesting.

Finally, the code creates a compact visual observatory: the S&P 500 price is shown against regime-colored backgrounds, while transition intensity and model disagreement are plotted separately. The graphics are diagnostics, not trading signals. They make it easier to inspect whether the inferred states correspond to recognizable structural episodes and whether uncertainty rises around transitions.


```python
# ============================================================
# CODE CELL 8
# REGIME CHARACTERIZATION & TRANSITION DYNAMICS
#
# PURPOSE
# ------------------------------------------------------------
# Cell 7 inferred the latent market states.
#
# Cell 8 asks:
#
#   1. What does each regime economically look like?
#   2. How persistent is each regime?
#   3. How does the market move between regimes?
#   4. Does observable state-space movement accelerate
#      around regime transitions?
#   5. Does model disagreement rise near transitions?
#
# No new clustering is performed here.
# ============================================================

import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. PRECONDITIONS
# ============================================================

required_objects = [
    "state",
    "X",
    "px",
    "K"
]

missing_objects = [
    name
    for name in required_objects
    if name not in globals()
]

if missing_objects:

    raise RuntimeError(
        "Cell 8 requires Cell 7 outputs. "
        f"Missing objects: {missing_objects}"
    )


required_state_columns = [
    "regime",
    "disagreement"
]

missing_columns = [
    col
    for col in required_state_columns
    if col not in state.columns
]

if missing_columns:

    raise RuntimeError(
        "The state dataframe is missing required columns: "
        f"{missing_columns}"
    )


if "SPY" not in px.columns:

    raise RuntimeError(
        "SPY is required for the market-regime observatory."
    )


print("=" * 72)
print("CELL 8 — REGIME CHARACTERIZATION & TRANSITION DYNAMICS")
print("=" * 72)

print(f"\nObservations        : {len(state):,}")
print(f"Latent regimes (K) : {K}")
print(f"Features            : {len(X.columns)}")


# ============================================================
# 2. CLEAN REGIME SERIES
# ============================================================

regime = (
    state["regime"]
    .dropna()
    .astype(int)
)


if len(regime) == 0:

    raise RuntimeError(
        "No valid regime observations were found."
    )


unexpected_states = (
    set(regime.unique())
    -
    set(range(K))
)

if unexpected_states:

    raise RuntimeError(
        "Unexpected regime labels found: "
        f"{sorted(unexpected_states)}"
    )


# ============================================================
# 3. FEATURE PROFILES BY REGIME
# ============================================================
#
# We characterize each latent state using the median of the
# original market features.
#
# Median profiles are preferred here because several market
# variables — volatility, VIX changes, drawdowns and credit
# stress — can be highly skewed.
# ============================================================

feature_profiles = (

    state

    .groupby("regime")[
        list(X.columns)
    ]

    .median()

    .reindex(
        range(K)
    )
)


feature_means = (

    state

    .groupby("regime")[
        list(X.columns)
    ]

    .mean()

    .reindex(
        range(K)
    )
)


regime_counts = (

    state["regime"]

    .value_counts()

    .reindex(
        range(K),
        fill_value=0
    )

    .sort_index()
)


regime_shares = (
    regime_counts
    /
    regime_counts.sum()
)


print("\n" + "=" * 72)
print("MEDIAN MARKET FEATURE PROFILE BY REGIME")
print("=" * 72)

display(
    feature_profiles.round(3)
)


# ============================================================
# 4. ECONOMIC SUMMARY TABLE
# ============================================================

summary_rows = []


for k in range(K):

    mask = (
        state["regime"] == k
    )

    row = {

        "regime":
            k,

        "observations":
            int(mask.sum()),

        "share":
            float(mask.mean()),

        "mean_disagreement":
            float(
                state.loc[
                    mask,
                    "disagreement"
                ].mean()
            )
    }


    # --------------------------------------------------------
    # Include economically interpretable variables when they
    # exist in the feature set.
    # --------------------------------------------------------

    useful_features = [

        "spy_ret_5d",

        "spy_trend_63d",

        "spy_vol_21d",

        "spy_downvol_63d",

        "spy_drawdown",

        "cross_asset_dispersion",

        "avg_corr_63d",

        "breadth_63d",

        "vix_level",

        "vix_change_21d",

        "credit_risk_21d",

        "bond_equity_rotation_21d",

        "gold_equity_rotation_21d"
    ]


    for feature in useful_features:

        if feature in state.columns:

            row[feature] = float(

                state.loc[
                    mask,
                    feature
                ].median()
            )


    summary_rows.append(
        row
    )


regime_summary = pd.DataFrame(
    summary_rows
)


print("\n" + "=" * 72)
print("ECONOMIC REGIME SUMMARY")
print("=" * 72)

display(
    regime_summary.round(4)
)


# ============================================================
# 5. TRANSITION MATRIX
# ============================================================
#
# T[i,j] estimates:
#
#       P(S_t = j | S_{t-1} = i)
#
# Diagonal values measure one-day regime persistence.
# Off-diagonal values describe observed transitions.
# ============================================================

previous_regime = (
    regime.shift(1)
)


transition_counts = pd.crosstab(

    previous_regime,

    regime
)


transition_counts = (
    transition_counts
    .reindex(
        index=range(K),
        columns=range(K),
        fill_value=0
    )
)


row_totals = (
    transition_counts
    .sum(axis=1)
    .replace(0, np.nan)
)


transition_matrix = (
    transition_counts
    .div(
        row_totals,
        axis=0
    )
    .fillna(0.0)
)


print("\n" + "=" * 72)
print("CONSENSUS TRANSITION MATRIX")
print("=" * 72)

display(
    transition_matrix.round(3)
)


# ============================================================
# 6. REGIME PERSISTENCE
# ============================================================

persistence_by_regime = pd.DataFrame({

    "regime":
        range(K),

    "one_day_persistence": [

        float(
            transition_matrix.loc[
                k,
                k
            ]
        )

        for k in range(K)
    ]
})


print("\n" + "=" * 72)
print("ONE-DAY REGIME PERSISTENCE")
print("=" * 72)

display(
    persistence_by_regime.round(4)
)


# ============================================================
# 7. DWELL-TIME ANALYSIS
# ============================================================
#
# A dwell is a consecutive sequence of observations in the
# same latent regime.
#
# This is distinct from the diagonal transition probability:
# it gives the empirical duration of complete regime episodes.
# ============================================================

regime_full = (
    state["regime"]
    .astype(int)
)


run_id = (
    regime_full
    !=
    regime_full.shift(1)
).cumsum()


dwell = (

    pd.DataFrame({

        "regime":
            regime_full,

        "run_id":
            run_id
    })

    .groupby("run_id")

    .agg(

        regime=(
            "regime",
            "first"
        ),

        length=(
            "regime",
            "size"
        ),

        start=(
            "regime",
            lambda x: x.index.min()
        ),

        end=(
            "regime",
            lambda x: x.index.max()
        )
    )
)


dwell_summary = (

    dwell

    .groupby("regime")[
        "length"
    ]

    .agg(
        [
            "count",
            "mean",
            "median",
            "max"
        ]
    )

    .reindex(
        range(K)
    )
)


dwell_summary.columns = [

    "episodes",

    "mean_days",

    "median_days",

    "max_days"
]


print("\n" + "=" * 72)
print("REGIME DWELL TIMES")
print("=" * 72)

display(
    dwell_summary.round(2)
)


# ============================================================
# 8. STATE-SPACE VELOCITY
# ============================================================
#
# Regime labels are discrete.
#
# But the observable market system moves continuously through
# feature space.
#
# We therefore construct:
#
#       velocity_t =
#       sqrt(mean_j[(X_j,t - X_j,t-1)^2])
#
# using standardized features.
#
# This is a model-independent measure of how rapidly the
# observable market state is changing.
# ============================================================

dX = (
    X
    .diff()
)


state_space_velocity = (

    np.sqrt(
        np.mean(
            np.square(dX),
            axis=1
        )
    )
)


state_space_velocity = pd.Series(

    state_space_velocity,

    index=X.index,

    name="state_space_velocity"
)


# ============================================================
# 9. TRANSITION INTENSITY
# ============================================================
#
# Smooth short-term state-space movement.
#
# The purpose is not to predict a transition automatically,
# but to measure pressure or instability in the observable
# system.
# ============================================================

TRANSITION_WINDOW = 10


transition_intensity = (

    state_space_velocity

    .rolling(
        TRANSITION_WINDOW,
        min_periods=3
    )

    .mean()
)


transition_intensity.name = (
    "transition_intensity"
)


# ============================================================
# 10. ROLLING ROBUST NORMALIZATION
# ============================================================
#
# A raw velocity of 0.5 may be extreme in one historical
# period and ordinary in another.
#
# We therefore normalize transition intensity relative to a
# trailing one-year window.
# ============================================================

NORMALIZATION_WINDOW = 252


rolling_median = (

    transition_intensity

    .rolling(
        NORMALIZATION_WINDOW,
        min_periods=60
    )

    .median()
)


rolling_std = (

    transition_intensity

    .rolling(
        NORMALIZATION_WINDOW,
        min_periods=60
    )

    .std()
)


transition_z = (

    (
        transition_intensity
        -
        rolling_median
    )

    /
    (
        rolling_std
        +
        1e-9
    )
)


transition_z.name = (
    "transition_z"
)


# ============================================================
# 11. ADD DYNAMICS TO MASTER STATE DATAFRAME
# ============================================================

state[
    "state_space_velocity"
] = state_space_velocity


state[
    "transition_intensity"
] = transition_intensity


state[
    "transition_z"
] = transition_z


# ============================================================
# 12. IDENTIFY ACTUAL REGIME SWITCHES
# ============================================================

state[
    "regime_switch"
] = (

    state["regime"]
    !=
    state["regime"].shift(1)
)


# First observation is not a genuine transition.

state.iloc[
    0,
    state.columns.get_loc(
        "regime_switch"
    )
] = False


state[
    "regime_switch"
] = (

    state[
        "regime_switch"
    ]

    .astype(bool)
)


n_switches = int(

    state[
        "regime_switch"
    ].sum()
)


print(
    f"\nObserved consensus regime switches: "
    f"{n_switches:,}"
)


# ============================================================
# 13. TRANSITION INTENSITY:
# SWITCH DAYS VS NON-SWITCH DAYS
# ============================================================

switch_intensity = (

    state.loc[
        state["regime_switch"],
        "transition_intensity"
    ]

    .dropna()
)


quiet_intensity = (

    state.loc[
        ~state["regime_switch"],
        "transition_intensity"
    ]

    .dropna()
)


if (
    len(switch_intensity) > 0
    and
    len(quiet_intensity) > 0
):

    switch_mean = float(
        switch_intensity.mean()
    )

    quiet_mean = float(
        quiet_intensity.mean()
    )


    transition_intensity_ratio = (

        switch_mean
        /
        (
            quiet_mean
            +
            1e-12
        )
    )

else:

    switch_mean = np.nan

    quiet_mean = np.nan

    transition_intensity_ratio = np.nan


transition_test = pd.DataFrame({

    "group": [
        "Regime-switch observations",
        "Non-switch observations"
    ],

    "n": [
        len(switch_intensity),
        len(quiet_intensity)
    ],

    "mean_transition_intensity": [
        switch_mean,
        quiet_mean
    ],

    "median_transition_intensity": [

        float(
            switch_intensity.median()
        )
        if len(switch_intensity)
        else np.nan,

        float(
            quiet_intensity.median()
        )
        if len(quiet_intensity)
        else np.nan
    ]
})


print("\n" + "=" * 72)
print("TRANSITION-INTENSITY TEST")
print("=" * 72)

display(
    transition_test.round(4)
)


print(
    f"\nSwitch / non-switch intensity ratio: "
    f"{transition_intensity_ratio:.3f}"
)


# ============================================================
# 14. MODEL DISAGREEMENT AROUND TRANSITIONS
# ============================================================
#
# The ensemble may begin to disagree before or during periods
# in which the inferred market state is changing.
#
# This gives disagreement an epistemic interpretation:
#
#     high disagreement =
#     lower confidence in a unique latent-state assignment.
# ============================================================

switch_disagreement = (

    state.loc[
        state["regime_switch"],
        "disagreement"
    ]

    .dropna()
)


quiet_disagreement = (

    state.loc[
        ~state["regime_switch"],
        "disagreement"
    ]

    .dropna()
)


if (
    len(switch_disagreement) > 0
    and
    len(quiet_disagreement) > 0
):

    disagreement_ratio = (

        switch_disagreement.mean()

        /
        (
            quiet_disagreement.mean()
            +
            1e-12
        )
    )

else:

    disagreement_ratio = np.nan


disagreement_test = pd.DataFrame({

    "group": [
        "Regime-switch observations",
        "Non-switch observations"
    ],

    "mean_disagreement": [

        switch_disagreement.mean()
        if len(switch_disagreement)
        else np.nan,

        quiet_disagreement.mean()
        if len(quiet_disagreement)
        else np.nan
    ],

    "median_disagreement": [

        switch_disagreement.median()
        if len(switch_disagreement)
        else np.nan,

        quiet_disagreement.median()
        if len(quiet_disagreement)
        else np.nan
    ]
})


print("\n" + "=" * 72)
print("MODEL DISAGREEMENT AROUND REGIME CHANGES")
print("=" * 72)

display(
    disagreement_test.round(4)
)


# ============================================================
# 15. HIGH TRANSITION-PRESSURE OBSERVATIONS
# ============================================================

TRANSITION_Z_THRESHOLD = 2.0


state[
    "high_transition_pressure"
] = (

    state[
        "transition_z"
    ]
    >
    TRANSITION_Z_THRESHOLD
)


high_pressure_share = float(

    state[
        "high_transition_pressure"
    ].mean()
)


print(
    f"\nShare of observations with "
    f"transition_z > {TRANSITION_Z_THRESHOLD}: "
    f"{100 * high_pressure_share:.2f}%"
)


# ============================================================
# 16. SPY MARKET SERIES
# ============================================================

spy = (

    px["SPY"]

    .reindex(
        state.index
    )

    .ffill()
)


state[
    "SPY"
] = spy


# ============================================================
# 17. MARKET REGIME OBSERVATORY
# ============================================================
#
# Each observation is colored by the inferred consensus
# latent state.
#
# The plot does not assign semantic names such as "crisis" or
# "expansion" automatically. Economic interpretation should
# come from the regime profiles, not from arbitrary labels.
# ============================================================

plt.figure(
    figsize=(15, 6)
)


for k in range(K):

    mask = (
        state["regime"] == k
    )


    plt.scatter(

        state.index[
            mask
        ],

        state.loc[
            mask,
            "SPY"
        ],

        s=9,

        label=f"Regime {k}"
    )


plt.yscale(
    "log"
)


plt.xlabel(
    "Date"
)


plt.ylabel(
    "SPY — log scale"
)


plt.title(
    "Market Regime Observatory — Consensus Latent States"
)


plt.legend(
    ncol=min(
        K,
        4
    )
)


plt.grid(
    alpha=0.2
)


plt.show()


# ============================================================
# 18. TRANSITION-PRESSURE OBSERVATORY
# ============================================================

plt.figure(
    figsize=(15, 5)
)


plt.plot(

    state.index,

    state[
        "transition_z"
    ],

    linewidth=1.0,

    label="Transition pressure"
)


plt.axhline(

    TRANSITION_Z_THRESHOLD,

    linestyle="--",

    linewidth=1.0,

    label=f"Threshold = {TRANSITION_Z_THRESHOLD}"
)


plt.axhline(

    0,

    linewidth=0.8
)


plt.xlabel(
    "Date"
)


plt.ylabel(
    "Rolling normalized transition intensity"
)


plt.title(
    "Observable Market State-Space Transition Pressure"
)


plt.legend()


plt.grid(
    alpha=0.2
)


plt.show()


# ============================================================
# 19. MODEL-DISAGREEMENT OBSERVATORY
# ============================================================

plt.figure(
    figsize=(15, 4)
)


plt.plot(

    state.index,

    state[
        "disagreement"
    ],

    linewidth=1.0
)


plt.xlabel(
    "Date"
)


plt.ylabel(
    "Model disagreement"
)


plt.title(
    "Epistemic Uncertainty Across Latent-State Models"
)


plt.grid(
    alpha=0.2
)


plt.show()


# ============================================================
# 20. JOINT TRANSITION DIAGNOSTIC
# ============================================================
#
# Transition pressure:
#     movement in observable market feature space.
#
# Disagreement:
#     uncertainty across state-inference models.
#
# They are conceptually different and should not be collapsed
# into one number.
# ============================================================

joint_transition_diagnostic = (

    state[
        [
            "regime",
            "regime_switch",
            "transition_intensity",
            "transition_z",
            "disagreement",
            "high_transition_pressure"
        ]
    ]

    .copy()
)


print("\n" + "=" * 72)
print("MOST EXTREME TRANSITION-PRESSURE OBSERVATIONS")
print("=" * 72)


display(

    joint_transition_diagnostic

    .sort_values(
        "transition_z",
        ascending=False
    )

    .head(20)

    .round(4)
)


# ============================================================
# 21. TRANSITION EVENT TABLE
# ============================================================
#
# Record every consensus state change.
# ============================================================

transition_events = (

    state.loc[
        state["regime_switch"],
        [
            "regime",
            "transition_intensity",
            "transition_z",
            "disagreement",
            "SPY"
        ]
    ]

    .copy()
)


transition_events[
    "previous_regime"
] = (

    state[
        "regime"
    ]

    .shift(1)

    .reindex(
        transition_events.index
    )
)


transition_events = transition_events[
    [
        "previous_regime",
        "regime",
        "transition_intensity",
        "transition_z",
        "disagreement",
        "SPY"
    ]
]


transition_events = transition_events.rename(

    columns={
        "regime":
            "new_regime"
    }
)


print("\n" + "=" * 72)
print("RECENT REGIME TRANSITIONS")
print("=" * 72)


display(

    transition_events
    .tail(20)
    .round(4)
)


# ============================================================
# 22. SAVE OUTPUT TABLES
# ============================================================

feature_profiles.to_csv(

    OUT /
    "regime_feature_profiles.csv"
)


regime_summary.to_csv(

    OUT /
    "regime_economic_summary.csv",

    index=False
)


transition_counts.to_csv(

    OUT /
    "transition_counts.csv"
)


transition_matrix.to_csv(

    OUT /
    "transition_matrix.csv"
)


dwell.to_csv(

    OUT /
    "regime_dwell_events.csv"
)


dwell_summary.to_csv(

    OUT /
    "regime_dwell_summary.csv"
)


transition_test.to_csv(

    OUT /
    "transition_intensity_test.csv",

    index=False
)


disagreement_test.to_csv(

    OUT /
    "transition_disagreement_test.csv",

    index=False
)


transition_events.to_csv(

    OUT /
    "transition_events.csv"
)


state.to_csv(

    OUT /
    "state_history_with_dynamics.csv"
)


# ============================================================
# 23. CELL 8 REPORT
# ============================================================

cell8_report = {

    "number_of_regimes":
        int(K),

    "observations":
        int(len(state)),

    "number_of_regime_switches":
        int(n_switches),

    "consensus_persistence":
        float(
            np.mean(
                regime.values[1:]
                ==
                regime.values[:-1]
            )
        ),

    "mean_transition_intensity_switch":
        (
            float(switch_mean)
            if np.isfinite(switch_mean)
            else None
        ),

    "mean_transition_intensity_non_switch":
        (
            float(quiet_mean)
            if np.isfinite(quiet_mean)
            else None
        ),

    "transition_intensity_switch_ratio":
        (
            float(
                transition_intensity_ratio
            )
            if np.isfinite(
                transition_intensity_ratio
            )
            else None
        ),

    "mean_model_disagreement":
        float(
            state[
                "disagreement"
            ].mean()
        ),

    "disagreement_switch_ratio":
        (
            float(
                disagreement_ratio
            )
            if np.isfinite(
                disagreement_ratio
            )
            else None
        ),

    "transition_z_threshold":
        float(
            TRANSITION_Z_THRESHOLD
        ),

    "high_transition_pressure_share":
        float(
            high_pressure_share
        )
}


cell8_path = (

    OUT /
    "cell8_transition_dynamics_report.json"
)


cell8_path.write_text(

    json.dumps(
        cell8_report,
        indent=2
    ),

    encoding="utf-8"
)


# ============================================================
# 24. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 72)

print(
    "CELL 8 COMPLETE — REGIME DYNAMICS CHARACTERIZED"
)

print("=" * 72)


print(
    f"\nLatent regimes                : {K}"
)

print(
    f"Consensus regime switches     : {n_switches:,}"
)

print(
    f"Overall regime persistence    : "
    f"{cell8_report['consensus_persistence']:.3f}"
)


if np.isfinite(
    transition_intensity_ratio
):

    print(
        f"Transition intensity ratio    : "
        f"{transition_intensity_ratio:.3f}"
    )


print(
    f"Mean model disagreement       : "
    f"{cell8_report['mean_model_disagreement']:.3f}"
)

print(
    f"High transition-pressure share: "
    f"{100 * high_pressure_share:.2f}%"
)


print(
    "\nInterpretation:"
)

print(
    "Cell 7 inferred WHERE the market is in latent-state space."
)

print(
    "Cell 8 measures WHAT those states look like and HOW the "
    "system moves between them."
)

print(
    "Transition pressure and model disagreement remain separate:"
)

print(
    "the first measures movement in observable market space; "
    "the second measures uncertainty about latent-state identity."
)

print(
    "\nCell 9 can now test whether these structures survive "
    "perturbation and whether transitions have empirical content."
)

print("=" * 72)
```

### CODE UNIT 9 — Falsification, robustness and swarm-convergence diagnostics

CODE UNIT 9 is the falsification and robustness layer. Because the true regime is unobserved, validation must ask whether the inferred partition has properties that a useful latent market state should possess. The first test is conditional economic differentiation. For each regime, the code measures subsequent S&P 500 return and realized volatility over a fixed horizon. A regime representation that produces indistinguishable forward distributions may still describe the present, but it is less useful for investment and risk applications.

The second test is stability under feature perturbation. A small amount of noise is added to the standardized feature matrix, the Gaussian-mixture model is refitted, and the resulting labels are compared with the original partition using Adjusted Rand Index. Repeating this procedure creates a distribution of stability rather than one lucky score. The third test examines transition concentration: are actual regime switches associated with higher transition intensity than ordinary days? This links the state classifier to the independent change-pressure measure.

The fourth test examines scientific convergence across swarm size. Rather than making additional API calls, the cell takes prefixes of the existing fifty-agent corpus and compares their distributions of proposed state counts and mathematical vocabulary with the full swarm. This asks whether the intellectual architecture stabilizes as the population grows. It is not a proof that fifty is optimal, but it reveals whether the final synthesis depends on the last few agents.

The code produces a validation report without declaring success automatically. Thresholds are shown as diagnostics, not as a hidden optimization target. A researcher should inspect failures and revise the feature system or architecture rather than tuning until every metric looks attractive.


```python
# ============================================================
# CODE CELL 9
# VALIDATION, ROBUSTNESS & SWARM CONVERGENCE
#
# PURPOSE
# ------------------------------------------------------------
# Cells 6–8 established:
#
#   Cell 6 — Cross-science synthesis
#   Cell 7 — Empirical latent-state inference
#   Cell 8 — Regime characterization and transition dynamics
#
# Cell 9 asks whether the resulting latent-state architecture
# survives empirical scrutiny.
#
# VALIDATION LAYERS
#
#   1. Conditional forward market behavior
#   2. Cross-regime economic separation
#   3. Perturbation stability
#   4. Transition-intensity validation
#   5. Model-disagreement validation
#   6. 50-agent swarm convergence
#   7. Consolidated validation dashboard
#
# IMPORTANT
# ------------------------------------------------------------
# This is NOT a trading backtest.
#
# The objective is scientific validation of the inferred
# latent-state architecture.
# ============================================================


# ============================================================
# 0. IMPORTS
# ============================================================

import json
import re

from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.decomposition import PCA
from sklearn.mixture import GaussianMixture
from sklearn.metrics import adjusted_rand_score


# ============================================================
# 1. PRECONDITIONS
# ============================================================

required_objects = [
    "state",
    "X",
    "px",
    "K",
    "proposals",
    "synthesis"
]


missing_objects = [

    obj

    for obj in required_objects

    if obj not in globals()
]


if missing_objects:

    raise RuntimeError(
        "Cell 9 requires outputs from Cells 5–8. "
        f"Missing objects: {missing_objects}"
    )


required_state_columns = [
    "regime",
    "disagreement",
    "transition_intensity",
    "transition_z",
    "regime_switch"
]


missing_state_columns = [

    col

    for col in required_state_columns

    if col not in state.columns
]


if missing_state_columns:

    raise RuntimeError(
        "The state dataframe is missing Cell 8 outputs: "
        f"{missing_state_columns}"
    )


if "SPY" not in px.columns:

    raise RuntimeError(
        "SPY is required for forward-behavior validation."
    )


if len(proposals) != 50:

    raise RuntimeError(
        "Swarm-convergence analysis requires all 50 proposals. "
        f"Currently available: {len(proposals)}."
    )


if not (
    2 <= int(K) <= 8
):

    raise RuntimeError(
        f"Invalid number of latent states: K={K}"
    )


print("=" * 72)

print(
    "CELL 9 — VALIDATION, ROBUSTNESS & SWARM CONVERGENCE"
)

print("=" * 72)


print(
    f"\nObservations       : {len(state):,}"
)

print(
    f"Latent regimes     : {K}"
)

print(
    f"Scientific agents  : {len(proposals)}"
)

print(
    f"Architecture       : "
    f"{synthesis['architecture_name']}"
)


# ============================================================
# 2. FORWARD MARKET BEHAVIOR
# ============================================================
#
# For every observation t:
#
#     regime_t
#
# is compared with market behavior over the NEXT 21 trading
# days.
#
# We calculate:
#
#     R(t,t+H) = log(P[t+H]) - log(P[t])
#
# and forward realized volatility from:
#
#     r[t+1], r[t+2], ..., r[t+H]
#
# The regime is therefore defined at time t and the validation
# quantities occur strictly after t.
# ============================================================

FORWARD_HORIZON = 21


# ------------------------------------------------------------
# Align SPY exactly to state history
# ------------------------------------------------------------

spy = (

    px["SPY"]

    .reindex(
        state.index
    )

    .ffill()
)


if spy.isna().all():

    raise RuntimeError(
        "SPY could not be aligned to the state dataframe."
    )


log_spy = np.log(
    spy
)


daily_spy_return = (
    log_spy.diff()
)


# ============================================================
# 3. FORWARD RETURN
# ============================================================

future_return = (

    log_spy.shift(
        -FORWARD_HORIZON
    )

    -

    log_spy
)


future_return.name = (
    "future_return"
)


# ============================================================
# 4. FORWARD REALIZED VOLATILITY
# ============================================================
#
# Construct explicitly:
#
#   r[t+1]
#   r[t+2]
#   ...
#   r[t+21]
#
# This avoids accidental backward-looking contamination.
# ============================================================

future_return_matrix = pd.concat(

    [

        daily_spy_return.shift(
            -h
        )

        for h in range(
            1,
            FORWARD_HORIZON + 1
        )

    ],

    axis=1
)


future_return_matrix.columns = [

    f"r_plus_{h}"

    for h in range(
        1,
        FORWARD_HORIZON + 1
    )
]


# ------------------------------------------------------------
# Require a complete 21-day future window
# ------------------------------------------------------------

complete_forward_window = (

    future_return_matrix

    .notna()

    .sum(axis=1)

    ==
    FORWARD_HORIZON
)


# ------------------------------------------------------------
# Annualized realized volatility
# ------------------------------------------------------------

future_volatility = (

    future_return_matrix

    .std(
        axis=1,
        ddof=1
    )

    *
    np.sqrt(252)
)


future_volatility.name = (
    "future_volatility"
)


# ------------------------------------------------------------
# Mask incomplete tail observations
# ------------------------------------------------------------

future_return = (

    future_return

    .where(
        complete_forward_window
    )
)


future_volatility = (

    future_volatility

    .where(
        complete_forward_window
    )
)


# ============================================================
# 5. FORWARD VALIDATION DATAFRAME
# ============================================================

forward_df = pd.DataFrame({

    "regime":
        state["regime"],

    "future_return":
        future_return,

    "future_volatility":
        future_volatility
})


forward_valid = (

    forward_df

    .dropna(
        subset=[
            "regime",
            "future_return",
            "future_volatility"
        ]
    )

    .copy()
)


forward_valid[
    "regime"
] = (

    forward_valid[
        "regime"
    ]

    .astype(int)
)


# ============================================================
# 6. FORWARD-DATA SANITY CHECK
# ============================================================

print("\n" + "=" * 72)

print(
    "FORWARD-VALIDATION DATA CHECK"
)

print("=" * 72)


print(
    f"\nState observations       : "
    f"{len(state):,}"
)

print(
    f"Complete forward windows : "
    f"{len(forward_valid):,}"
)

print(
    f"Forward horizon          : "
    f"{FORWARD_HORIZON} trading days"
)


if len(forward_valid) == 0:

    raise RuntimeError(
        "Forward-validation sample is empty. "
        "Check SPY/state date alignment."
    )


forward_counts = (

    forward_valid[
        "regime"
    ]

    .value_counts()

    .sort_index()
)


print(
    "\nComplete forward observations by regime:"
)


print(
    forward_counts
)


missing_forward_regimes = (

    set(
        range(K)
    )

    -

    set(
        forward_valid[
            "regime"
        ].unique()
    )
)


if missing_forward_regimes:

    print(
        "\nWARNING:"
    )

    print(
        "No complete forward observations exist for regimes:",
        sorted(
            missing_forward_regimes
        )
    )


# ============================================================
# 7. CONDITIONAL FORWARD BEHAVIOR
# ============================================================

conditional_forward = (

    forward_valid

    .groupby(
        "regime"
    )

    .agg(

        n=(
            "future_return",
            "size"
        ),

        mean_forward_return=(
            "future_return",
            "mean"
        ),

        median_forward_return=(
            "future_return",
            "median"
        ),

        std_forward_return=(
            "future_return",
            "std"
        ),

        mean_forward_volatility=(
            "future_volatility",
            "mean"
        ),

        median_forward_volatility=(
            "future_volatility",
            "median"
        )
    )

    .reindex(
        range(K)
    )
)


# ------------------------------------------------------------
# Descriptive annualization
# ------------------------------------------------------------

conditional_forward[
    "annualized_forward_return_approx"
] = (

    conditional_forward[
        "mean_forward_return"
    ]

    *
    (
        252
        /
        FORWARD_HORIZON
    )
)


print("\n" + "=" * 72)

print(
    f"CONDITIONAL FORWARD BEHAVIOR — "
    f"{FORWARD_HORIZON} DAYS"
)

print("=" * 72)


display(
    conditional_forward.round(4)
)


# ============================================================
# 8. CROSS-REGIME ECONOMIC SEPARATION
# ============================================================
#
# These are descriptive separation measures.
#
# We are NOT claiming causality or forecasting power.
# ============================================================

valid_forward_summary = (

    conditional_forward

    .dropna(
        subset=[
            "mean_forward_return",
            "mean_forward_volatility"
        ]
    )
)


if len(
    valid_forward_summary
) > 1:

    forward_return_range = float(

        valid_forward_summary[
            "mean_forward_return"
        ].max()

        -

        valid_forward_summary[
            "mean_forward_return"
        ].min()
    )


    forward_vol_range = float(

        valid_forward_summary[
            "mean_forward_volatility"
        ].max()

        -

        valid_forward_summary[
            "mean_forward_volatility"
        ].min()
    )


else:

    forward_return_range = np.nan

    forward_vol_range = np.nan


print("\n" + "=" * 72)

print(
    "CROSS-REGIME ECONOMIC SEPARATION"
)

print("=" * 72)


print(
    f"\nMean forward-return range     : "
    f"{forward_return_range:.4f}"
)


print(
    f"Mean forward-volatility range : "
    f"{forward_vol_range:.4f}"
)


# ============================================================
# 9. BASELINE REPRESENTATION FOR ROBUSTNESS
# ============================================================
#
# Reconstruct the PCA + GMM component of Cell 7.
#
# This becomes the baseline partition against which slightly
# perturbed versions of the feature matrix are compared.
# ============================================================

N_CLUSTER_PCS = min(
    4,
    X.shape[1],
    len(X)
)


base_pca = PCA(
    n_components=N_CLUSTER_PCS
)


base_Z = (

    base_pca

    .fit_transform(
        X.values
    )
)


base_gmm = GaussianMixture(

    n_components=K,

    covariance_type="full",

    reg_covar=1e-5,

    n_init=10,

    max_iter=750,

    random_state=SEED
)


base_labels = (

    base_gmm

    .fit_predict(
        base_Z
    )
)


# ============================================================
# 10. PERTURBATION STABILITY
# ============================================================
#
# A scientifically useful latent-state architecture should not
# collapse when the standardized market features are perturbed
# slightly.
#
# For each experiment:
#
#   X* = X + epsilon
#
# where epsilon is small Gaussian noise.
#
# We then refit:
#
#   PCA -> GMM
#
# and compare the resulting partition with the baseline using
# Adjusted Rand Index.
#
# ARI is permutation invariant, which is essential because
# cluster labels themselves are arbitrary.
# ============================================================

N_PERTURBATIONS = 20

PERTURBATION_SCALE = 0.05


rng = np.random.default_rng(
    SEED
)


ari_scores = []


for run in range(
    N_PERTURBATIONS
):

    noise = rng.normal(

        loc=0.0,

        scale=PERTURBATION_SCALE,

        size=X.shape
    )


    X_perturbed = (
        X.values
        +
        noise
    )


    pca_perturbed = PCA(
        n_components=N_CLUSTER_PCS
    )


    Z_perturbed = (

        pca_perturbed

        .fit_transform(
            X_perturbed
        )
    )


    gmm_perturbed = GaussianMixture(

        n_components=K,

        covariance_type="full",

        reg_covar=1e-5,

        n_init=5,

        max_iter=500,

        random_state=(
            SEED
            +
            run
            +
            1
        )
    )


    labels_perturbed = (

        gmm_perturbed

        .fit_predict(
            Z_perturbed
        )
    )


    ari = adjusted_rand_score(

        base_labels,

        labels_perturbed
    )


    ari_scores.append(
        float(ari)
    )


ari_scores = np.asarray(
    ari_scores,
    dtype=float
)


perturbation_report = pd.DataFrame({

    "run":
        np.arange(
            1,
            N_PERTURBATIONS + 1
        ),

    "ARI":
        ari_scores
})


print("\n" + "=" * 72)

print(
    "PERTURBATION STABILITY"
)

print("=" * 72)


print(
    f"\nNoise scale      : "
    f"{PERTURBATION_SCALE}"
)

print(
    f"Perturbations    : "
    f"{N_PERTURBATIONS}"
)

print(
    f"Mean ARI         : "
    f"{ari_scores.mean():.3f}"
)

print(
    f"Median ARI       : "
    f"{np.median(ari_scores):.3f}"
)

print(
    f"Minimum ARI      : "
    f"{ari_scores.min():.3f}"
)

print(
    f"Maximum ARI      : "
    f"{ari_scores.max():.3f}"
)


# ============================================================
# 11. PERTURBATION DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(9, 5)
)


plt.hist(

    ari_scores,

    bins=min(
        10,
        N_PERTURBATIONS
    )
)


plt.xlabel(
    "Adjusted Rand Index"
)

plt.ylabel(
    "Frequency"
)

plt.title(
    "Latent-State Stability Under Small Feature Perturbations"
)

plt.grid(
    alpha=0.2
)

plt.show()


# ============================================================
# 12. TRANSITION-INTENSITY VALIDATION
# ============================================================
#
# Cell 8 defined transition intensity independently of the
# regime labels.
#
# We test whether realized consensus regime changes occur
# during periods of greater observable state-space movement.
# ============================================================

switch_mask = (

    state[
        "regime_switch"
    ]

    .fillna(False)

    .astype(bool)
)


switch_intensity = (

    state.loc[
        switch_mask,
        "transition_intensity"
    ]

    .dropna()
)


quiet_intensity = (

    state.loc[
        ~switch_mask,
        "transition_intensity"
    ]

    .dropna()
)


if (
    len(switch_intensity) > 0
    and
    len(quiet_intensity) > 0
):

    switch_intensity_mean = float(
        switch_intensity.mean()
    )


    quiet_intensity_mean = float(
        quiet_intensity.mean()
    )


    transition_intensity_ratio = (

        switch_intensity_mean

        /

        (
            quiet_intensity_mean
            +
            1e-12
        )
    )


else:

    switch_intensity_mean = np.nan

    quiet_intensity_mean = np.nan

    transition_intensity_ratio = np.nan


transition_validation = pd.DataFrame({

    "sample": [
        "Regime switch",
        "No regime switch"
    ],

    "observations": [
        len(switch_intensity),
        len(quiet_intensity)
    ],

    "mean_intensity": [
        switch_intensity_mean,
        quiet_intensity_mean
    ],

    "median_intensity": [

        (
            float(
                switch_intensity.median()
            )
            if len(switch_intensity)
            else np.nan
        ),

        (
            float(
                quiet_intensity.median()
            )
            if len(quiet_intensity)
            else np.nan
        )
    ]
})


print("\n" + "=" * 72)

print(
    "TRANSITION-INTENSITY VALIDATION"
)

print("=" * 72)


display(
    transition_validation.round(4)
)


print(
    f"\nSwitch / quiet intensity ratio: "
    f"{transition_intensity_ratio:.3f}"
)


# ============================================================
# 13. MODEL-DISAGREEMENT VALIDATION
# ============================================================
#
# Transition pressure and disagreement are distinct:
#
# transition intensity:
#     movement in observable market feature space
#
# disagreement:
#     uncertainty across latent-state estimators
#
# We test whether model disagreement rises on regime-switch
# observations.
# ============================================================

switch_disagreement = (

    state.loc[
        switch_mask,
        "disagreement"
    ]

    .dropna()
)


quiet_disagreement = (

    state.loc[
        ~switch_mask,
        "disagreement"
    ]

    .dropna()
)


if (
    len(switch_disagreement) > 0
    and
    len(quiet_disagreement) > 0
):

    switch_disagreement_mean = float(
        switch_disagreement.mean()
    )


    quiet_disagreement_mean = float(
        quiet_disagreement.mean()
    )


    disagreement_ratio = (

        switch_disagreement_mean

        /

        (
            quiet_disagreement_mean
            +
            1e-12
        )
    )


else:

    switch_disagreement_mean = np.nan

    quiet_disagreement_mean = np.nan

    disagreement_ratio = np.nan


disagreement_validation = pd.DataFrame({

    "sample": [
        "Regime switch",
        "No regime switch"
    ],

    "observations": [
        len(switch_disagreement),
        len(quiet_disagreement)
    ],

    "mean_disagreement": [
        switch_disagreement_mean,
        quiet_disagreement_mean
    ],

    "median_disagreement": [

        (
            float(
                switch_disagreement.median()
            )
            if len(switch_disagreement)
            else np.nan
        ),

        (
            float(
                quiet_disagreement.median()
            )
            if len(quiet_disagreement)
            else np.nan
        )
    ]
})


print("\n" + "=" * 72)

print(
    "MODEL-DISAGREEMENT VALIDATION"
)

print("=" * 72)


display(
    disagreement_validation.round(4)
)


print(
    f"\nSwitch / quiet disagreement ratio: "
    f"{disagreement_ratio:.3f}"
)


# ============================================================
# 14. SWARM SIGNATURE
# ============================================================
#
# We now validate the scientific swarm itself.
#
# Does the collective scientific picture stabilize as the
# number of agents increases?
#
# We examine:
#
#   A. preferred latent-state count distribution
#   B. mathematical vocabulary
# ============================================================

def swarm_signature(
    proposal_subset,
    vocabulary_size=20
):

    # --------------------------------------------------------
    # Preferred state-count distribution
    # --------------------------------------------------------

    state_counts = Counter(

        int(
            p["state_count"]
        )

        for p in proposal_subset
    )


    state_distribution = np.array([

        state_counts.get(
            k,
            0
        )

        for k in range(
            2,
            9
        )

    ], dtype=float)


    if state_distribution.sum() > 0:

        state_distribution = (

            state_distribution

            /
            state_distribution.sum()
        )


    # --------------------------------------------------------
    # Mathematical vocabulary
    # --------------------------------------------------------

    vocabulary = Counter()


    for p in proposal_subset:

        for phrase in p[
            "mathematical_core"
        ]:

            words = re.findall(

                r"[A-Za-z][A-Za-z\-]+",

                str(
                    phrase
                ).lower()
            )


            for word in words:

                if len(word) > 4:

                    vocabulary[
                        word
                    ] += 1


    top_vocabulary = {

        word

        for word, _

        in vocabulary.most_common(
            vocabulary_size
        )
    }


    return {

        "state_distribution":
            state_distribution,

        "top_vocabulary":
            top_vocabulary
    }


# ============================================================
# 15. FULL-SWARM REFERENCE SIGNATURE
# ============================================================

full_signature = swarm_signature(
    proposals
)


full_state_distribution = (

    full_signature[
        "state_distribution"
    ]
)


full_vocabulary = (

    full_signature[
        "top_vocabulary"
    ]
)


# ============================================================
# 16. PREFIX CONVERGENCE
# ============================================================
#
# Evaluate the collective scientific signature after:
#
#     5
#     10
#     20
#     30
#     40
#     50
#
# agents.
#
# State-count convergence:
#
#     L1 distance from final 50-agent distribution.
#
# Vocabulary convergence:
#
#     Jaccard similarity with final top vocabulary.
# ============================================================

prefix_sizes = [
    5,
    10,
    20,
    30,
    40,
    50
]


convergence_rows = []


for n_agents in prefix_sizes:

    signature_n = swarm_signature(

        proposals[
            :n_agents
        ]
    )


    state_dist_n = (

        signature_n[
            "state_distribution"
        ]
    )


    vocabulary_n = (

        signature_n[
            "top_vocabulary"
        ]
    )


    # --------------------------------------------------------
    # State-count L1 distance
    # --------------------------------------------------------

    state_l1_distance = float(

        np.abs(

            state_dist_n

            -

            full_state_distribution

        ).sum()
    )


    # --------------------------------------------------------
    # Vocabulary Jaccard similarity
    # --------------------------------------------------------

    vocabulary_union = (

        vocabulary_n

        |

        full_vocabulary
    )


    vocabulary_intersection = (

        vocabulary_n

        &

        full_vocabulary
    )


    if len(
        vocabulary_union
    ) > 0:

        vocabulary_jaccard = float(

            len(
                vocabulary_intersection
            )

            /

            len(
                vocabulary_union
            )
        )


    else:

        vocabulary_jaccard = np.nan


    convergence_rows.append({

        "agents":
            n_agents,

        "state_count_L1_distance":
            state_l1_distance,

        "vocabulary_Jaccard":
            vocabulary_jaccard
    })


swarm_convergence = pd.DataFrame(
    convergence_rows
)


print("\n" + "=" * 72)

print(
    "SWARM CONVERGENCE"
)

print("=" * 72)


display(
    swarm_convergence.round(4)
)


# ============================================================
# 17. STATE-COUNT CONVERGENCE PLOT
# ============================================================

plt.figure(
    figsize=(9, 5)
)


plt.plot(

    swarm_convergence[
        "agents"
    ],

    swarm_convergence[
        "state_count_L1_distance"
    ],

    marker="o"
)


plt.xlabel(
    "Number of scientific agents"
)

plt.ylabel(
    "L1 distance from full-swarm distribution"
)

plt.title(
    "Swarm Convergence — Preferred Number of Latent States"
)

plt.grid(
    alpha=0.2
)

plt.show()


# ============================================================
# 18. VOCABULARY CONVERGENCE PLOT
# ============================================================

plt.figure(
    figsize=(9, 5)
)


plt.plot(

    swarm_convergence[
        "agents"
    ],

    swarm_convergence[
        "vocabulary_Jaccard"
    ],

    marker="o"
)


plt.xlabel(
    "Number of scientific agents"
)

plt.ylabel(
    "Jaccard similarity with full-swarm vocabulary"
)

plt.title(
    "Swarm Convergence — Mathematical Vocabulary"
)

plt.grid(
    alpha=0.2
)

plt.show()


# ============================================================
# 19. CORE LATENT-STATE METRICS
# ============================================================

if "gmm_silhouette" in globals():

    final_gmm_silhouette = float(
        gmm_silhouette
    )


else:

    final_gmm_silhouette = np.nan


if "consensus_persistence" in globals():

    final_consensus_persistence = float(
        consensus_persistence
    )


else:

    regime_values = (

        state[
            "regime"
        ]

        .dropna()

        .astype(int)

        .values
    )


    if len(regime_values) > 1:

        final_consensus_persistence = float(

            np.mean(

                regime_values[1:]

                ==

                regime_values[:-1]
            )
        )


    else:

        final_consensus_persistence = np.nan


final_mean_disagreement = float(

    state[
        "disagreement"
    ].mean()
)


# ============================================================
# 20. VALIDATION REPORT
# ============================================================
#
# We deliberately DO NOT collapse all metrics into one
# arbitrary "validation score."
#
# Each statistic answers a different scientific question.
# ============================================================

validation_report = {

    "architecture":
        synthesis[
            "architecture_name"
        ],

    "target_regimes":
        int(K),

    "number_of_observations":
        int(
            len(state)
        ),

    "number_of_agents":
        int(
            len(proposals)
        ),

    "forward_horizon_days":
        int(
            FORWARD_HORIZON
        ),

    "complete_forward_windows":
        int(
            len(forward_valid)
        ),

    "gmm_silhouette":
        (
            float(
                final_gmm_silhouette
            )

            if np.isfinite(
                final_gmm_silhouette
            )

            else None
        ),

    "consensus_persistence":
        (
            float(
                final_consensus_persistence
            )

            if np.isfinite(
                final_consensus_persistence
            )

            else None
        ),

    "perturbation_ARI_mean":
        float(
            ari_scores.mean()
        ),

    "perturbation_ARI_median":
        float(
            np.median(
                ari_scores
            )
        ),

    "perturbation_ARI_min":
        float(
            ari_scores.min()
        ),

    "perturbation_ARI_max":
        float(
            ari_scores.max()
        ),

    "transition_intensity_switch_ratio":
        (
            float(
                transition_intensity_ratio
            )

            if np.isfinite(
                transition_intensity_ratio
            )

            else None
        ),

    "mean_model_disagreement":
        float(
            final_mean_disagreement
        ),

    "switch_disagreement_ratio":
        (
            float(
                disagreement_ratio
            )

            if np.isfinite(
                disagreement_ratio
            )

            else None
        ),

    "cross_regime_forward_return_range":
        (
            float(
                forward_return_range
            )

            if np.isfinite(
                forward_return_range
            )

            else None
        ),

    "cross_regime_forward_volatility_range":
        (
            float(
                forward_vol_range
            )

            if np.isfinite(
                forward_vol_range
            )

            else None
        ),

    "status":
        "EDUCATIONAL RESEARCH — NOT INVESTMENT ADVICE"
}


# ============================================================
# 21. VALIDATION DASHBOARD
# ============================================================

dashboard_rows = [

    {
        "Metric":
            "Number of regimes",

        "Value":
            K
    },

    {
        "Metric":
            "Complete forward windows",

        "Value":
            len(forward_valid)
    },

    {
        "Metric":
            "GMM silhouette",

        "Value":
            final_gmm_silhouette
    },

    {
        "Metric":
            "Consensus persistence",

        "Value":
            final_consensus_persistence
    },

    {
        "Metric":
            "Perturbation ARI — mean",

        "Value":
            ari_scores.mean()
    },

    {
        "Metric":
            "Perturbation ARI — minimum",

        "Value":
            ari_scores.min()
    },

    {
        "Metric":
            "Transition-intensity switch ratio",

        "Value":
            transition_intensity_ratio
    },

    {
        "Metric":
            "Mean model disagreement",

        "Value":
            final_mean_disagreement
    },

    {
        "Metric":
            "Switch disagreement ratio",

        "Value":
            disagreement_ratio
    },

    {
        "Metric":
            "Cross-regime forward-return range",

        "Value":
            forward_return_range
    },

    {
        "Metric":
            "Cross-regime forward-volatility range",

        "Value":
            forward_vol_range
    },

    {
        "Metric":
            "Scientific agents",

        "Value":
            len(proposals)
    }
]


validation_dashboard = pd.DataFrame(
    dashboard_rows
)


print("\n" + "=" * 72)

print(
    "VALIDATION DASHBOARD"
)

print("=" * 72)


display(
    validation_dashboard.round(4)
)


# ============================================================
# 22. VALIDATED MARKET-REGIME OBSERVATORY
# ============================================================

plt.figure(
    figsize=(15, 6)
)


for k in range(K):

    mask = (
        state[
            "regime"
        ]

        ==

        k
    )


    plt.scatter(

        state.index[
            mask
        ],

        spy.reindex(
            state.index[
                mask
            ]
        ),

        s=8,

        label=f"Regime {k}"
    )


plt.yscale(
    "log"
)


plt.xlabel(
    "Date"
)

plt.ylabel(
    "SPY — log scale"
)

plt.title(
    "Validated Consensus Market-Regime Observatory"
)

plt.legend(
    ncol=min(
        K,
        4
    )
)

plt.grid(
    alpha=0.2
)

plt.show()


# ============================================================
# 23. TRANSITION-PRESSURE OBSERVATORY
# ============================================================

plt.figure(
    figsize=(15, 5)
)


plt.plot(

    state.index,

    state[
        "transition_z"
    ],

    label="Transition pressure"
)


plt.axhline(

    2.0,

    linestyle="--",

    linewidth=1.0,

    label="High-pressure threshold"
)


plt.axhline(

    0.0,

    linewidth=0.8
)


plt.xlabel(
    "Date"
)

plt.ylabel(
    "Transition-pressure z-score"
)

plt.title(
    "Observable Market State-Space Transition Pressure"
)

plt.legend()

plt.grid(
    alpha=0.2
)

plt.show()


# ============================================================
# 24. MODEL-DISAGREEMENT OBSERVATORY
# ============================================================

plt.figure(
    figsize=(15, 4)
)


plt.plot(

    state.index,

    state[
        "disagreement"
    ]
)


plt.xlabel(
    "Date"
)

plt.ylabel(
    "Model disagreement"
)

plt.title(
    "Epistemic Uncertainty Across Latent-State Models"
)

plt.grid(
    alpha=0.2
)

plt.show()


# ============================================================
# 25. SAVE VALIDATION OUTPUTS
# ============================================================

conditional_forward.to_csv(

    OUT /
    "conditional_forward_behavior.csv"
)


forward_valid.to_csv(

    OUT /
    "forward_validation_sample.csv"
)


perturbation_report.to_csv(

    OUT /
    "perturbation_stability.csv",

    index=False
)


transition_validation.to_csv(

    OUT /
    "transition_validation_cell9.csv",

    index=False
)


disagreement_validation.to_csv(

    OUT /
    "disagreement_validation_cell9.csv",

    index=False
)


swarm_convergence.to_csv(

    OUT /
    "swarm_convergence.csv",

    index=False
)


validation_dashboard.to_csv(

    OUT /
    "validation_dashboard.csv",

    index=False
)


validation_path = (

    OUT /
    "validation_report.json"
)


validation_path.write_text(

    json.dumps(
        validation_report,
        indent=2
    ),

    encoding="utf-8"
)


# ============================================================
# 26. FINAL CELL 9 REPORT
# ============================================================

print("\n" + "=" * 72)

print(
    "CELL 9 COMPLETE — VALIDATION LAYER FINISHED"
)

print("=" * 72)


print(
    f"\nLatent regimes                  : "
    f"{K}"
)

print(
    f"Complete forward windows        : "
    f"{len(forward_valid):,}"
)


if np.isfinite(
    final_gmm_silhouette
):

    print(
        f"GMM silhouette                  : "
        f"{final_gmm_silhouette:.3f}"
    )


if np.isfinite(
    final_consensus_persistence
):

    print(
        f"Consensus persistence           : "
        f"{final_consensus_persistence:.3f}"
    )


print(
    f"Perturbation ARI — mean         : "
    f"{ari_scores.mean():.3f}"
)

print(
    f"Perturbation ARI — minimum      : "
    f"{ari_scores.min():.3f}"
)


if np.isfinite(
    transition_intensity_ratio
):

    print(
        f"Transition-intensity ratio      : "
        f"{transition_intensity_ratio:.3f}"
    )


print(
    f"Mean model disagreement         : "
    f"{final_mean_disagreement:.3f}"
)


if np.isfinite(
    disagreement_ratio
):

    print(
        f"Switch disagreement ratio       : "
        f"{disagreement_ratio:.3f}"
    )


if np.isfinite(
    forward_return_range
):

    print(
        f"Forward-return regime range     : "
        f"{forward_return_range:.4f}"
    )


if np.isfinite(
    forward_vol_range
):

    print(
        f"Forward-volatility regime range : "
        f"{forward_vol_range:.4f}"
    )


print(
    f"Forward horizon                 : "
    f"{FORWARD_HORIZON} trading days"
)

print(
    f"Scientific agents               : "
    f"{len(proposals)}"
)


print("\n" + "-" * 72)

print(
    "SCIENTIFIC INTERPRETATION"
)

print("-" * 72)


print(
    "\nThe latent-state architecture has now been examined "
    "along several independent dimensions."
)


print(
    "\nFirst, conditional forward behavior tests whether states "
    "identified at time t correspond to distinguishable market "
    "behavior during the following 21 trading days."
)


print(
    "\nSecond, perturbation analysis tests whether small changes "
    "in the observed feature matrix materially alter the "
    "latent-state partition."
)


print(
    "\nThird, transition-intensity analysis tests whether regime "
    "changes coincide with stronger movement through observable "
    "market state space."
)


print(
    "\nFourth, model disagreement measures epistemic uncertainty "
    "across alternative latent-state estimators."
)


print(
    "\nFinally, swarm-convergence analysis asks whether the "
    "scientific conclusions stabilize as the number of "
    "independent disciplinary agents increases."
)


print(
    "\nNo single metric is treated as proof that the inferred "
    "regimes constitute a unique or true market ontology."
)


print(
    "The architecture is retained only to the extent that "
    "multiple independent tests support its robustness, "
    "interpretability and empirical usefulness."
)


print(
    "\nCell 10 can now consolidate the complete experiment, "
    "provenance, outputs and final research conclusions."
)


print("\n" + "=" * 72)

print(
    "READY FOR CELL 10"
)

print("=" * 72)
```

CODE UNIT 10 closes the research loop by exporting a board-level scientific report, a machine-readable research package and the full provenance trail. This is the operational form of the Market Regime Observatory: latest consensus state, model disagreement, transition pressure and the most distinctive standardized features. That separation avoids contaminating regime discovery with strategy optimization.

This makes the experiment portable and auditable.

The conclusion embedded below emphasizes the main intellectual result: the swarm is a mechanism for systematic scientific transfer. Its purpose is not to replace HMMs with a fashionable metaphor. ## The Narrative: From Fifty Scientific Minds to a Market-Regime Observatory

## A 50-Agent Cross-Science Experiment in Market-Regime Discovery

Financial professionals have always understood that markets behave differently under different conditions. A portfolio that performs well during a stable expansion may behave very differently during a liquidity shock. Correlations that appear dependable in ordinary markets can suddenly converge toward one during periods of stress. Credit, equities, bonds, commodities and volatility can move from relatively independent sources of information to components of a single systemic event.

We usually describe these different configurations with the word **regime**.

The difficulty is that the regime itself is not observable. We observe prices, returns, volatility, correlations, spreads, trends and rotations. We do not observe a variable called *the market regime*. Regime discovery is therefore an inverse problem: from a collection of observable financial signals, we attempt to infer the hidden state of the system that may have generated them.

Traditional finance already possesses sophisticated methods for attacking this problem. Hidden Markov Models, Markov-switching models, mixture models and change-point methods are familiar examples. The notebook developed here begins from a different question:

**What if finance has been searching too narrowly?**

Physics, chemistry, electrical engineering, control theory, information theory, mathematics, network science, astrophysics and geophysics all confront versions of the same fundamental problem. They observe complex systems whose underlying states cannot be measured directly. They must infer those states from noisy, incomplete and sometimes contradictory signals.

The notebook therefore does not begin by selecting a favorite financial model. It begins by constructing a **swarm of fifty scientific agents** and asking them to search a much larger intellectual universe.

The result is a progression from scientific diversity to empirical discipline:

> **50 scientific conjectures → cross-disciplinary synthesis → executable architecture → latent market states → transition dynamics → robustness and falsification.**

That sequence is the key to understanding the experiment.

---

## 1. Why a Swarm Rather Than One Powerful Model?

Imagine asking a single highly capable AI system:

> “Find the best way to identify market regimes.”

The system might produce an intelligent answer. But there is a substantial risk that it would gravitate toward the most familiar concepts in its training distribution: Hidden Markov Models, clustering, volatility states, perhaps neural networks. The answer could be sophisticated while remaining intellectually narrow.

The swarm changes the problem.

Instead of asking one model to be universally intelligent, the notebook deliberately gives fifty agents **different scientific identities and different mathematical starting points**. They are organized into ten disciplinary families, with five specialists inside each family.

The families cover statistical physics, dynamical systems, signal engineering, control theory, information theory, applied mathematics, chemistry and statistical mechanics, astrophysics, network science, and geophysics/seismology.

The distinction matters because each discipline sees the hidden-state problem differently.

A statistical physicist might ask whether financial markets occupy **metastable phases** separated by transitions. Another may search for an **order parameter** that summarizes collective market organization.

A dynamical-systems specialist might think in terms of **attractors, basins of attraction, bifurcations, Lyapunov stability or slow-fast systems**.

A signal engineer may regard the observed market as a noisy mixture of latent signals and ask whether spectral decomposition, wavelets, blind-source separation or change detection can recover the hidden structure.

A control theorist naturally thinks about **state-space representation, observability, Kalman filtering, switching systems and system identification**.

An information theorist asks a different question: when does the information structure of the market change? Entropy, mutual information, information bottlenecks, minimum-description-length ideas, transfer entropy and distributional divergence become candidate tools.

Applied mathematics contributes operator methods, topology, optimal transport, manifold learning and inverse problems.

Chemistry brings a particularly powerful analogy: a complex system can remain for long periods inside a **metastable basin**, occasionally acquire enough pressure to cross an energy barrier, pass through a transition state, and settle into another basin. That is strikingly close to the intuitive language financial professionals already use when describing stable markets interrupted by abrupt transitions.

Astrophysics contributes methods designed to recover weak latent sources from extremely noisy observations. Network science asks whether the regime is encoded not in individual asset returns but in the **topology of relationships among assets**. Geophysics and seismology contribute ideas about stress accumulation, precursor signals, critical transitions, rupture and event clustering.

The fifty agents therefore do not represent fifty repetitions of the same analysis. They represent **fifty different lenses through which the same hidden-state problem can be viewed**.

That is the first important role of the swarm:

> **Expansion of the hypothesis space.**

---

## 2. Giving Every Scientist the Same Empirical Reality

Scientific imagination is useful only if it remains connected to the data.

The notebook therefore constructs a common empirical environment before asking the agents to reason. The market universe includes major equity exposures such as SPY, QQQ and IWM; Treasury instruments such as TLT and IEF; investment-grade and high-yield credit through LQD and HYG; gold through GLD; commodities through DBC; the dollar through UUP; and the VIX as a direct market-volatility indicator.

From this multivariate system, the notebook constructs thirteen observable market features.

These include short-horizon equity return, medium-horizon trend, realized volatility, downside volatility, drawdown, cross-asset dispersion, average correlation, market breadth, the VIX level and its change, credit risk, bond-versus-equity rotation, and gold-versus-equity rotation.

This feature system is important pedagogically because a market regime should not be confused with the direction of the S&P 500 alone.

A genuine market state may involve simultaneously rising volatility, deteriorating credit, increasing cross-asset correlation, weakening breadth, defensive bond rotation and demand for gold. The regime is therefore conceived as a **multidimensional configuration of the financial system**.

The features are robustly standardized. This prevents a variable with a naturally large numerical scale from dominating the representation merely because of its units.

The fifty agents consequently face the same empirical reality. They can propose radically different theories, but they cannot invent different markets.

This creates a productive separation:

> **The swarm is free in theory, but constrained by evidence.**

---

## 3. What Exactly Does Each Agent Produce?

Each agent receives a structured scientific assignment. It must explain what it believes the latent object actually is, which scientific principle it proposes transferring to finance, the mathematical core of that transfer, which observable features would be required, how transitions should be detected, how many states might be appropriate, how the proposal should be validated, how it could fail, and how novel, plausible and confident the agent considers its hypothesis.

The output is not free-form prose. Claude Sonnet 5 is required to return a defined structured object.

This is a crucial engineering choice.

Without a common schema, fifty scientists would produce fifty essays that would be difficult to compare systematically. Structured output transforms the swarm from a collection of interesting conversations into a **research dataset containing fifty hypotheses**.

The notebook also uses a canary call before launching the complete swarm. One agent is executed first. Only after the API response has passed the structured-output and parsing checks are the other forty-nine agents launched.

Each successful proposal is checkpointed. A failed connection or interrupted Colab session therefore does not destroy the intellectual work already completed.

The individual agents are deliberately run with relatively low reasoning effort. This reflects a deeper design principle: the intelligence of this experiment does not come primarily from making one agent think for an extremely long time. It comes from **diversity across the population**.

We are purchasing breadth before depth.

---

## 4. Fifty Ideas Do Not Solve the Problem

At this point the notebook has deliberately created a new difficulty.

We began with uncertainty about market regimes. We now possess fifty competing scientific descriptions of them.

A naïve solution would be majority voting: count how many agents favor each idea and select the winner.

The notebook explicitly rejects that approach.

Scientific innovation is not necessarily democratic. An unusual proposal from geophysics or topology could contain precisely the idea that forty conventional proposals miss. If the swarm were reduced immediately to majority rule, diversity would have been generated only to be destroyed.

Instead, Cell 6 introduces a separate **scientific synthesizer**.

Its task is not to become agent fifty-one. Its task is architectural.

It receives the proposal corpus and searches for structures that recur across disciplines, complementary ideas that can coexist, and minority insights worth preserving.

The synthesis that emerged from the notebook is called the **Metastable Basin-Transition Architecture, or MBTA**.

This is a revealing result because the architecture is not simply the favorite model of one discipline. It is a convergence of ideas that appear under different names across multiple sciences.

The synthesis identifies six major recurring motifs.

The first is the idea of a **free-energy or potential landscape containing metastable basins separated by barriers**. Versions of this appear in statistical physics, chemistry and geophysics.

The second is **critical slowing down**: before some complex systems undergo a major transition, their ability to recover from disturbances weakens, potentially producing changes in variance and autocorrelation.

The third is **spectral or operator decomposition**, appearing through concepts such as Koopman operators, graph Laplacians and correlation eigenstructures.

The fourth involves **self-exciting cascades and triggered events**, connecting nucleation, aftershocks and propagation processes.

The fifth is **information-theoretic divergence** between local distributions as a model-free signal that the system itself is changing.

The sixth is **low-dimensional order-parameter reduction**: the possibility that many observable variables are manifestations of a smaller number of hidden collective coordinates.

Notice what has happened. The swarm did not simply tell us, “Use clustering.” It gave us a scientific interpretation of what clustering is attempting to recover.

---

## 5. How Was the “Best” Architecture Chosen?

It is important to be precise here. The notebook does **not** claim to have discovered the universally best theory of market regimes.

Instead, it selects the most promising architecture for this experiment using three filters.

### Cross-disciplinary recurrence

Ideas that independently emerge in unrelated sciences deserve attention because they may capture structural properties of hidden-state problems rather than conventions specific to finance.

### Complementarity

Representation, state identification, transition detection and uncertainty are different problems. A useful architecture can therefore combine ideas rather than forcing one algorithm to perform every task.

### Empirical implementability and falsifiability

An elegant scientific metaphor that cannot be connected to observable data is not enough. The synthesis is constrained to an executable empirical vocabulary available in the notebook.

This produces four layers.

The **representation layer** uses PCA to compress the thirteen-dimensional standardized market system into a smaller number of dominant coordinates. A physicist can interpret these as approximate order parameters; a signal engineer as dominant modes; a financial professional can simply regard them as the principal common dimensions along which the market system moves.

The **state-inference layer** deliberately uses several competing algorithms. Gaussian mixtures identify probabilistic basins. K-means provides a simple geometric partition. Spectral clustering contributes a neighborhood-based nonlinear perspective.

The **transition layer** is separated from state classification. The notebook constructs transition intensity from the speed at which the observable feature vector moves through standardized state space.

Finally, the **uncertainty layer** uses disagreement among models as information. If GMM, K-means and spectral clustering disagree, the notebook does not hide the disagreement by pretending that one answer is certain. It records the disagreement as epistemic uncertainty.

This is the architecture selected from the scientific universe opened by the swarm.

Importantly, the synthesis also preserves minority ideas that were not implemented in the first prototype: transfer-entropy networks, persistent homology, nucleation dynamics, robust-control fragility measures and switched dynamical systems with explicit guard boundaries.

They are not discarded. They become part of the research frontier.

---

## 6. Turning the Scientific Architecture into Market Regimes

The empirical implementation begins with PCA.

The reduced representation is passed through Gaussian Mixture Models, K-means and spectral clustering. Because unsupervised labels are arbitrary—a model's “state 0” has no reason to correspond to another model's “state 0”—the notebook uses the Hungarian assignment algorithm to align the classifications.

A consensus state can then be constructed.

The experiment settles on **four latent regimes**, but this number is treated as a working scientific hypothesis rather than a metaphysical fact about markets.

The notebook contains **3,641 market observations** and thirteen features. Rather than naming the four states prematurely as “bull,” “bear,” “crisis” and “recovery,” it examines their empirical feature profiles. This is good scientific discipline. A cluster should earn its economic interpretation from the data.

The resulting sequence is also persistent. The measured one-period consensus persistence is:

> **0.873**

That matters enormously.

If the algorithm changed its mind every day, it might be identifying clusters without identifying regimes. A regime, almost by definition, should have some temporal persistence.

A persistence of 0.873 therefore indicates substantial continuity in the inferred state sequence.

---

## 7. State and Transition Are Not the Same Thing

One of the strongest conceptual features of the notebook is the distinction between **where the market is** and **how rapidly it is moving**.

Cell 7 answers the first question through latent-state inference.

Cell 8 answers the second through transition intensity.

Transition intensity measures the velocity of the standardized market-feature vector. It therefore captures movement through observable state space without requiring the regime label to change immediately.

This distinction is intuitive for a financial professional.

A market can remain classified as the same regime while its internal conditions deteriorate rapidly. Conversely, a classification boundary can occasionally be crossed without an economically dramatic event.

The experiment records **461 consensus regime switches**. Overall persistence remains **0.873**. Transition intensity on switch observations is **1.054 times** that on non-switch observations.

That ratio is above one, but only modestly so. This should not be exaggerated. It provides some support for the idea that switches occur during stronger movement in observable state space, but it is not overwhelming evidence.

The disagreement result is more striking.

Average model disagreement is **0.144**, while disagreement on switch observations relative to ordinary observations rises by a factor of:

> **2.141**

This is economically interesting. The models become much less unanimous around the points where the consensus regime changes.

In other words:

> **Uncertainty itself appears to contain information about transitions.**

## 8. Does the Architecture Actually Perform?

Because the true regime is hidden, there is no externally supplied answer key. We cannot calculate ordinary classification accuracy.

The notebook therefore evaluates performance through several independent falsification-oriented tests.

### Geometric separation

The GMM silhouette score is:

> **0.176**

This is modest rather than spectacular. That is informative.

Financial regimes overlap. The market does not divide itself into four perfectly separated clouds for our convenience. A low-to-moderate silhouette warns us against pretending that the states are naturally isolated categories.

### Temporal persistence

The consensus persistence is:

> **0.873**

This is considerably stronger. The inferred states demonstrate meaningful temporal continuity.

### Perturbation stability

The notebook adds small random disturbances to the feature system and repeatedly rebuilds the PCA-GMM partition.

Across twenty perturbation experiments:

- Mean Adjusted Rand Index: **0.870**
- Median Adjusted Rand Index: **0.949**
- Maximum Adjusted Rand Index: **0.965**
- Minimum Adjusted Rand Index: **0.402**

This is encouraging.

It suggests that the architecture is generally not destroyed by small measurement changes.

But the minimum ARI of 0.402 is equally important. Robustness is strong on average but not universal. Some perturbations materially change the inferred partition.

A scientifically responsible interpretation must preserve both observations.

### Economic differentiation

The notebook then asks whether the regimes possess economic differentiation after they are identified.

Using **3,620 observations with complete forward windows**, it calculates subsequent 21-trading-day S&P 500 return and realized volatility conditional on the state observed today.

Across regimes, the range of mean 21-day forward log returns is:

> **0.0297**

while the range in mean forward realized volatility is:

> **0.1487**

Those numbers do not prove predictability.

The analysis uses overlapping windows and is descriptive rather than causal. But they do show that the inferred states are associated with materially different subsequent market environments.

That is precisely what we would hope to see if the latent states contain economically relevant information.

> **Scientific convergence among agents**

## 10. What Has the Experiment Actually Demonstrated?

The strongest conclusion is not that we have finally discovered the four true regimes of financial markets.

That would go well beyond the evidence.

The deeper achievement is methodological.

We started with an enormous intellectual search space. Fifty agents explored hidden-state inference through ten scientific traditions and fifty specialized perspectives.

Structured outputs converted that exploration into a comparable hypothesis corpus.

A separate synthesizer searched for recurring structures without simply counting votes.

Minority ideas were preserved rather than erased.

The resulting architecture was then constrained to algorithms that could actually be executed.

Only after that intellectual expansion did empirical compression begin.

PCA created a reduced market representation.

GMM, K-means and spectral clustering supplied competing views of the hidden states.

Consensus produced a practical state estimate.

Disagreement measured epistemic uncertainty.

Transition intensity measured movement independently of classification.

Persistence, forward behavior, perturbation tests and swarm convergence then challenged the result.

The final architecture—MBTA—is therefore not merely another clustering exercise.

It embodies a different philosophy of AI-assisted financial research:

> **Use artificial intelligence first to enlarge the universe of scientific possibilities, then use mathematical structure and empirical evidence to reduce it.**

That ordering is essential.

An LLM should not be allowed to manufacture empirical truth. Its comparative advantage here is scientific imagination, translation across disciplines and hypothesis generation. Python and market data then impose discipline.

For financial professionals, this is perhaps the most important lesson of the notebook.

The future of AI in quantitative finance may not consist simply of replacing one econometric model with a larger AI model.

A more powerful architecture may consist of **societies of specialized intelligences that explore competing explanations, followed by explicit mechanisms for synthesis, implementation, uncertainty measurement and falsification**.

The market-regime problem is an ideal demonstration because no one observes the hidden state directly. We are forced to distinguish hypothesis from evidence.

And that is precisely what the swarm makes possible.

It does not eliminate uncertainty.

**It organizes uncertainty.**

It does not prove that four regimes are the final description of markets.

It gives us a disciplined way to discover, compare, challenge and eventually replace regime descriptions.

The natural next step is therefore not to declare the problem solved. It is to make the observatory more sophisticated: out-of-sample estimation; explicit comparison with HMM and Markov-switching benchmarks; hierarchical regimes operating at different time scales; factorial states in which several hidden conditions coexist; richer credit, options and liquidity information; precursor analysis; and eventually recursive systems capable of recognizing when their own regime vocabulary has become obsolete.

At that point, the architecture moves beyond a static regime classifier.

It becomes an:

> **Adaptive scientific observatory of the market itself.**




## CONCLUSION — FROM REGIME LABELS TO A MARKET-STATE OBSERVATORY

The central result of this notebook is methodological. Market-regime discovery is reframed from a narrow model-selection exercise into a cross-science latent-state inference problem. The fifty-agent swarm is valuable not because fifty opinions are automatically better than one, but because the agents are assigned genuinely different mathematical traditions. Their collective task is to enlarge the hypothesis space before empirical discipline compresses it again.

That distinction changes the research workflow. A conventional project might begin with an HMM, choose the number of states, fit the model and then interpret the resulting clusters. Here the sequence is reversed. We begin by asking what a “hidden regime” could mean under multiple scientific ontologies. It might be a metastable phase, a state-space variable, a spectral configuration, a network topology, an information regime, a basin of attraction, a chemical-like transition state, a weak latent source, or a precursor-and-rupture process. Each interpretation suggests different observables and different failure tests. Only after collecting these possibilities do we build a common empirical representation.

The resulting architecture is intentionally hybrid. Persistence matters because market conditions do not normally redraw themselves independently every day. Multiscale structure matters because slow macroeconomic conditions and fast liquidity disturbances coexist. Network structure matters because crises often appear as changes in collective dependence rather than merely changes in average returns. Transition detection matters because the economically important moment is often not the stable regime itself but the movement between regimes. Uncertainty matters because a market can sit near a boundary where competing models disagree. A useful observatory therefore should report state, transition pressure and model disagreement separately.

The notebook also establishes a practical principle for AI-assisted quantitative research: language models should expand scientific imagination without being allowed to fabricate the empirical world. Claude Sonnet 5 generates hypotheses under a strict JSON schema, and the code checks completion status before parsing structured output. The notebook deliberately avoids obsolete sampling parameters and old JSON-extraction tricks. A one-agent canary call is executed before the remaining swarm, preventing an API mismatch from being multiplied fifty times. Low effort and concise response contracts keep the individual agents inexpensive; collective diversity, rather than maximal reasoning per call, is the source of intelligence. The market data, transformations, clustering, validation and exports remain deterministic Python objects.

No unsupervised regime model can be validated in the same way as a supervised classifier with known labels. That is why the notebook uses a portfolio of falsification-oriented diagnostics. Persistence asks whether states last long enough to deserve the word regime. Conditional forward distributions ask whether the states have economic consequences. Perturbation stability asks whether small changes in inputs destroy the partition. Transition-intensity analysis asks whether inferred switches coincide with large changes in the observable market system. Cross-agent convergence asks whether the scientific architecture is dependent on one idiosyncratic prompt. A serious extension should add walk-forward estimation, transaction-cost-aware strategy tests, alternative universes, macroeconomic releases, options surfaces, intraday liquidity measures and explicit out-of-sample crisis episodes.

The most important conceptual move is to stop treating regime as a perfectly observed categorical truth waiting to be uncovered. In real markets, regimes may overlap, nest and evolve at different speeds. A slow inflationary state can coexist with a fast liquidity event; a risk-on equity trend can coexist with deteriorating credit structure. Future versions should therefore allow hierarchical and factorial latent states, where several hidden processes jointly generate the observations. The current consensus representation is a bridge toward that richer formulation.

There is also a natural path toward recursive adaptation. Once the observatory can estimate state and transition uncertainty, downstream portfolio agents can condition their models, risk budgets and execution logic on those estimates. More importantly, the observatory itself can monitor when its regime vocabulary becomes inadequate. Persistent disagreement, falling cluster separation, unstable transition probabilities or new feature relationships can become triggers for scientific re-inquiry. At that point a new swarm could be convened to propose revised state representations. The architecture would then move from regime detection to regime-model adaptation.

The exercise therefore points toward a different vision of quantitative finance. Rather than selecting one favored model and defending it indefinitely, we can build institutions that maintain a population of scientific lenses, expose their disagreements, test their implications and revise their ontology when the market changes. Artificial intelligence becomes most interesting not when it imitates a single analyst, but when it makes structured intellectual pluralism operational.

This notebook should be read as a research platform, not as trading advice and not as evidence that any discovered state will remain stable in live markets. Its strongest claim is that the hidden-regime problem deserves a broader mathematical search process than finance normally gives it. Physics, engineering, chemistry, astrophysics, mathematics, information theory and network science all contain mature ideas about invisible states and transitions. A fifty-agent swarm gives us a practical way to interrogate those traditions simultaneously. The resulting Market Regime Observatory is the first synthesis: a transparent, falsifiable and extensible architecture for turning that collective scientific imagination into measurable hypotheses.
